# API (Application Programming Interface)

* Think of an API as a waiter in a restaurant. You look at the menu, tell the waiter what you want, and the waiter goes to the kitchen, delivers your order, and brings back the dish. You never enter the kitchen or need to know the recipe. Technically, an API is a defined interface that allows one software system to communicate with another without exposing its internal implementation.

* Web API

    * Like ordering delivery from another restaurant. Your waiter calls the other restaurant, places the order on your behalf, and brings the food to your table.

    * Your application communicates with another application's server over the internet to request or send data. This is the most common meaning of API in Data Engineering. Examples include the Spotify API, GitHub API, and Weather APIs.

* Library API

    * Like asking the bartender inside your own restaurant to make you a cocktail. You don't need to know the recipe — you just ask for it by name and it arrives at your table.

    * Your code calls functions provided by a library within the same application without needing to know how those functions are implemented. Examples include the Pandas API and NumPy API.

* Database API

    * Like asking the waiter to check the wine cellar. "Do you have a 2018 Merlot? Bring me two bottles and update the inventory." The waiter goes down, finds what you need, and brings it back.

    * Your application communicates with a database through a standard interface such as JDBC or ODBC to execute SQL queries and retrieve or modify data.

* Operating System API

    * Like asking the restaurant staff for building services — adjust the air conditioning, turn on the lights at your table, or unlock the restroom. These aren't food requests; they're requests to the building itself.

    * Your program requests services from the operating system, such as opening files, creating processes, accessing hardware, or allocating memory.

* Cloud API

    * Like hiring a catering company for your event. You don't build a new kitchen — you tell the catering service how many guests, what menu, and what equipment you need, and they provision everything for you.

    * Your application communicates with cloud providers such as AWS, Azure, or GCP to create and manage resources like virtual machines, storage, databases, and networking.

* AI API

    * Like asking the head chef to improvise a custom dish that isn't on the menu. You describe what you're in the mood for, and the chef uses their expertise and creativity to craft something and send it to your table.

    * Your application sends prompts to an AI service, which processes the request and returns generated text, images, code, or other AI outputs.

* Graphics API

    * Like giving plating instructions to the pastry chef. "Layer the chocolate here, fan the strawberries there, drizzle the sauce in this pattern." You describe the visual result, and the specialist renders it precisely.

    * Your application communicates with the graphics driver and GPU to render 2D and 3D graphics. Examples include OpenGL, Vulkan, and DirectX.

* Internal API

    * Like the kitchen's internal communication system — the expeditor calling orders to the grill station, the grill station signaling the plating station. Customers never see or interact with this; it's how the kitchen coordinates behind closed doors.

    * Internal services communicate with each other through APIs that are only available inside the organization and are not exposed to external users.

* Related API Terms

    * API

        * A contract that defines how software components exchange data by specifying inputs, outputs, protocols, and data formats, enabling communication without exposing internal implementation.

    * Web API

        * An API that operates over HTTP/HTTPS, accepting requests at URLs and returning responses (typically JSON or XML), allowing applications to communicate over the internet.

        * REST API

            * A web API design style where resources are identified by URLs and accessed using standard HTTP methods (GET, POST, PUT, DELETE). It provides a uniform, stateless, cacheable, and scalable interface for CRUD operations, typically exchanging data in formats such as JSON or XML.
            
            > Similar to browsing the web, a client sends an HTTP request for a resource, and the server responds with the requested data. APIs also commonly provide metadata, documentation, authentication, and error handling to simplify integration.

        * GraphQL API

            * A query language and runtime where the client specifies exactly which fields and nested relations to return, eliminating over-fetching and under-fetching.

        * gRPC API

            * A Remote Procedure Call (RPC) framework using HTTP/2 and Protocol Buffers with shared `.proto` schemas, providing high-throughput, low-latency communication with strongly typed contracts and bi-directional streaming.

    * SDK (Software Development Kit)

        * A package containing libraries, documentation, code samples, and tools that wrap raw API calls into language-native functions, simplifying authentication, requests, and error handling.

    * Endpoint

        * A specific URL combined with an HTTP method that maps to a server operation and routes incoming requests to the appropriate handler.

    * Request

        * An HTTP message sent from a client containing a method, URL, headers, and optionally a body, specifying the operation and any required data.

    * Response

        * An HTTP message returned by the server containing a status code, headers, and optionally a body, providing the requested data or error information.

    * Client

        * Any software that initiates API calls, such as a browser, mobile app, CLI tool, or another server, by sending requests and consuming responses.

    * Server

        * The software that listens for API requests, executes business logic, and returns responses to clients.

* FastAPI

    * A modern Python Web Framework built on Starlette (ASGI) and Pydantic (data validation). It uses Python type hints to automatically validate requests, serialize responses, and generate OpenAPI documentation.

    * Framework Hierarchy

        * Software Framework

            * Web Frameworks – Handle HTTP requests, routing, middleware, validation, and responses.

                * FastAPI – Python, ASGI, async-first, automatic OpenAPI documentation.

                * Flask – Python, WSGI, lightweight, extensible through plugins.

                * Django – Python, WSGI, full-stack framework with ORM, authentication, admin panel, and migrations.

                * Express.js – JavaScript, middleware-based framework for Node.js.

                * Spring Boot – Java framework using dependency injection and convention over configuration.

                * ASP.NET Core – Cross-platform C# web framework with built-in dependency injection.

            * GUI Frameworks – Build graphical user interfaces and handle user interactions.

            * ML Frameworks – Define, train, and deploy machine learning models.

            * Testing Frameworks – Write, execute, and report automated tests.

    * Other Python Web Frameworks

        * FastAPI – ASGI framework using async/await, Pydantic validation, and automatic OpenAPI generation. Primarily used for high-performance asynchronous APIs.

        * Flask – Lightweight WSGI micro-framework commonly used for small web applications and APIs.

        * Django – Full-stack WSGI framework with built-in ORM, authentication, admin panel, templates, and migrations.

        * Tornado – Asynchronous networking framework designed for long-lived connections such as WebSockets.

        * Falcon – Lightweight WSGI/ASGI framework optimized for high-performance REST APIs.

# HTTP & HTTPS

* HTTP (Hypertext Transfer Protocol)

    * The standard communication protocol used to transfer data between a client and a server over the Internet.

    * Communication over HTTP involves two parties:

        * Client → Sends an HTTP request.

        * Server → Processes the request and returns an HTTP response.

* HTTPS (Hypertext Transfer Protocol Secure)

    * The secure version of HTTP that encrypts data during transmission.

    > The client encrypts the request before sending it, the server decrypts it, processes it, encrypts the response, and sends it back. The client then decrypts the response.

    > HTTPS should always be used when transmitting sensitive information such as passwords, personal information, or credit card details.

* HTTP Methods (HTTP Verbs)

    * HTTP methods specify the operation to perform on a resource.

        > An endpoint (URL) can support multiple operations. The URL identifies the resource, while the HTTP method specifies the action to perform.

        * GET → Retrieve a resource.

        * POST → Create a new record (resource).

        * PUT → Replace an entire resource.

        * PATCH → Partially update a resource.

        * DELETE → Delete a resource.

        > A resource is anything identified by a URL. It can represent either a collection (e.g., `/users`) or a single record (e.g., `/users/5`).

        > A record is a single instance of a resource (e.g., one user in the users resource).

    * HTTP Method Notes

        * GET requests do not have a request body.

        * GET requests commonly use query parameters to filter, search, sort, or paginate results.

        * POST requests include a payload (typically JSON or form data) used to create a resource.

        * PUT requests replace an entire single resource and usually require all of its fields.

        * PATCH requests update only selected fields of a single resource.

        * DELETE can target:

            * A collection (e.g., `/users`) to delete the collection.

            * A single resource (e.g., `/users/5`) to delete one record.

* HTTP Request

    * An HTTP request is sent from the client to the server and typically contains:

        * HTTP version (e.g., HTTP/1.1, HTTP/2).

        * URL (path to the resource).

        * HTTP method.

        * HTTP request headers.

        * Optional HTTP request body.

    * HTTP Request Headers

        * Contain metadata (information about the request rather than the resource itself) that helps the server process the request.

        > Common examples include Authorization, Cookies, User-Agent, Content-Type, Accept, and Referer.

    * HTTP Request Body

        * Contains the data (payload) sent to the server.

        > Commonly used with POST, PUT, and PATCH requests.

        > The body is typically sent as JSON or form-encoded data.

* HTTP Response

    * An HTTP response is sent from the server back to the client and typically contains:

        * HTTP status code.

        * HTTP response headers.

        * Response body.

    * HTTP Response Headers

        * Contain metadata about the response.

        > Common examples include Content-Type, Content-Length, Set-Cookie, ETag, Last-Modified, Cache-Control, and Date.

    * HTTP Response Body

        * Contains the requested resource or the result of the requested operation.

* HTTP Response Types

    * REST APIs commonly return responses in:

        * JSON

        * XML

        * HTML

        * Plain Text

        * YAML

        > The client specifies the preferred response format using the `Accept` request header (e.g., `Accept: application/json`).

* HTTP Status Codes

    * HTTP status codes indicate the outcome of an HTTP request.

        * 1xx → Informational responses.

        * 2xx → Successful responses.

        * 3xx → Redirection responses.

        * 4xx → Client errors.

        * 5xx → Server errors.

    * Common status codes

        * 102 Processing

            > The server is still processing the request.

        * 200 OK

            > The request completed successfully.

        * 201 Created

            > A new resource was successfully created.

        * 301 Moved Permanently

            > The requested resource has permanently moved to a new URL.

        * 303 See Other

            > The client should retrieve the requested resource from a different URL.

        * 304 Not Modified

            > The requested resource has not changed since the client's last request, so the cached version can be used.

        * 400 Bad Request

            > The client sent an invalid request (e.g., malformed syntax or missing required data).

        * 401 Unauthorized

            > Authentication is required or the provided credentials are invalid.

        * 403 Forbidden

            > The client is authenticated but does not have permission to access the resource.

        * 404 Not Found

            > The requested resource does not exist.

        * 429 Too Many Requests

            > The client has exceeded the API's rate limit and should retry later.

        * 500 Internal Server Error

            > An unexpected error occurred on the server.

        > The meaning of a status code depends on the HTTP method. For example, `200 OK` means a GET request successfully retrieved a resource, a PUT request successfully updated a resource, or a DELETE request successfully deleted a resource.

    * Client Errors (4xx)

        > Usually occur because the client sent an invalid request, requested a non-existent resource, or lacks authentication/authorization.

    * Server Errors (5xx)

        > Usually occur because of bugs, configuration issues, dependency failures, or other server-side problems. Developers should validate requests and handle errors properly to avoid exposing 5xx errors.

# CRUD

* CRUD (Create, Read, Update, Delete)

    * The four fundamental operations that can be performed on a resource.

    * CRUD maps directly to HTTP methods and SQL statements.

        * Create → POST → INSERT.

        * Read → GET → SELECT.

        * Update → PUT / PATCH → UPDATE.

        * Delete → DELETE → DELETE.

    > CRUD is the foundation of every REST API. The HTTP method on a request determines which CRUD operation to perform on the resource.

    > In data engineering, CRUD is what ETL does to a database, but exposed over HTTP.

# HTTP Method Semantics

* Safe and Idempotent

    * HTTP methods have semantic properties that define their expected behavior.

    * Safe

        * A method is safe if it does not modify server state.

        > GET is safe — calling it any number of times changes nothing on the server.

    * Idempotent

        * A method is idempotent if calling it once or multiple times produces the same result.

        > PUT and DELETE are idempotent — repeating the same request has the same effect as executing it once.

        > POST is not idempotent — sending the same POST request twice may create two records.

    * Method Properties

        * GET → Safe, Idempotent.

        * POST → Not safe, Not idempotent.

        * PUT → Not safe, Idempotent.

        * PATCH → Not safe, Not idempotent.

        * DELETE → Not safe, Idempotent.

    > Understanding these properties helps in designing retry logic, caching strategies, and error handling.

# REST API

* REST (Representational State Transfer)

    * An architectural style for designing web APIs.

    * An API is considered RESTful only if it follows REST constraints.

    > REST is popular because it is simple, scalable, and easy to develop and maintain.

* Purpose of a REST API

    * Provides a standardized way for clients to communicate with servers and access or modify resources.

* REST Constraints

    * Client-Server Architecture

        * Separates the client (consumer) from the server (resource provider).

        > The client sends requests, and the server processes them and returns responses.

    * Stateless

        * The server does not store client state between requests.

        * Every request must contain all the information needed to process it (e.g., authentication).

        > Previous requests do not affect future requests unless the client sends the required information again.

    * Cacheable

        * Responses can be cached by browsers, proxies, or servers.

        > Caching reduces server load and improves response time by reusing previously retrieved responses.

    * Layered System

        * The API can be organized into independent layers.

        * Layers can be added, removed, or modified without affecting other layers.

        > Typical layers include Firewall → Load Balancer → Web Server → Database.

    * Uniform Interface

        * Provides a consistent way to identify and manipulate resources.

        * Each resource should have a unique URL.

        * Resources should be represented using standard formats such as JSON or XML.

        * Standard HTTP methods (GET, POST, PUT, PATCH, DELETE) are used consistently across resources.

    * Code on Demand (Optional)

        * The server may send executable code to the client.

        > Example: Returning JavaScript that runs in the client's browser to add functionality.

        > This is the only optional REST constraint and is rarely used in modern REST APIs.

# REST Resources

* A resource is any data exposed by a REST API and identified by a unique URL.

    > A resource can represent either a collection (e.g., `/orders`) or a single record (e.g., `/orders/16`).

* The resource type depends on the endpoint being accessed.

    * `/orders` → Collection of order objects.

    * `/orders/16` → Single order object.

    * `/orders/16/customer` → Customer object associated with order 16.

    * `/orders/16/menu-items` → Collection of menu item objects belonging to order 16.

    * `/menu-items` → Collection of all menu item objects.

* Query parameters filter a resource without changing its type.

    * `/menu-items?category=appetizers`

        > Returns only menu items in the "appetizers" category, but the resource type is still menu items.

* Resource hierarchy

    * Resources can contain or be related to other resources.

    > Example: An order resource can expose its customer or its menu items through nested endpoints such as `/orders/16/customer` and `/orders/16/menu-items`.

* Statelessness (Practical Meaning)

    * The server only processes the information included in the current request.

        * It does not remember previous requests or client state.

        > If you first request `/orders/16` and then request `/menu-items`, the server returns all menu items because it does not remember that you were previously viewing order 16.

        > To retrieve the menu items for order 16, the request must explicitly specify `/orders/16/menu-items`.

    * Every request must include all information required to process it.

        > This includes resource identifiers (e.g., `/orders/16`) and, when required, authentication information.

# REST API Endpoint Naming Conventions

* API endpoints (URIs or URL paths) should clearly describe the resources they represent.

    > Well-designed endpoints improve readability, consistency, and maintainability.

* Use lowercase letters.

    * Good: `/orders`

    * Bad: `/Orders`

* Use hyphens (`-`) to separate multiple words.

    * Good: `/menu-items`

    * Bad: `/menu_items`, `/menuItems`, `/MenuItems`

* Path parameters (variables)

    * Use camelCase and wrap them in curly braces.

    * Examples:

        * `/orders/{orderId}`

        * `/customers/{customerId}/orders`

* Use forward slashes (`/`) to represent hierarchical relationships between resources.

    * `/orders/{orderId}/menu-items`

    * `/orders/{orderId}/customer`

    * `/books/{bookId}/author`

    * `/authors/{authorId}/books`

* Use nouns for resource names.

    * Good:

        * `/books`

        * `/books/{bookId}`

        * `/books/{bookId}/price`

    * Bad:

        * `/getAllBooks`

        * `/getUser/{userId}`

        * `/users/{userId}/delete`

        * `/orders/{orderId}/save`

    > The HTTP method (GET, POST, PUT, PATCH, DELETE) specifies the action, so the endpoint should only represent the resource.

* Avoid CRUD verbs in endpoint names.

    * Bad:

        * `/create`

        * `/read`

        * `/update`

        * `/delete`

    > Use the appropriate HTTP method instead.

* Do not include file extensions in endpoints.

    * Bad:

        * `/orders/1.json`

        * `/orders/1.xml`

    * Good:

        * `/orders/1?format=json`

        * `/orders/1?format=xml`

* Use query parameters for filtering, searching, pagination, or selecting the response format.

    * Filtering:

        * `/menu-items?category=appetizers`

    * Pagination:

        * `/menu-items?page=2&perPage=10`

    * Response format:

        * `/orders/1?format=json`

* Do not add a trailing slash.

    * Good:

        * `/orders`

        * `/orders/{orderId}`

    * Bad:

        * `/orders/`

        * `/orders/{orderId}/`

* Follow a consistent naming strategy throughout the API.

    > Consistent endpoint naming makes APIs easier to understand, use, and maintain.

# REST API Best Practices

* KISS (Keep It Simple, Stupid)

    * Each API endpoint should perform one specific task and do it well.

    > Avoid combining multiple responsibilities into a single endpoint.

* Filtering

    * Allow clients to filter large collections using query parameters.

    > Example: `/menu-items?category=desserts`

* Ordering

    * Allow clients to sort results in ascending or descending order.

    > Common query parameters include `sort`, `order`, or `ordering`.

* Pagination

    * Return large datasets in smaller chunks instead of all at once.

    > Common query parameters include `page`, `perPage`, `limit`, and `offset`.

    > Pagination improves performance and reduces network usage.

* Versioning

    * Version APIs when introducing breaking changes.

    > Common approaches include `/v1/menu-items` and `/v2/menu-items`.

    > Maintaining too many API versions increases complexity and maintenance costs.

* Caching

    * Cache frequently requested responses to reduce server and database load.

    * Update or invalidate the cache when the underlying data changes.

    > Cached responses improve response time and reduce unnecessary database queries.

* Rate Limiting

    * Limit how many requests a client can make within a given time period.

    > Helps prevent abuse and protects API resources.

* Monitoring

    * Continuously monitor API performance and health.

    * Track:

        * Response latency.

        * HTTP status codes (especially 4xx and 5xx errors).

        * Network bandwidth and API usage.

    > Monitoring helps detect performance issues, errors, and potential abuse early.

# REST API Security

* Use HTTPS

    * APIs should always be served over HTTPS to encrypt data transmitted between the client and server.

    > HTTPS uses SSL(Secure Socket Layer)/TLS certificates to provide secure communication.

* Signed URLs

    * Signed URLs grant temporary access to a specific resource.

    * A signature is included with each request URL and verified by the server.

    > Commonly used to ensure requests come from trusted clients.

    > HMAC (Hash-based Message Authentication Code) is a common mechanism for generating and verifying signatures, ensuring message authenticity and integrity.

* Authentication

    * Verifies the identity of the client or user.

    * Token-based authentication is preferred over Basic Authentication.

    > Basic Authentication requires sending the username and password with every request.

    > Token-based authentication requires the user to authenticate once, receive a token, and include that token in subsequent requests (typically in the `Authorization` header).

    > JSON Web Token (JWT) is a common industry-standard format for authentication tokens.

* Authentication Status Codes

    * 401 Unauthorized

        > Authentication failed (e.g., invalid username/password or invalid token).

    * 403 Forbidden

        > Authentication succeeded, but the user does not have permission to perform the requested action.

* CORS (Cross-Origin Resource Sharing)

    * Controls which domains are allowed to access an API.

    > Configured using HTTP headers to allow or restrict requests from specific origins.

* Firewalls

    * Restrict API access based on network rules, such as allowed IP addresses.

    > Provides an additional layer of protection by blocking unauthorized network traffic.

# API Access Control

* Access Control (Authorization)

    * Determines which authenticated users are allowed to access specific API resources and perform specific actions.

    > Protects sensitive data by ensuring only authorized users can access it.

* Privileges (Permissions)

    * A privilege defines a single action a user is allowed to perform.

    > Examples: Create menu items, view orders, delete users.

* Roles

    * A role is a collection of related privileges.

    > Examples: Customer, Manager, Delivery Crew, Administrator.

* Authentication vs Authorization

    * Authentication verifies who the user is.

    * Authorization determines what the authenticated user is allowed to do.

    > Authentication gets you in; authorization determines what you can access.

* Designing Roles

    * Keep privileges small and specific.

    > Fine-grained permissions make the access control system easier to manage and more secure.

* Multiple Approaches to Role Assignment

    * Create broader roles by combining many privileges into a single role.

    * Assign multiple roles to a single user.

    > Assigning multiple roles is generally more flexible because users automatically inherit new privileges added to those roles.

* Benefits of Good Access Control

    * Protects sensitive data.

    * Enforces the principle of least privilege.

    * Simplifies maintenance as the application grows.

    * Reduces future debugging and refactoring.

# Serialization & Deserialization

* Serialization

    * The process of converting a programming language object into a format that can be transmitted over a network or stored.

    > In APIs, serialization converts an internal data structure (e.g., a Python object) into JSON or XML for the response.

* Deserialization

    * The reverse process — converting transmitted data (e.g., JSON from a request body) back into a programming language object.

    > Deserialization typically includes validation to ensure the incoming data meets the expected format and constraints.

* The Data Conversion Chain

    * Outgoing (Serialization): Database Row → Language Object → Dictionary → JSON String.

    * Incoming (Deserialization): JSON String → Dictionary → Validated Data → Language Object → Database Row.

    > A serializer acts as a bidirectional translator and a security layer, controlling exactly which fields are exposed to the outside world.

# ORM (Object-Relational Mapping)

* An ORM is a layer that translates between a programming language and a relational database.

    * Instead of writing SQL directly, developers define database tables as classes and interact with them using language-native methods.

    > Example: Calling `.objects.all()` on a model class translates to `SELECT * FROM table;`.

* How the ORM Works

    * A class represents a database table.

    * A class attribute represents a column.

    * An instance of the class represents a single row.

    * The ORM translates method calls into SQL, executes them, and wraps the results back into objects.

* Benefits

    * No raw SQL required for common operations.

    * Database-agnostic — switching databases requires changing configuration, not rewriting queries.

    * Migrations — the ORM tracks schema changes over time as version-controlled files.

    * Protection against SQL injection by default.

    > Common ORMs include Django ORM, SQLAlchemy (Flask/FastAPI), Hibernate (Java), Entity Framework (C#), and Sequelize (Node.js).

# URL Routing & Views

* URL Routing

    * The process of mapping incoming request URLs to the appropriate handler (view) that will process the request.

    * A routing table defines URL patterns and associates each pattern with a specific view.

    > When a request arrives, the router checks the URL against its patterns top-to-bottom and forwards the request to the first matching view.

    * Multi-Level Routing

        * Large applications typically split routing into two levels: a project-level router that delegates to app-level routers.

        > This keeps each module responsible for its own routes, improving organization and maintainability.

    * URL Parameters

        * Dynamic segments in a URL pattern that capture values from the request path.

        > Example: In `/orders/{orderId}`, the `{orderId}` segment captures the specific order ID from the URL and passes it to the view.

* Views (Handlers)

    * A view is a function or class that receives an HTTP request, executes business logic, and returns an HTTP response.

    * In class-based views, each HTTP method (GET, POST, PUT, DELETE) maps to a corresponding method on the class.

    > The view is where routing, serialization, and database access come together — it orchestrates the full request-to-response cycle.

    * Abstraction Levels

        * Most frameworks provide increasing levels of abstraction for views.

            * Manual views — full control, you write every step (query, serialize, validate, respond).

            * Generic views — pre-built views for common patterns (list, create, retrieve, update, delete), configured with a model and serializer.

            * ViewSets — combine all CRUD operations into a single class with auto-generated URL patterns.

        > Higher abstraction reduces boilerplate but trades off granular control. Choose the level that matches the complexity of the endpoint.

# The Request Lifecycle

* When a client sends an API request, it passes through several layers before a response is returned.

    * Client → Sends an HTTP request (method, URL, headers, optional body).

    * Web Server → Receives the raw request and forwards it to the application.

    * Middleware → Processes the request through a chain (e.g., authentication, logging, CORS).

    * Router (URL Resolver) → Matches the URL to the appropriate view or handler.

    * View / Handler → Executes business logic, queries the database, and prepares the response.

    * Serializer → Converts database objects to the response format (e.g., JSON) or validates incoming request data.

    * Response → Sent back through the middleware chain (in reverse) to the web server and then to the client.

    > Each layer has a single responsibility. This separation of concerns makes the system easier to test, maintain, and scale.

* Middleware

    * Software components that sit between the request and the view, processing every request and response.

    > Common middleware includes authentication, logging, CORS handling, and request parsing.

    > Middleware runs in order on the way in (request) and in reverse order on the way out (response).
