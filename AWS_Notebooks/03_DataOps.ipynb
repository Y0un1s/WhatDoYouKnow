# DataOps Overview

* **DataOps** is a collection of practices and cultural habits focused on building reliable data systems and delivering high-quality data products.

* DataOps evolved from **DevOps**, adapting software engineering practices to the needs of data engineering.

> While DataOps and DevOps have different goals, they share many automation and operational practices.

---

### The Three Pillars of DataOps

DataOps is built on three core pillars:

* Automation.
* Observability and Monitoring.
* Incident Response.

```text
DataOps
├── Automation
├── Observability & Monitoring
└── Incident Response
```

---

### Automation

* Automation reduces manual work and improves consistency across the data lifecycle.

Common automation practices include:

* CI/CD.
* Infrastructure as Code (IaC).
* Pipeline scheduling.
* Workflow orchestration.

---

### Observability & Monitoring

* Observability helps engineers understand the health of their data pipelines.

Typical monitoring includes:

* Data quality.
* Pipeline execution.
* System health.
* Performance metrics.

> Observability helps detect problems before they affect downstream users.

---

### Incident Response

* Incident Response focuses on identifying, responding to, and recovering from pipeline failures.

It includes:

* Investigating issues.
* Restoring pipeline functionality.
* Preventing similar failures in the future.

> Incident response is as much about team processes and communication as it is about technology.

---

### Continuous Integration & Continuous Delivery (CI/CD)

* DataOps adopts **Continuous Integration (CI)** and **Continuous Delivery (CD)** from DevOps.

#### Continuous Integration (CI)

* Automatically reviews and tests code whenever changes are made.

Benefits:

* Detects bugs early.
* Prevents broken code from reaching production.

---

#### Continuous Delivery (CD)

* Automatically deploys tested code into production.

Benefits:

* Faster releases.
* Consistent deployments.
* Reduced manual effort.

---

### CI/CD in DataOps

CI/CD applies to both:

* Pipeline code.
* Data.

Examples include:

* ETL scripts.
* SQL transformations.
* Database population scripts.
* Data validation logic.

> Data pipelines can be maintained just like software applications.

---

### Pipeline Automation

There are several ways to execute a data pipeline.

#### Manual Execution

```text
Engineer
    │
    ▼
Run Pipeline
```

* Every stage is started manually.

---

#### Scheduled Execution

```text
Every Day at 2:00 AM
        │
        ▼
Run Pipeline
```

* Pipelines run automatically according to a schedule.

---

#### Workflow Orchestration

Pipelines can be represented as a **Directed Acyclic Graph (DAG)**.

Example:

```text
Extract
    │
    ▼
Transform
    │
    ▼
Load
```

Workflow orchestration tools manage:

* Task execution.
* Dependencies.
* Retries.
* Scheduling.

Example:

* Apache Airflow.

---

### Version Control

* Version control records every change made to code.

Benefits:

* Track history.
* Compare versions.
* Roll back to previous versions.

Common example:

* GitHub.

---

### Data Versioning

DataOps extends version control beyond code.

It also tracks:

* Datasets.
* Pipeline outputs.
* Intermediate data.

Benefits:

* Recover previous data versions.
* Investigate data issues.
* Improve reproducibility.

> Data can be versioned just like application code.

---

### Infrastructure as Code (IaC)

* **Infrastructure as Code (IaC)** means defining infrastructure using code instead of manually creating cloud resources.

Instead of manually creating resources:

```text
AWS Console

Create EC2
Create S3
Create RDS
```

You write configuration files:

```text
Terraform Code

↓

terraform apply

↓

Infrastructure Created
```

Benefits:

* Repeatable deployments.
* Consistent environments.
* Easy automation.
* Infrastructure version control.

---

### Infrastructure Versioning

Because infrastructure is defined as code:

* Infrastructure changes are tracked.
* Previous infrastructure versions can be restored.
* Deployments become reproducible.

> Rolling back infrastructure is simply rolling back the infrastructure code.

---

### Common Infrastructure as Code Tools

Examples include:

* Terraform.
* AWS CloudFormation.

These tools can automatically provision cloud resources such as:

* EC2 instances.
* Databases.
* Networking.
* Storage.
* IAM resources.

---

### DataOps and Other Data Engineering Undercurrents

DataOps overlaps with several areas of data engineering.

#### Software Engineering

DataOps borrows:

* CI/CD.
* Version control.
* Infrastructure as Code.
* Automated testing.

---

#### Data Management

DataOps improves:

* Data versioning.
* Data lifecycle management.
* Data reliability.
* Data quality.

> DataOps combines software engineering best practices with data engineering to build reliable, automated, and maintainable data systems.

# Infrastructure as Code (IaC)

* **Infrastructure as Code (IaC)** is the practice of defining, deploying, and managing infrastructure using code instead of manual configuration.

* IaC automates the creation of cloud resources such as:

  * Networking.
  * Security.
  * Compute.
  * Storage.
  * Databases.
  * Analytics services.

> Instead of manually creating infrastructure through a cloud console, you define the desired infrastructure in configuration files.

---

### History of Infrastructure as Code

* The idea of IaC existed before cloud computing.

* In the 1970s, engineers wrote **Bash scripts** to automate server configuration.

* As cloud computing grew, dedicated IaC tools emerged.

Major milestones:

* **2006** – AWS introduced Amazon EC2.
* **Early 2010s** – Tools such as Terraform, CloudFormation, and Ansible became popular.

> Modern IaC tools replaced manual configuration and complex scripting with declarative configuration files.

---

### Benefits of Infrastructure as Code

IaC provides several advantages.

* Automation.
* Repeatable deployments.
* Consistent environments.
* Version control.
* Faster infrastructure provisioning.
* Easier collaboration.

---

### Infrastructure Managed by IaC

Infrastructure as Code can provision resources including:

* Virtual Private Clouds (VPCs).
* Subnets.
* EC2 instances.
* S3 buckets.
* Databases.
* IAM resources.
* ETL jobs.
* Networking components.

---

### Popular Infrastructure as Code Tools

Common IaC tools include:

* Terraform.
* AWS CloudFormation.
* Ansible.

---

### Terraform vs CloudFormation

| Terraform | CloudFormation |
|-----------|----------------|
| Multi-cloud | AWS only |
| Created by HashiCorp | Native AWS service |
| Uses HCL | Uses JSON or YAML templates |

> Many organizations use Terraform, CloudFormation, or both depending on their infrastructure requirements.

---

### Terraform Configuration Files

Terraform infrastructure is defined using configuration files.

A resource follows this general structure:

```hcl
resource "<resource_type>" "<resource_name>" {

  configuration

}
```

Example:

```hcl
resource "aws_s3_bucket" "data_lake" {

  bucket = "my-data-lake"

}
```

Components:

* `resource`

  * Declares a new infrastructure resource.

* `aws_s3_bucket`

  * Resource type.
  * Specifies the provider (AWS) and resource (S3 Bucket).

* `data_lake`

  * Logical name used within Terraform.

* Configuration block

  * Defines the resource settings using key-value pairs.

---

### Resource Type

The resource type identifies:

* Cloud provider.
* Resource being created.

Examples:

| Resource Type | Creates |
|--------------|----------|
| `aws_s3_bucket` | Amazon S3 Bucket |
| `aws_instance` | EC2 Instance |
| `aws_vpc` | Virtual Private Cloud |
| `google_compute_instance` | GCP Virtual Machine |

---

### HashiCorp Configuration Language (HCL)

* Terraform uses **HashiCorp Configuration Language (HCL).**

Characteristics:

* Human-readable.
* Declarative.
* Cross-platform.
* Supports multiple cloud providers.

Example providers:

* AWS.
* Google Cloud Platform (GCP).
* Microsoft Azure.

---

### Declarative vs Imperative

Terraform uses a **Declarative** approach.

With declarative programming:

* You specify **what** the infrastructure should look like.

Terraform determines:

* How to create it.
* How to update it.
* How to maintain it.

This desired configuration is called the:

* **Desired State**

---

### Desired State

The desired state describes the final infrastructure you want.

Example:

```text
Desired State

5 EC2 Instances
1 VPC
2 S3 Buckets
1 RDS Database
```

Terraform compares:

* Desired State.
* Current Infrastructure.

It then performs only the necessary changes.

---

### Idempotency

Terraform is **idempotent**.

**Idempotent** means:

> Running the same Terraform configuration multiple times always produces the same final infrastructure.

Example:

First execution:

```text
Terraform Apply

↓

Creates 5 EC2 Instances
```

Second execution:

```text
Terraform Apply

↓

No Changes Needed
```

Terraform detects that the infrastructure already matches the desired state.

---

### Infrastructure Reconciliation

Before making changes, Terraform compares:

```text
Current Infrastructure

        │

Compare

        │

Desired State
```

Terraform then:

* Creates missing resources.
* Updates resources with incorrect configurations.
* Leaves correctly configured resources unchanged.

---

### Declarative Example

Desired configuration:

```text
5 EC2 Instances
```

Current infrastructure:

```text
3 EC2 Instances
```

Terraform automatically creates:

```text
2 Additional EC2 Instances
```

Result:

```text
5 EC2 Instances
```

---

### Imperative Configuration

Imperative languages require specifying every step.

Example using Bash:

```text
Create EC2 #1

Create EC2 #2

Create EC2 #3

Create EC2 #4

Create EC2 #5
```

If the script is executed again:

```text
Creates 5 More EC2 Instances
```

The script does not automatically determine whether the resources already exist.

---

### Declarative vs Imperative

| Declarative (Terraform) | Imperative (Bash) |
|--------------------------|-------------------|
| Specify the desired result | Specify every step |
| Automatically determines required actions | Executes commands exactly as written |
| Idempotent | Not inherently idempotent |
| Safe to execute repeatedly | May create duplicate resources |
| Maintains desired infrastructure state | Performs commands regardless of current state |

> Terraform focuses on describing **what** infrastructure should exist, while imperative tools focus on **how** to create it.

# Terraform Workflow & Configuration

* Terraform follows a consistent workflow for provisioning infrastructure.

The workflow is:

```text
Write Configuration

        │

terraform init

        │

terraform plan

        │

terraform apply

        │

Infrastructure Created
```

> Terraform first prepares the environment, then creates an execution plan, and finally provisions the infrastructure.

---

### Step 1: Write Configuration Files

* Infrastructure is defined using Terraform configuration files.

* Every Terraform configuration file uses the `.tf` extension.

Example:

```text
main.tf
```

These files describe the infrastructure Terraform should create.

---

### Typical Terraform File Structure

A Terraform configuration commonly contains five sections.

1. Terraform settings.
2. Providers.
3. Resources.
4. Input Variables (Optional).
5. Output Values (Optional).

```text
main.tf

Terraform Settings

Providers

Resources

Variables

Outputs
```

---

### Terraform Blocks

Terraform configuration files are organized into **Blocks**.

A block follows this general structure.

```hcl
keyword "label" {

    arguments

}
```

A block consists of:

* Keyword.
* Optional labels.
* Body enclosed in curly braces.
* Arguments (key-value pairs).
* Nested blocks (optional).

---

### Terraform Block

The **Terraform Block** defines Terraform settings.

Typical settings include:

* Required providers.
* Required Terraform version.

Example:

```hcl
terraform {

    required_providers {

    }

}
```

---

### Providers

A **Provider** is a plugin that allows Terraform to communicate with external systems.

Examples include:

* AWS.
* Azure.
* Google Cloud.

> A Terraform provider is **not** the cloud provider itself. It is the plugin Terraform uses to interact with that cloud provider.

---

### Required Providers

Each provider declaration specifies:

* Local name.
* Source.
* Optional version.

Example:

```hcl
terraform {

  required_providers {

    aws = {

      source = "hashicorp/aws"

    }

  }

}
```

---

### Provider Block

After declaring the provider, you configure it.

Example:

```hcl
provider "aws" {

    region = "us-east-1"

}
```

The provider block specifies settings such as:

* Region.
* Credentials.
* Other provider-specific configuration.

---

### Resource Block

A **Resource Block** tells Terraform what infrastructure to create.

General syntax:

```hcl
resource "<resource_type>" "<resource_name>" {

    configuration

}
```

Example:

```hcl
resource "aws_instance" "web_server" {

    ...

}
```

---

### Resource Type

The resource type identifies:

* Provider.
* Resource.

Examples:

| Resource Type | Creates |
|--------------|----------|
| `aws_instance` | EC2 Instance |
| `aws_s3_bucket` | S3 Bucket |
| `aws_vpc` | Virtual Private Cloud |

The format is:

```text
provider_resource
```

Example:

```text
aws_instance
```

---

### Resource Name

The second string is the logical name assigned to the resource.

Example:

```hcl
resource "aws_instance" "web_server"
```

Here:

* Resource Type

```text
aws_instance
```

* Resource Name

```text
web_server
```

The resource can later be referenced as:

```text
aws_instance.web_server
```

---

### Resource Arguments

Inside a resource block, you configure the resource using arguments.

Example:

```hcl
resource "aws_instance" "web_server" {

    ami = "ami-xxxx"

    instance_type = "t2.micro"

}
```

Most arguments are optional.

Required arguments depend on the resource.

---

### EC2 Required Arguments

An EC2 instance requires at least:

* AMI.
* Instance Type.

#### AMI (Amazon Machine Image)

An **AMI** is a software template that defines the operating environment of an EC2 instance.

An AMI includes:

* Operating System.
* System Architecture.
* Pre-installed Software.

Example:

```text
Amazon Linux

Ubuntu

Windows Server
```

---

#### Instance Type

The instance type defines the hardware configuration.

Examples:

```text
t2.micro

t3.small

m5.large
```

It determines:

* CPU.
* Memory.
* Network performance.

---

### Tags

Resources can also include optional tags.

Example:

```hcl
tags = {

    Name = "Example Server"

}
```

Tags help organize cloud resources.

---

### Terraform Commands

Terraform uses three primary commands.

---

#### terraform init

```bash
terraform init
```

Purpose:

* Initializes the Terraform project.
* Downloads required providers.
* Creates the hidden `.terraform` directory.

> Run this command before any other Terraform command.

---

#### terraform plan

```bash
terraform plan
```

Purpose:

* Compares the desired infrastructure with the current infrastructure.
* Generates an execution plan.

The execution plan shows what Terraform intends to:

* Create.
* Update.
* Destroy.

---

### Plan Symbols

Terraform uses symbols to indicate planned actions.

| Symbol | Meaning |
|---------|----------|
| `+` | Create |
| `~` | Update |
| `-` | Destroy |

---

#### terraform apply

```bash
terraform apply
```

Purpose:

* Displays the execution plan.
* Requests user confirmation.
* Applies the approved changes.
* Creates or updates the infrastructure.

---

### Complete Terraform Workflow

```text
Write Configuration Files

        │

terraform init

Downloads Providers

        │

terraform plan

Generates Execution Plan

        │

terraform apply

Applies Changes

        │

Cloud Infrastructure Created
```

> Terraform always follows the sequence: **Write → Initialize → Plan → Apply**.

# Advanced Terraform Concepts

* Terraform supports additional block types beyond resources.

These include:

* Variables.
* Output Values.
* Data Sources.
* Modules.

These features make Terraform configurations reusable, maintainable, and easier to organize.

---

### Input Variables

* **Input Variables** allow you to parameterize your Terraform configuration.

Instead of hardcoding values:

```hcl
region = "us-east-1"
```

You define a variable:

```hcl
variable "region" {

  type = string

}
```

Then reference it:

```hcl
provider "aws" {

  region = var.region

}
```

> Variables allow the same configuration to be reused with different values.

---

### Variable Declaration

A variable typically includes:

* Name.
* Description (Optional).
* Type (Optional).
* Default Value (Optional).

Example:

```hcl
variable "server_name" {

  description = "EC2 Server Name"

  type = string

  default = "web-server"

}
```

---

### Variable References

Variables are accessed using:

```text
var.<variable_name>
```

Example:

```hcl
var.region

var.server_name
```

---

### Assigning Variable Values

There are two common ways to assign values.

#### Command Line

```bash
terraform apply \
-var="region=us-east-1"
```

---

#### tfvars File

Terraform automatically loads files ending with:

```text
.tfvars
```

Example:

```text
terraform.tfvars
```

Contents:

```hcl
region = "us-east-1"

server_name = "production-server"
```

> Using `.tfvars` files is the preferred approach for managing configuration values.

---

### Default Values

Variables may include default values.

Example:

```hcl
variable "region" {

    default = "us-east-1"

}
```

If no value is provided:

* Terraform uses the default value.

If no default exists:

* Terraform prompts the user for a value.

---

### Resource Attributes

Every Terraform resource exposes attributes.

Examples for an EC2 instance include:

* Instance ID.
* ARN.
* Public IP.
* Private IP.

These attributes can be referenced elsewhere in the configuration.

---

### Output Values

* **Output Values** expose resource information after Terraform finishes provisioning.

Common uses:

* Display resource information.
* Share values with other Terraform configurations.
* Reference outputs from modules.

---

### Output Block

Example:

```hcl
output "server_id" {

    value = aws_instance.web_server.id

}
```

Components:

* Output Name.
* Value.

---

### Accessing Resource Attributes

Resource attributes follow this syntax:

```text
resource_type.resource_name.attribute
```

Example:

```hcl
aws_instance.web_server.id

aws_instance.web_server.arn

aws_instance.web_server.public_ip
```

---

### Terraform Output Command

Display every output:

```bash
terraform output
```

Display a single output:

```bash
terraform output server_id
```

---

### Organizing Terraform Files

Large Terraform projects are typically split into multiple files.

Common layout:

```text
providers.tf

main.tf

variables.tf

outputs.tf

terraform.tfvars
```

Purpose of each file:

| File | Purpose |
|------|----------|
| `providers.tf` | Terraform settings and providers |
| `main.tf` | Resource definitions |
| `variables.tf` | Input variables |
| `outputs.tf` | Output values |
| `terraform.tfvars` | Variable values |

> Terraform automatically loads every `.tf` file in the directory.

---

### Data Sources

* A **Data Source** allows Terraform to read information about existing infrastructure instead of creating it.

Terraform refers to these existing resources as **Data Sources**.

---

### Resource vs Data Source

| Resource | Data Source |
|----------|-------------|
| Creates infrastructure | Reads existing infrastructure |
| Managed by Terraform | Managed outside the current Terraform configuration |

---

### Data Block

General syntax:

```hcl
data "<resource_type>" "<name>" {

    ...

}
```

Example:

```hcl
data "aws_subnet" "selected" {

    id = "subnet-123"

}
```

---

### Accessing Data Source Attributes

Use:

```text
data.resource_type.name.attribute
```

Example:

```hcl
data.aws_subnet.selected.id
```

---

### Common Uses of Data Sources

Data sources allow Terraform to:

* Read existing VPCs.
* Read existing Subnets.
* Read existing Security Groups.
* Find the latest AMI automatically.

Example:

```text
Terraform

↓

Find Latest Amazon Linux AMI

↓

Launch EC2 Instance
```

> Data sources make Terraform configurations more dynamic and reusable.

---

### Modules

A **Module** is a reusable collection of Terraform configuration files.

Think of a module as a reusable package of infrastructure.

Example:

```text
Website Module

├── main.tf

├── variables.tf

├── outputs.tf

└── providers.tf
```

---

### Why Use Modules?

Modules improve:

* Reusability.
* Maintainability.
* Organization.
* Separation of concerns.

Instead of copying infrastructure repeatedly, create it once as a module.

---

### Module Block

Modules are called from the root configuration.

Example:

```hcl
module "website" {

    source = "./website"

}
```

The `source` specifies the module directory.

---

### Passing Variables to Modules

Modules receive input values.

Example:

```hcl
module "website" {

    source = "./website"

    server_name = var.server_name

}
```

---

### Accessing Module Outputs

Module outputs are referenced using:

```text
module.<module_name>.<output_name>
```

Example:

```hcl
module.website.server_id

module.website.server_arn
```

---

### Root Directory vs Module

The root configuration cannot directly access files inside a module.

Communication occurs through:

* Input Variables.
* Output Values.

```text
Root Configuration

        │

Input Variables

        ▼

Module

        │

Output Values

        ▼

Root Configuration
```

---

### Updating Modules

Whenever modules are:

* Added.
* Removed.
* Modified.

Run:

```bash
terraform init
```

This allows Terraform to download or update the required modules.

---

### Terraform Project Structure

A well-organized Terraform project typically looks like:

```text
terraform-project/

├── main.tf
├── providers.tf
├── variables.tf
├── outputs.tf
├── terraform.tfvars
└── modules/
    └── website/
        ├── main.tf
        ├── providers.tf
        ├── variables.tf
        └── outputs.tf
```

> As Terraform projects grow, separating configurations into variables, outputs, providers, data sources, and reusable modules makes the infrastructure easier to maintain and reuse.

# Data Observability & Monitoring

* **Data Observability** is the practice of monitoring the health and quality of data throughout its lifecycle.

* Its goal is to detect data issues as early as possible before they impact downstream users.

> A data pipeline can appear to be functioning correctly while still producing poor-quality data.

---

### Why Data Quality Matters

High-quality data should be:

* Accurate.
* Complete.
* Discoverable.
* Available when needed.
* Consistent with stakeholder expectations.

It should also:

* Follow the expected schema.
* Use the correct data definitions.
* Be suitable for downstream use.

> High-quality data creates business value by enabling trustworthy analytics and decision-making.

---

### Low-Quality Data

Low-quality data may be:

* Inaccurate.
* Incomplete.
* Outdated.
* Inconsistent.
* Missing required values.
* Using an unexpected schema.

> Providing incorrect data is often worse than providing no data at all because it can lead to incorrect business decisions.

---

### Data Failures vs System Failures

A software failure is usually obvious.

Examples:

* Application crash.
* HTTP 404 error.
* Service unavailable.

These failures immediately alert engineers.

---

A data failure is often hidden.

Example:

```text
Pipeline

✓ Running Successfully

↓

Dashboard Updated

↓

Incorrect Data
```

The pipeline appears healthy, but the data is wrong.

> Hidden data quality issues are often more dangerous than complete system failures.

---

### Example of a Hidden Data Quality Issue

Suppose a company sells products in Europe.

Originally:

```text
Price = USD
```

The source system changes to:

```text
Price = EUR
```

The pipeline continues running normally.

Analytics dashboards now report:

```text
Revenue ↓ 15%
```

Management believes sales have dropped dramatically.

In reality:

* Sales are unchanged.
* Only the currency changed.

> The pipeline never failed—only the meaning of the data changed.

---

### Responsibility for Data Quality

Upstream systems may change unexpectedly.

Examples:

* New columns.
* Renamed fields.
* Different units.
* Schema changes.
* Business logic changes.

Even when upstream systems introduce the problem:

* Data engineers are responsible for detecting it.
* Data engineers are responsible for protecting downstream users.

> Upstream changes should be expected and monitored.

---

### What is Data Observability?

Data Observability answers questions such as:

* Is the data current?
* Is it complete?
* Is it accurate?
* Has the schema changed?
* Is the pipeline producing the expected data?

Like application monitoring tracks software health, data observability tracks data health.

---

### Common Data Quality Questions

A good monitoring system should answer questions like:

* Is the data up to date?
* Is the data complete?
* Are values within expected ranges?
* Is the NULL rate normal?
* Has the schema changed?

---

### The Five Pillars of Data Observability

Barr Moses defines five pillars that describe the health of data.

```text
Data Observability

├── Distribution (Internal Quality)
├── Freshness
├── Volume
├── Lineage
└── Schema
```

---

### 1. Distribution (Internal Quality)

Distribution measures the internal quality of the data.

Typical checks include:

* NULL percentage.
* Unique value percentage.
* Summary statistics.
* Expected value ranges.
* Data distributions.

Example:

```text
Age

Expected:

18–100

Received:

-15
```

This indicates a data quality issue.

> Distribution monitoring ensures the data matches expected characteristics.

---

### 2. Freshness

Freshness measures how recently the data was updated.

Typical questions:

* When was the last update?
* Is the data arriving on schedule?

Example:

```text
Expected

Hourly Updates

↓

Last Update

8 Hours Ago
```

This indicates stale data.

> Freshness monitoring detects delayed or missing data updates.

---

### 3. Volume

Volume measures how much data is being ingested.

Unexpected changes may indicate problems.

Examples:

* Sudden drop in records.
* Sudden spike in records.

Example:

```text
Expected

100,000 Records

↓

Received

15,000 Records
```

Possible causes:

* Pipeline failure.
* Missing data.
* Source system outage.

---

### 4. Lineage

Lineage describes the complete journey of data.

Example:

```text
Source Database

↓

Kafka

↓

Spark

↓

Data Warehouse

↓

Power BI
```

Lineage helps answer:

> "Where did the problem originate?"

It allows engineers to trace data from its source through every transformation.

---

### 5. Schema

Schema monitoring detects structural changes.

Examples:

* Added columns.
* Removed columns.
* Renamed columns.
* Changed data types.

Example:

```text
Before

price FLOAT

↓

After

price STRING
```

Unexpected schema changes can break pipelines or produce incorrect results.

> Schema monitoring detects structural changes before they affect downstream systems.

---

### Summary of the Five Pillars

| Pillar | What It Monitors |
|---------|------------------|
| Distribution | Internal data quality (NULLs, uniqueness, ranges, statistics) |
| Freshness | How recent the data is |
| Volume | Amount of incoming data |
| Lineage | End-to-end data flow |
| Schema | Structure and data type changes |

# Data Observability & Monitoring

* **Data Observability** is the practice of monitoring the health and quality of data throughout its lifecycle.

* Its goal is to detect data issues before they affect downstream users.

> A pipeline can be running successfully while still producing poor-quality data.

---

### Why Data Quality Matters

High-quality data should be:

* Accurate.
* Complete.
* Discoverable.
* Available when needed.
* Consistent with stakeholder expectations.
* Follow the expected schema.

> Poor-quality data is often worse than no data because it can lead to incorrect business decisions.

---

### Data Failures vs System Failures

**System failures** are usually obvious.

Examples:

* Application crash.
* HTTP 404 error.
* Service unavailable.

**Data failures** are often hidden.

The pipeline continues running, but the data becomes incorrect.

> Hidden data quality issues are often more dangerous than complete system failures.

---

### Hidden Data Quality Issues

A pipeline may continue working even after upstream changes.

Examples:

* Currency changes (USD → EUR).
* Unit changes.
* Business logic changes.
* Schema changes.

> The pipeline doesn't fail—the meaning or quality of the data changes.

---

### Responsibility for Data Quality

Data engineers are responsible for ensuring data quality, even when issues originate from upstream systems.

Unexpected changes should be:

* Expected.
* Detected.
* Monitored.

---

### What Does Data Observability Monitor?

Data observability helps answer questions such as:

* Is the data current?
* Is it complete?
* Is it accurate?
* Has the schema changed?
* Is the pipeline producing the expected data?

---

### Five Pillars of Data Observability

```text
Data Observability

├── Distribution
├── Freshness
├── Volume
├── Lineage
└── Schema
```

---

### 1. Distribution

Measures the internal quality of the data.

Checks include:

* NULL percentage.
* Unique values.
* Value ranges.
* Summary statistics.

---

### 2. Freshness

Measures how recent the data is.

Checks include:

* Last update time.
* Update frequency.
* Data latency.

---

### 3. Volume

Measures the amount of incoming data.

Looks for:

* Unexpected spikes.
* Unexpected drops.

---

### 4. Lineage

Tracks the complete journey of data from source to destination.

Helps identify:

* Where data originated.
* Where transformations occurred.
* Where an issue was introduced.

---

### 5. Schema

Monitors changes to the data structure.

Checks include:

* Added or removed columns.
* Renamed columns.
* Data type changes.

---

### Summary of the Five Pillars

| Pillar | What It Monitors |
|---------|------------------|
| Distribution | Internal data quality (NULLs, uniqueness, ranges, statistics) |
| Freshness | How recent the data is |
| Volume | Amount of incoming data |
| Lineage | End-to-end data flow |
| Schema | Structure and data type changes |

> Data observability continuously monitors these five pillars to detect data issues before they impact downstream systems or business decisions.

# Great Expectations Workflow

* **Great Expectations (GX)** is an open-source framework for validating data quality.

* It allows you to define **Expectations** (data quality rules) and validate your data against them.

---

### Typical Workflow

```text
Data Context
      ↓
Data Source
      ↓
Data Asset
      ↓
Batch Request
      ↓
Expectation Suite
      ↓
Validator
      ↓
Checkpoint
      ↓
Validation Results
      ↓
Data Docs
```

---

### 1. Data Context

The **Data Context** is the entry point to a Great Expectations project.

It is used to:

* Connect to data sources.
* Create expectations.
* Run validations.
* Manage project configuration.

---

### 2. Data Source

A **Data Source** defines where the data comes from.

Examples:

* SQL database.
* CSV file.
* Local file system.
* Amazon S3.
* Pandas DataFrame.

---

### 3. Data Asset

A **Data Asset** is the specific data you want to validate.

Examples:

* A database table.
* A CSV file.
* A SQL query.
* A collection of files.

---

### 4. Batch

A **Batch** is a subset of a data asset.

Examples:

* Monthly sales.
* Daily transactions.
* Data for a specific store.
* Entire dataset.

---

### 5. Batch Request

A **Batch Request** specifies which batch of data should be validated.

It is passed to the validator during validation.

---

### 6. Expectations

An **Expectation** is a rule that data must satisfy.

Common expectations:

* Values are unique.
* No NULL values.
* Values are within a valid range.
* Column exists.
* Correct data type.

---

### 7. Expectation Suite

An **Expectation Suite** is a collection of expectations for a dataset.

Example:

* Payment ID is unique.
* Customer ID is not NULL.
* Amount is non-negative.

---

### 8. Validator

The **Validator** applies an Expectation Suite to a Batch Request and checks whether the data passes the tests.

---

### 9. Checkpoint

A **Checkpoint** automates the validation process.

It combines:

* Batch Request.
* Expectation Suite.

Then runs all validations automatically.

---

### 10. Backend Stores

Great Expectations stores project metadata in different stores.

| Store | Purpose |
|--------|---------|
| Expectation Store | Stores Expectation Suites |
| Validation Store | Stores validation results |
| Checkpoint Store | Stores checkpoint configurations |
| Data Docs Store | Stores validation reports |

---

### Data Docs

**Data Docs** are automatically generated reports showing:

* Validation results.
* Passed and failed expectations.
* Statistics.
* Metadata.

---

### Common Workflow Commands

Initialize a project:

```bash
great_expectations init
```

Get the Data Context:

```python
context = gx.get_context()
```

Add a SQL data source:

```python
context.sources.add_sql(...)
```

Create a data asset:

```python
datasource.add_table_asset(...)
```

Build a batch request:

```python
asset.build_batch_request()
```

Create an Expectation Suite:

```python
context.add_or_update_expectation_suite(...)
```

Create a validator:

```python
context.get_validator(...)
```

Add expectations:

```python
validator.expect_column_values_to_be_unique(...)
validator.expect_column_values_to_not_be_null(...)
validator.expect_column_min_to_be_between(...)
```

Save the Expectation Suite:

```python
validator.save_expectation_suite()
```

Create a checkpoint:

```python
context.add_or_update_checkpoint(...)
```

Run validation:

```python
checkpoint.run()
```

Generate Data Docs:

```python
context.build_data_docs()
```

### Summary

A typical Great Expectations workflow is:

1. Create a **Data Context**.
2. Connect to a **Data Source**.
3. Select a **Data Asset**.
4. Build a **Batch Request**.
5. Create an **Expectation Suite**.
6. Validate data using a **Validator** or **Checkpoint**.
