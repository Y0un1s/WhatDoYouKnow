# Data Ingestion

* In some sense you can think of nearly all data as representing a continuous stream of events, these events might be smth happening in the real world like a stock price changing or your favorite sports team scoring a goal or user clicking a button on a website

* When it comes to data ingestion, whatever system you are working with, you have data generated somewhere as a continuous stream of events that is unbounded, meaning the stream doesn't have a particular beginning or end, if you ingest those events individually one at a time as they are generated then that's an example of stream ingestion, if instead you enforce some boundaries on that stream and ingest all the data within those boundaries as a single unit then that's batch ingestion
    > you can bound data by many ways for example size or time 

* at some point as you continue to increase the frequency of ingestion `with super high frequency batch ingestion` you will eventually be implementing streaming ingestion

* so batch ingestion is not very different than streaming ingestion

* The choice of ingestion frequency depends on:
    
    * The source system you are working with

    * The end use case you are serving

* Batch processing of large chunks of data was the only possible way to ingest data, but as technologies advanced in recent years micro-batch processing became possible to blur the line between batch and streaming
    
    > Batch processing was the dominant model. Streaming processing already existed for specialized applications, and later micro-batching emerged as a practical bridge between batch and low-latency processing. Modern stream processing frameworks then made true large-scale event-by-event streaming more accessible.

    > There is no rule where you can call the official cutoff between batch and micro-batch or exactly how close to real time data needs to be processed to be considered streaming 

* ways to ingest data from databases:
    
    * You can use connectors like JDBC or ODBC

        > JDBC is the standard Java API for connecting Java applications to relational databases. It uses database-specific JDBC drivers to execute SQL queries and retrieve or modify data.

        > ODBC is a language-independent standard for connecting applications to relational databases. It uses ODBC drivers to provide a common interface for applications such as Excel, Power BI, Python, and C/C++ to access different database systems.

    * Use a serverless ingestion tool, such as AWS Glue ETL, which can be configured to connect to a source database and ingest data automatically.

* Once connected, you can choose an ingestion strategy:
    
    * running ingestion at regular intervals (batch) 
    
    * triggering ingestion after a certain amount of new data has been recorded (micro-batch)
    
    * processing changes continuously if the source supports streaming.

* Ingesting data from an API:
  
    * You will be required to set up a connection based on this API's protocol and you will be subject to that API's various constraints and limitations when it comes to how much data you can ingest in one go or how frequently you can call the API

    * This process will be somewhat frustrating and it will involve:
        
        * reading documentation
        
        * communicating with external data owners
        
        * and writing and maintaining your own custom API connection code
    
    * use existing solutions to connecting to an API whenever possible

* Ways to ingest data from files:

    * Use an object storage system as the data source.

        > Examples include Amazon S3, Azure Blob Storage, and Google Cloud Storage.

    * Manually download files when ingestion cannot be fully automated.

        > Some business processes still require someone to generate and send the file manually before it can be ingested.

    * Use secure file transfer protocols such as SFTP or SCP.

        > SFTP (Secure File Transfer Protocol) securely transfers files over SSH.

        > SCP (Secure Copy) securely copies files between systems over SSH.

* Ways to ingest data from IoT devices and sensors:

    * Use a message queue or another streaming system to ingest the data.

        > IoT devices continuously produce events, making streaming systems the preferred ingestion method regardless of whether the data will later be processed in batch or as a stream.


* The appropriate ingestion method depends on the source system and business requirements.

* Understanding ingestion patterns and their trade-offs comes primarily through hands-on experience and real-world use cases.

* Batch Ingestion Tools

    * AWS Glue ETL

        > Serverless ETL service that ingests data from sources such as RDS, S3, Redshift, and DynamoDB, performs transformations using Apache Spark, and loads the data into a destination.

    * Amazon EMR

        > Managed platform for running big data frameworks such as Apache Spark and Hadoop on petabyte-scale data. Offers more control over compute resources than Glue but requires more configuration.

    * AWS DMS (Database Migration Service)

        > Synchronizes or migrates data between databases and AWS services (e.g., S3 or Redshift). Best suited for data migration and replication when transformations are not required.

    * AWS Snow Family

        > Physical transfer appliances (e.g., Snowball, Snowcone) used to migrate very large datasets to or from AWS when internet transfer is impractical.

    * AWS Transfer Family

        > Managed service for transferring files to and from Amazon S3 using protocols such as SFTP and FTP.

    * Other Ingestion Tools

        > Third-party connector platforms such as Airbyte, Matillion, and Fivetran simplify data ingestion between different sources and destinations.

* Streaming Ingestion Tools

    * Amazon Kinesis Data Streams

        > Managed AWS platform for real-time data ingestion and stream processing.

    * Amazon MSK (Managed Streaming for Apache Kafka)

        > Fully managed Apache Kafka service for building real-time event streaming pipelines.

* AWS Glue ETL vs Amazon EMR

    * AWS Glue ETL

        > Serverless, easier to configure, and ideal for most ETL workloads.

    * Amazon EMR

        > Greater control over compute and memory resources, suitable for large-scale Spark/Hadoop workloads, but requires more management.

* Choosing Batch vs Streaming Ingestion

    * Use Cases

        > Determine whether real-time data enables better business decisions than periodic batch updates.

    * Latency

        > Choose streaming for low-latency requirements; otherwise, batch or micro-batch is often sufficient.

    * Cost & Complexity

        > Streaming generally costs more and is more complex to build, maintain, and operate than batch.

    * Existing Systems

        > Ensure both source and destination systems can support the chosen ingestion pattern and expected data rate.

    * Reliability & Availability

        > Streaming systems require high availability and fault tolerance, whereas batch workloads are generally less demanding.

* Best Practices

    * Adopt true real-time streaming only when the business value justifies the additional cost and complexity.

    * IoT and sensor data are natural candidates for streaming ingestion using platforms such as Amazon Kinesis or Apache Kafka.

    * When both real-time and historical processing are required, architectures such as Lambda Architecture can support both streaming and batch ingestion.

    

# ETL vs ELT (Batch Ingestion Patterns)

* ETL = Extract → Transform → Load

    * Transform data before loading into storage.

    * Best when business requirements are well understood.

    > Popular when storage and compute resources were limited.

* ELT = Extract → Load → Transform

    * Load raw data first, then transform it inside the data warehouse.

    > Became practical with cheap cloud storage and powerful cloud data warehouses.

* ETL

    * Stores only transformed data.

    * More storage efficient.

    * Requires planning before ingestion.

    * If transformation logic changes, data often needs to be re-extracted and reloaded.

    * Can clean, validate, and mask sensitive data before storing it.

* ELT

    * Stores raw data.

    * Faster to implement.

    * Raw data is loaded faster since no transformations are performed beforehand.

    * Preserves all original data.

    * Allows transformations to be rerun without re-ingesting the data.

    * Provides flexibility for future analysis.

    * Supports structured, semi-structured, and unstructured data.

* ELT downside

    * Without proper data management, it can lead to a data swamp.

        > A data swamp is an unorganized collection of raw data that becomes difficult to manage or use.

    * Data quality and security transformations are applied after the data is loaded.

* Modern ETL vs ELT

    * The line between ETL and ELT has blurred because both modern ETL engines (Spark, Glue, Databricks, etc.) and cloud data warehouses provide distributed computing.

    * ETL transformations use the compute resources of the ETL engine.

    * ELT transformations use the compute resources of the data warehouse or lakehouse.

    * Neither ETL nor ELT performs heavy transformations on the OLTP database; both only extract data from it.

    * Today, the choice is less about processing power and more about architecture and business requirements.

* Choosing between ETL and ELT

    * Use ETL when:

        * Business requirements are stable and well defined.

        * You only want to store curated data.

        * You want to reduce storage costs.

        * Sensitive or unnecessary data should be removed before storage.

    * Use ELT when:

        * Requirements are likely to change.

        * Analysts need flexibility to create new transformations.

        * You want to preserve the raw data for future reprocessing.

        * Storage is relatively inexpensive compared to engineering effort.

        > ELT follows a "store first, decide later" philosophy, while ETL follows a "decide first, store later" philosophy.

* API ingestion often favors ELT.

    > APIs usually return semi-structured data (such as JSON), making future transformation requirements difficult to predict.

* Hybrid approach (EtLT)

    * Perform minimal transformations before loading (e.g., masking sensitive data or removing duplicates), then perform business transformations after loading.

* Cost comparison (Historically)

    * ETL was the preferred approach because storage and compute resources were expensive.

    * Data warehouses often cost millions of dollars and had limited processing power.

    * Since storing large amounts of raw data was impractical, organizations transformed data before loading to minimize storage and compute costs.

* Cost comparison (Today)

    * Cloud object storage (e.g., Amazon S3) is inexpensive.

    * Modern cloud data warehouses provide scalable compute that can be provisioned on demand.

    * As a result, storing raw data is often cheaper than the engineering effort required to carefully design ETL pipelines.

    * The cost difference between ETL and ELT now depends more on the chosen tools, cloud services, data volume, and compute usage than on the ingestion pattern itself.

* ETL vs ELT costs today

    * ETL may reduce storage costs because only transformed data is stored.

    * ELT generally increases storage costs because both raw and transformed data are often retained.

    * ELT can reduce development and maintenance costs by allowing transformations to be modified without re-ingesting data.

    * In modern cloud environments, the additional storage cost is often outweighed by the flexibility and reduced engineering effort provided by ELT.

# API (Application Programming Interface)

* Think of an API as a waiter in a restaurant. You look at the menu, tell the waiter what you want, and the waiter goes to the kitchen, delivers your order, and brings back the dish. You never enter the kitchen or need to know the recipe. Technically, an API is a defined interface that allows one software system to communicate with another without exposing its internal implementation.

* Web API

    * Like ordering delivery from another restaurant. Your waiter calls the other restaurant, places the order on your behalf, and brings the food to your table.

    * Your application communicates with another application's server over the internet to request or send data. This is the most common meaning of API in Data Engineering. Examples include the Spotify API, GitHub API, and Weather APIs.

* Library API

    * Like asking the bartender inside your own restaurant to make you a cocktail. You don't need to know the recipe — you just ask for it by name and it arrives at your table.

    * Your code calls functions provided by a library within the same application without needing to know how those functions are implemented. Examples include the Pandas API and NumPy API.

* Database API

    * Like asking the waiter to check the wine cellar. "Do you have a 2018 Merlot? Bring me two bottles and update the inventory." The waiter goes down, finds what you need, and brings it back.

    * Your application communicates with a database through a standard interface such as JDBC or ODBC to execute SQL queries and retrieve or modify data.

* Operating System API

    * Like asking the restaurant staff for building services — adjust the air conditioning, turn on the lights at your table, or unlock the restroom. These aren't food requests; they're requests to the building itself.

    * Your program requests services from the operating system, such as opening files, creating processes, accessing hardware, or allocating memory.

* Cloud API

    * Like hiring a catering company for your event. You don't build a new kitchen — you tell the catering service how many guests, what menu, and what equipment you need, and they provision everything for you.

    * Your application communicates with cloud providers such as AWS, Azure, or GCP to create and manage resources like virtual machines, storage, databases, and networking.

* AI API

    * Like asking the head chef to improvise a custom dish that isn't on the menu. You describe what you're in the mood for, and the chef uses their expertise and creativity to craft something and send it to your table.

    * Your application sends prompts to an AI service, which processes the request and returns generated text, images, code, or other AI outputs.

* Graphics API

    * Like giving plating instructions to the pastry chef. "Layer the chocolate here, fan the strawberries there, drizzle the sauce in this pattern." You describe the visual result, and the specialist renders it precisely.

    * Your application communicates with the graphics driver and GPU to render 2D and 3D graphics. Examples include OpenGL, Vulkan, and DirectX.

* Internal API

    * Like the kitchen's internal communication system — the expeditor calling orders to the grill station, the grill station signaling the plating station. Customers never see or interact with this; it's how the kitchen coordinates behind closed doors.

    * Internal services communicate with each other through APIs that are only available inside the organization and are not exposed to external users.

* Related API Terms

    * API

        * A contract that defines how software components exchange data by specifying inputs, outputs, protocols, and data formats, enabling communication without exposing internal implementation.

    * Web API

        * An API that operates over HTTP/HTTPS, accepting requests at URLs and returning responses (typically JSON or XML), allowing applications to communicate over the internet.

        * REST API

            * A web API design style where resources are identified by URLs and accessed using standard HTTP methods (GET, POST, PUT, DELETE). It provides a uniform, stateless, cacheable, and scalable interface for CRUD operations, typically exchanging data in formats such as JSON or XML.
            
            > Similar to browsing the web, a client sends an HTTP request for a resource, and the server responds with the requested data. APIs also commonly provide metadata, documentation, authentication, and error handling to simplify integration.

        * GraphQL API

            * A query language and runtime where the client specifies exactly which fields and nested relations to return, eliminating over-fetching and under-fetching.

        * gRPC API

            * A Remote Procedure Call (RPC) framework using HTTP/2 and Protocol Buffers with shared `.proto` schemas, providing high-throughput, low-latency communication with strongly typed contracts and bi-directional streaming.

    * SDK (Software Development Kit)

        * A package containing libraries, documentation, code samples, and tools that wrap raw API calls into language-native functions, simplifying authentication, requests, and error handling.

    * Endpoint

        * A specific URL combined with an HTTP method that maps to a server operation and routes incoming requests to the appropriate handler.

    * Request

        * An HTTP message sent from a client containing a method, URL, headers, and optionally a body, specifying the operation and any required data.

    * Response

        * An HTTP message returned by the server containing a status code, headers, and optionally a body, providing the requested data or error information.

    * Client

        * Any software that initiates API calls, such as a browser, mobile app, CLI tool, or another server, by sending requests and consuming responses.

    * Server

        * The software that listens for API requests, executes business logic, and returns responses to clients.

* FastAPI

    * A modern Python Web Framework built on Starlette (ASGI) and Pydantic (data validation). It uses Python type hints to automatically validate requests, serialize responses, and generate OpenAPI documentation.

    * Framework Hierarchy

        * Software Framework

            * Web Frameworks – Handle HTTP requests, routing, middleware, validation, and responses.

                * FastAPI – Python, ASGI, async-first, automatic OpenAPI documentation.

                * Flask – Python, WSGI, lightweight, extensible through plugins.

                * Django – Python, WSGI, full-stack framework with ORM, authentication, admin panel, and migrations.

                * Express.js – JavaScript, middleware-based framework for Node.js.

                * Spring Boot – Java framework using dependency injection and convention over configuration.

                * ASP.NET Core – Cross-platform C# web framework with built-in dependency injection.

            * GUI Frameworks – Build graphical user interfaces and handle user interactions.

            * ML Frameworks – Define, train, and deploy machine learning models.

            * Testing Frameworks – Write, execute, and report automated tests.

    * Other Python Web Frameworks

        * FastAPI – ASGI framework using async/await, Pydantic validation, and automatic OpenAPI generation. Primarily used for high-performance asynchronous APIs.

        * Flask – Lightweight WSGI micro-framework commonly used for small web applications and APIs.

        * Django – Full-stack WSGI framework with built-in ORM, authentication, admin panel, templates, and migrations.

        * Tornado – Asynchronous networking framework designed for long-lived connections such as WebSockets.

        * Falcon – Lightweight WSGI/ASGI framework optimized for high-performance REST APIs.
