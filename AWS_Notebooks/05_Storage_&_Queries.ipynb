# Data Storage

- `Storage` is one of the most complicated components of the `data engineering lifecycle` because data is stored multiple times as it moves through the lifecycle.

- Storage decisions affect:
  - `Cost`
  - `Performance`
  - `End-user experience`

- Data storage spans the lifecycle from:
  - `Source systems`
  - `Ingestion`
  - `Transformation`
  - `Serving`

  > As a data engineer, you usually have more direct control over storage from the ingestion stage onward, while source systems may be outside your control.

- Choosing the right storage solution requires understanding the characteristics of the data and how it will be used:
  - `Type`
  - `Format`
  - `Size`
  - How the data will be `accessed`
  - How the data will be `updated`
  - Who will access or update it
  - When it will be accessed or updated

- `Storage hierarchy:` Storage can be viewed as a hierarchy, starting from the physical components and building up into higher-level storage abstractions.

  - `Raw ingredients:` The physical components and processes that make storage possible.

    - Physical storage components include:
      - `Magnetic disks`
      - `Solid-state drives (SSDs)`
      - `Memory`

    - Processes used to store and transmit data include:
      - `Networking`
      - `Serialization`
      - `Compression`

    > Data engineers generally don't interact directly with these physical components. Instead, they interact with storage systems built on top of them.

  - `Storage systems:` Systems that internally manage and organize stored data and provide ways to interact with it.

    - Examples:
      - `Databases`
      - `Object storage`

  - `Storage abstractions:` Higher-level storage solutions built using underlying storage systems.

    - Examples:
      - `Cloud data warehouses`
      - `Data lakes`
      - `Data lakehouses`

- `OLTP (Online Transaction Processing):`
  - Designed for processing transactions.
  - Focuses on `read` and `write` operations with `low latency`.
  - Commonly deals with rapidly changing application state.

- `OLAP (Online Analytical Processing):`
  - Designed for analytical workloads.
  - Focuses on operations such as:
    - `Aggregation`
    - `Summarization`
  - Used to analyze data and support `business decisions`.

  > OLTP and OLAP have different storage requirements because they optimize for different workloads. OLTP prioritizes fast transactional operations, while OLAP prioritizes efficient analytical processing.

- Modern storage systems also include specialized databases for analytical and machine learning use cases:
  - `Graph databases`
  - `Vector databases`

  - `Graph databases` are designed around relationships between entities.
  - `Vector databases` support storing and searching vector representations, making them useful for machine learning and AI workloads.

- `Row-oriented vs column-oriented storage:`
  - Different physical data layouts have different performance characteristics.
  - The appropriate layout depends on the workload and whether the system is primarily supporting `OLTP` or `OLAP`.

  > Storage should be chosen based on workload characteristics rather than simply choosing the storage technology that is most popular or familiar.

- `Distributed storage systems` allow data to be stored across multiple machines rather than relying on a single machine.

- Cloud storage can be categorized into:
  - `Block storage`
  - `Object storage`
  - `File storage`

- `Storage trade-offs:` Storage decisions involve balancing:
  - `Cost`
  - `Performance`

  > There is no single storage solution that is optimal for every workload. The correct choice depends on the characteristics of the data and how stakeholders need to access, update, and process it.

- A storage decision should therefore consider both:
  - The characteristics of the `data`.
  - The characteristics of the `workload`.

- The overall goal of storage design is to choose solutions that provide the required performance and functionality while keeping storage costs appropriate for the workload.

# Physical Storage Technologies

* `Raw ingredients` are the physical components used to store data and form the foundation of all data storage systems.

* As a data engineer, you need to understand the:

  * `Characteristics`
  * `Performance`
  * `Data durability`
  * `Cost`

  of different storage technologies in order to select the appropriate storage system for a particular use case.

* As data moves through a data pipeline, it frequently passes through two main types of storage:

  * `Persistent storage`

    * Magnetic disks
    * SSDs
  * `Volatile memory`

    * RAM

    * CPU cache

  > Persistent storage retains data over time, while volatile memory can lose its contents when power is lost.

* `Magnetic disks (HDDs):`

  * Use rotating platters coated with a magnetic film to store data.

  * They work mechanically, similar to an old record player.

  * Each platter contains circular `tracks`.

  * Tracks are divided into smaller storage units called `sectors`.

  * The combination of a track number and sector number creates a unique address used to locate data.

  * `Writing data:`

    * A read/write head moves to the appropriate location.
    * The head magnetizes the film to physically encode binary data.
    * One magnetic direction represents a `1`.
    * The opposite magnetic direction represents a `0`.

  * `Reading data:`

    * The read/write head detects the magnetic field at the specified address.
    * The detected information is returned as a `bit stream`.

* `SSD (Solid-State Drive):`

  * Stores data as electrical charges in `flash memory cells`.

  * A charged cell represents a `1`.

  * An uncharged cell represents a `0`.

  * SSDs have no mechanical moving parts.

  * Data is read and written electronically.

  > Removing mechanical movement makes SSDs significantly faster than magnetic disks, especially for random access.

* `Magnetic disk performance:`

  * Data-fetch latency depends mainly on two mechanical operations:

    * `Seek time:` The time required for the read/write head to physically locate the correct track.
    * `Rotational latency:` The time required for the correct sector to rotate underneath the read/write head.

  * Both operations have physical limitations because they depend on mechanical movement.

  * A typical commercial HDD at the time of the recording:

    * Rotates at around `7,200 RPM`.
    * Has an average data-fetch latency of about `4 milliseconds`.
    * Can perform at most several hundred `IOPS` (Input/Output Operations Per Second).

* `SSD performance:`

  * Newer SSDs can typically:

    * Perform up to `tens of thousands of IOPS`.
    * Achieve data-fetch latency of about `0.1 milliseconds`.

  * This makes SSDs particularly effective for `random access`.

  > Random access means being able to quickly read or update data from any location rather than primarily processing large sequential blocks.

* `Data transfer speed:`

  * Magnetic disks can read and write up to approximately `300 MB/s` between disk and RAM.
  * SSDs can be `more than 10 times faster` in data transfer speed.

* `Distributed storage and parallel processing:`

  * Read/write performance can be increased beyond the limitations of a single storage device by processing data in parallel.

  * One approach is to:

    * Distribute data across many magnetic disks and machines in a cluster.

    * Read from multiple disks simultaneously.

    > When many disks are accessed in parallel, overall transfer speed becomes increasingly limited by `network performance` rather than the physical speed of an individual disk.

  * SSDs can also be scaled through parallelism:

    * Storage can be divided into multiple partitions.
    * Multiple storage controllers can operate in parallel.
    * More data transactions can therefore be handled simultaneously.

  * With parallel processing, SSD-based systems can achieve transfer speeds of `many gigabytes per second`.

* `Cost comparison:`

  * Commercial magnetic disks are typically around `2–3 times cheaper` than SSDs for the same amount of storage.

  * HDDs are therefore still widely used as the backbone of bulk data storage despite:

    * Higher latency.
    * Slower transfer speeds.

* `When to choose magnetic disks:`

  * Magnetic disks are a cost-efficient choice when:

    * Data is accessed `infrequently`.

    * Data is accessed in relatively large blocks, typically `1 MB or more` at a time.

    * The application does not require extremely fast read/write speeds.

  > HDDs are a good choice when minimizing storage cost is more important than achieving very low latency.

* `When to choose SSDs:`

  * SSDs are commonly used in commercial `OLTP` deployments.

  * Their high IOPS and low latency allow relational databases to handle `thousands of transactions per second`.

  > SSDs are not always the best choice for analytical storage because their higher cost can make them less economical for storing large amounts of analytical data.

* `RAM (Random Access Memory):`

  * Before the CPU can process data stored on persistent storage, the data generally needs to be transferred from disk storage, such as:

    * HDDs
    * SSDs

    to `RAM`.

  * RAM is typically attached closely to the CPU.

  * This allows RAM to provide much faster access than disk storage.

  * Its speed is better suited to the processing speed of the CPU.

  * At the time of the recording, RAM provided approximately:

    * `100 GB/s` data transfer speed.

    * `0.1 microseconds` data-fetch latency.

    * `Millions of IOPS`.

  > Exact RAM performance varies significantly depending on the hardware and system configuration, and these numbers can change as hardware technology improves.

* `RAM cost and durability:`

  * RAM is much more expensive than persistent storage because it is closely integrated with the CPU.

  * At the time of the recording, RAM was approximately `30–50 times more expensive than SSDs` per unit of stored data.

  * RAM is `volatile`.

    * If power is lost, data stored in RAM can disappear in less than a second.

  > RAM should therefore generally not be relied upon as the only place to store data that must persist over time.

* `Common uses of RAM:`

  * Storing code that the CPU executes.

  * Storing data that the currently executing code needs to process.

  * `Caching`

  * `Data processing`

  * `Indexing`

  * Some databases treat RAM as a `primary storage layer` because of its extremely fast read/write performance.

  > Even when RAM-based data is replicated across multiple machines, volatility remains important. A power outage affecting several nodes at once can still result in data loss.

* `CPU cache:`

  * CPU cache is another type of volatile memory.

  * It is located directly on the CPU processing chip.

  * It is even faster than RAM because of its physical proximity to the CPU.

  * CPU cache is used to store `frequently accessed data` so it can be retrieved extremely quickly during processing.

  * At the time of the recording, CPU cache provided approximately:

    * `1 nanosecond` data-fetch latency.
    * `1 TB/s` data transfer speed.

* `Caching:`

  * The concept of caching is not limited to CPU cache.

  * A `cache` is generally a fast-access storage layer used to store frequently or recently accessed data.

  * Examples:

    * `CPU cache`

      * Stores frequently needed data close to the CPU.
    * `Browser cache`

      * Stores downloaded web resources so pages can load faster.
    * `Database cache`

      * Stores results of frequently executed queries so they can be returned faster.

  > The main idea behind caching is to keep frequently needed data in a faster storage layer, reducing the need to repeatedly access slower storage.

* `General storage trade-off:`

  * Storage technologies differ in:

    * `Latency`
    * `IOPS`
    * `Transfer speed`
    * `Cost`
    * `Durability / volatility`

  * The general relationship is:

    * `HDD` → slower, cheaper, persistent

    * `SSD` → faster, more expensive, persistent

    * `RAM` → much faster, much more expensive, volatile

    * `CPU cache` → extremely fast, extremely expensive, volatile

  > There is a fundamental trade-off between `performance and cost`. Faster storage is generally more expensive, so the correct technology depends on the performance requirements and access patterns of the workload.

* As a data engineer, understanding these performance and cost trade-offs helps you evaluate storage technologies and select solutions that satisfy the requirements of your data processing workloads.


# Networking, Serialization, and Compression

* The bottom layer of the storage hierarchy includes not only the physical storage components such as `HDDs`, `SSDs`, and `RAM`, but also the components and processes required to store and transmit data.

* Important components and processes at this layer include:

  * `Networking`
  * `CPU processing`
  * `Serialization`
  * `Compression`

* `Distributed storage:`

  * Modern cloud storage systems are increasingly `distributed`.

  * Data can be:

    * Split across multiple servers.
    * Replicated across multiple servers.
    * Spread across connected servers.

  * Distributed storage can improve:

    * `Read performance`
    * `Write performance`
    * `Data durability`
    * `Availability`

  * Networking and CPU processing are therefore part of the raw ingredients of modern storage systems.

  * The CPU may be responsible for tasks such as:

    * Aggregating read results from multiple servers.

    * Distributing writes across multiple servers.

    * Handling the details of servicing read/write requests.

  > In a distributed storage system, storage performance is not determined only by the physical disks. Network performance and the CPU resources required to coordinate operations across servers can also become important factors.

* `Serialization:` The process of converting data from its in-memory representation into a standard format that can be stored or transmitted.

  * Data in memory is stored in data structures optimized for:

    * Efficient CPU access.
    * Efficient CPU manipulation.

  * This in-memory representation is not suitable for:

    * Persistent storage on disk.
    * Transmission over a network.

  * Serialization converts the data into a standard representation, usually a `sequence of bytes`, that can be:

    * Efficiently stored.
    * Transmitted over a network.
    * Shared between different systems.

* `Deserialization:` The process of reconstructing the original data structures from the serialized representation.

  > Serialization converts data structures into a transferable/storable representation, while deserialization converts that representation back into usable data structures.

* Data can be serialized using either a:

  * `Row-based approach`
  * `Column-based approach`

* `Row-based serialization:`

  * Data is encoded and stored `record by record`.

  * A consecutive sequence of bytes represents one complete row.

  * For semi-structured data:

    * Data is encoded object by object or document by document.
    * The data representing one complete object/document is stored as a consecutive sequence of bytes.

  * This is well suited to `transactional operations` where an entire row or object needs to be accessed.

* `Column-based serialization:`

  * Data is encoded and stored `column by column`.

  * A consecutive sequence of bytes represents one column.

  * For column-oriented semi-structured data:

    * Values for the same key across multiple objects are stored together as a consecutive sequence of bytes.

  * This is well suited to `analytical queries` that operate on specific columns.

  > Row-based storage is generally advantageous when accessing complete records, while column-based storage is advantageous when analytical queries need only a subset of columns.

* Common data serialization formats include:

  * `CSV`
  * `XML`
  * `JSON`
  * `Avro`
  * `Parquet`

* `CSV:`

  * A human-readable text format.

  * Commonly used for exchanging tabular data.

  * It is a `row-based` format.

  * A major limitation is that CSV does not have a defined schema.

  * The application therefore has to determine the meaning of each row and column.

  * Schema changes can be problematic:

    * If an application adds a new row or column, the change may need to be handled manually.

  > CSV is convenient and widely supported, but it is generally not a good choice for data pipelines when more robust formats are available because it lacks a defined schema.

* `XML (Extensible Markup Language):`

  * A human-readable text format.
  * Was widely used when the web and HTML were developing.
  * It is now generally considered a `legacy format` for modern data engineering applications.
  * Serialization and deserialization can be relatively slow.
  * It has largely been replaced by `JSON` for plain-text object serialization.

* `JSON:`

  * A human-readable text format.
  * Widely used for exchanging data between systems through `APIs`.
  * Also commonly used for data storage.
  * Has become a common standard for API data exchange.

* `Parquet:`

  * A `binary` serialization format.
  * Uses a `column-based` layout.
  * Designed for:

    * Efficient storage.
    * Big data processing.
    * Efficient analytical queries.

* `Avro:`

  * A `binary` serialization format.

  * Uses a `row-based` layout.

  * Uses a `schema` to define the structure of the data.

  * Supports `schema evolution`.

  > Schema evolution allows the structure of data to change over time while providing mechanisms for handling those changes.

* The choice of serialization format can have a significant impact on:

  * Storage efficiency.

  * Query performance.

  * Data processing performance.

  > Serialization is not just about making data readable or transferable. The way data is physically organized can directly affect how efficiently data can be stored and queried.

* A real-world example showed that simply changing a serialization format from `CSV` to `Parquet` improved job performance by a factor of `100`.

  > Choosing an appropriate serialization format can therefore produce very large performance improvements without changing the underlying processing logic.

* `Compression:` The process of reducing the number of bits required to represent data.

* Compression is particularly important as datasets become increasingly large because it can:

  * Reduce storage requirements.
  * Reduce the amount of data transferred.
  * Improve query performance.

* Compression algorithms work by identifying `redundancy` and `repetition` in data and encoding the information more efficiently.

* For text-based formats such as:

  * `CSV`
  * `JSON`
  * `XML`

  traditional compression algorithms can identify characters that occur frequently and encode them differently from less frequent characters.

  * Common characters can be represented using shorter bit sequences.

  * Less common characters can use longer sequences.

  * The overall file therefore requires fewer bits.

  > Compression takes advantage of patterns and redundancy in data so that the same information can be represented using less storage space.

* `Compression ratio:` The ratio of the compressed file size to the original uncompressed file size.

  * For example, if a file is reduced from `100 MB` to `20 MB`:

    * Compression ratio = `20 / 100 = 0.2`

* Compression reduces disk space, but it can also improve `query performance`.

  * When querying data, data often needs to be loaded from disk into memory.

  * This creates `I/O` operations.

  * A smaller compressed dataset requires less data to be read from disk.

  * Therefore, compression can reduce the I/O time required to load the necessary data.

  > Compression can improve performance even though the system has to perform decompression, because reducing the amount of data read from storage can save more time than the CPU work required for decompression.

* Modern compression algorithms increasingly prioritize:

  * `Compression/decompression speed`
  * `CPU efficiency`

  rather than maximizing the compression ratio.

* These fast compression algorithms are frequently used in:

  * `Data lakes`

  * `Columnar databases`

  > For analytical workloads, a slightly larger compressed file can be preferable if it can be decompressed much faster and with less CPU overhead.

* The main trade-offs when choosing storage-related formats and processes include:

  * `Storage size`
  * `I/O performance`
  * `CPU usage`
  * `Network transfer`
  * `Query performance`

* The raw storage layer therefore consists of more than physical storage devices. Modern storage systems depend on:

  * Physical storage such as `HDDs`, `SSDs`, and memory.

  * `Networking` to communicate between storage components.

  * `CPU processing` to coordinate storage operations.

  * `Serialization` to represent data in a format suitable for storage and transmission.

  * `Compression` to reduce storage and I/O requirements.

  > Understanding these components is important because decisions made at this low level can affect the performance, cost, and scalability of the entire data system.


# Cloud Storage Systems

* Cloud storage systems exist at a higher level of abstraction than the raw storage ingredients such as `HDDs`, `SSDs`, `RAM`, networking, serialization, and compression.

* The three common types of cloud storage are:

  * `File storage`
  * `Block storage`
  * `Object storage`

* Choosing between them requires considering:

  * `Performance`
  * `Scalability`
  * How frequently data is accessed.
  * How frequently data is modified.
  * The type of workload being supported.

* `File storage:`

  * Organizes files into a hierarchical `directory tree`, similar to folders on a laptop.

  * Each folder contains metadata about its files and subfolders, such as:

    * File names.
    * Owners.
    * Last modification dates.
    * Access permissions.
    * Location pointers to the actual files and subfolders.

  * Files are located using a `path`.

  * The operating system follows the hierarchy from the root directory through the relevant subdirectories until it reaches the requested file.

  * File storage is useful when you need:

    * Centralized access to files.
    * Files to be easily shared between multiple users.
    * Files to be accessed by multiple host machines.
    * Simple file management.

  * Managed cloud file storage services, such as `Amazon Elastic File System (EFS)`, provide shared network access to files without requiring you to manage:

    * Networking.
    * Disk cluster scaling.
    * Underlying configuration.

  * File storage is often built on top of `block storage`, with the complexity of the underlying storage mechanisms abstracted away.

  * The main disadvantage is that file storage does not provide the highest read/write performance.

    > File storage has to maintain and manage the file hierarchy, which introduces overhead compared with lower-level block storage.

* `Block storage:`

  * Divides files into small, fixed-size `blocks`.

  * Blocks can be stored on:

    * Magnetic disks.
    * SSDs.

  * Each block has a unique identifier or address.

  * The identifier allows individual blocks to be efficiently:

    * Located.
    * Retrieved.
    * Modified.

  * Block storage provides:

    * Higher performance than file storage.
    * Lower latency than file storage.
    * Precise allocation of storage space.
    * Efficient modification of individual pieces of data.

  * A file stored using block storage is divided across multiple blocks.

  * The storage application:

    * Writes the file's data into multiple blocks.
    * Records the identifiers of those blocks in a lookup table.

  * When the file is requested:

    * The application finds the relevant blocks.

    * Retrieves their data.

    * Combines the blocks back into the original sequence.

  > The application hides this complexity, allowing you to work with the file while the underlying system manages the individual blocks.

* Block storage is often implemented using a `distributed architecture`.

  * Blocks can be spread across multiple storage disks.
  * This provides:

    * Higher scalability.
    * Greater data durability.

* Block storage forms the foundation of many modern storage solutions, including:

  * Local file systems.
  * Transactional databases.
  * Virtual machine storage.

* Block storage is ideal when data is:

  * Accessed frequently.
  * Modified frequently.
  * Required to have low-latency access.

* `OLTP` systems commonly access storage at the block level.

  * This allows OLTP systems to perform:

    * Small reads.
    * Small writes.
    * Frequent operations.
    * Random access.
    * Low-latency transactions.

* Block storage is also commonly used as persistent storage for `virtual machines`.

  * For example, `Amazon EC2` instances use block storage for their persistent storage.

  * An EC2 instance automatically has a root storage device backed by a block storage volume.

  * The block storage volume can contain:

    * The operating system.
    * The file system.
    * Other computing resources.

  * The default AWS block storage service for EC2 is `Amazon Elastic Block Store (EBS)`.

  * EBS provides different volume types for different workloads:

    * High-performance `SSD` volumes for latency-sensitive workloads.
    * Cost-efficient `magnetic disk` volumes for infrequently accessed data.

* A limitation of block storage is that its scalability is usually connected to the compute resources to which the storage volumes are attached.

  * Because block storage volumes are typically attached to compute instances:

    * Storage scalability is constrained by compute scaling.
    * Block storage typically scales to only a few `terabytes`.

* `Object storage:`

  * Separates or `decouples` the storage layer from the compute layer.

  * This allows storage to scale independently of compute.

  * It can scale to `petabytes` or more.

  > With cloud object storage, storage capacity is generally limited more by your budget than by the physical storage capacity available to you.

* Because object storage is independent of compute:

  * You can use `ephemeral compute clusters`.

  * Clusters can be created when needed.

  * Clusters can be scaled up or down based on workload requirements.

  * The underlying infrastructure is managed by the cloud provider.

  > Separating storage from compute makes large-scale data processing accessible without requiring an organization to permanently own expensive hardware for workloads that may only run occasionally.

* Unlike traditional file storage, object storage does not use a hierarchical directory structure.

* Object storage stores data as `immutable objects` in a `flat structure`.

  * Objects are organized into top-level logical containers.

  * For example, an AWS `S3 bucket` acts as a container for objects.

  * Each object has a unique identifier called a `key`.

  * An S3 object identifier can conceptually contain:

    * The bucket name.
    * The object's key.

  * The bucket name must be unique across AWS.

* `Object immutability:`

  * Once an object is written, it is treated as immutable.

  * You cannot modify only a small part of an existing object.

  * For example:

    * If you have a `1 GB` object.

    * You want to change only one character.

    * You must rewrite the entire object.

  > This is very different from block storage, where a small portion of a file can be modified by updating only the relevant block.

* Object immutability is also an advantage for distributed storage.

  * Since objects are not modified in place:

    * The system does not need to synchronize small changes across many storage nodes.
    * Objects can be distributed across many nodes.
    * Each node can contain its own disks.

  * This reduces the overhead associated with propagating changes across nodes.

* Object storage can therefore:

  * Scale horizontally.
  * Perform highly parallel reads and writes.
  * Distribute data across many disks.

* Object storage systems divide data into `shards` that can be distributed across storage nodes.

  * Shards are replicated across multiple nodes.

  * Replication improves `data durability`.

  > Horizontal scaling and replication allow object storage to handle extremely large datasets while maintaining high availability and durability.

* Object storage is ideal for:

  * `Cloud data warehouses`
  * `Data lakes`
  * Massive analytical systems.
  * Large machine learning datasets.

* Object storage is particularly useful for `OLAP` workloads because these workloads are generally:

  * Read-heavy.
  * Analytical.
  * Performed over very large datasets.
  * Suitable for parallel processing.

* Object storage is not well suited to transactional workloads because objects are immutable.

  * Workloads requiring:

    * Many small updates.
    * Frequent modifications.
    * Very low-latency writes.

    are better suited to `block storage`.

* Object storage is also important in `machine learning pipelines`.

  * ML pipelines often require large amounts of unstructured training data, such as:

    * Raw text.
    * Images.
    * Videos.
    * Audio.

* `File vs block vs object storage:`

  * `File storage`

    * Hierarchical directory structure.
    * Simple to understand and manage.
    * Good for centralized file sharing.
    * Can be accessed by multiple users or machines.
    * Lower performance than block storage.
    * Best when performance and scalability requirements are relatively low.

  * `Block storage`

    * Data is divided into fixed-size blocks.
    * Individual blocks can be accessed and modified.
    * High performance and low latency.
    * Good for frequent reads and writes.
    * Commonly used by transactional databases and virtual machines.
    * Scalability is generally tied to compute resources.
    * Typically scales to a few terabytes.

  * `Object storage`

    * Stores immutable objects in a flat structure.
    * Objects are identified using unique keys.
    * Storage is decoupled from compute.
    * Extremely scalable.
    * Supports massive parallel processing.
    * Highly durable through replication.
    * Good for massive analytical datasets and ML data.
    * Poor fit for frequent small updates.

* A useful way to choose between the three:

  * Choose `file storage` when:

    * You primarily need simple file sharing.
    * Multiple users or machines need centralized access.
    * Performance and scalability requirements are relatively low.

  * Choose `block storage` when:

    * You need frequent reads and writes.
    * You need low-latency access.
    * Your workload is transactional.
    * You need to modify individual portions of data efficiently.
    * You are supporting workloads such as `OLTP` or virtual machines.

  * Choose `object storage` when:

    * You need to store massive amounts of data.

    * You need very high scalability.

    * Storage should scale independently from compute.

    * You are performing large-scale analytical processing.

    * You need highly parallel reads and writes.

    * You are storing large unstructured datasets for machine learning.

  > The key distinction is that `file storage` prioritizes simple shared access, `block storage` prioritizes low-latency and frequent modifications, while `object storage` prioritizes massive scalability, durability, and parallel analytical processing.

* Cloud providers typically offer different `storage tiers` within these storage types.

  > Storage tiers allow you to choose different cost and performance characteristics based on how frequently the data is accessed, such as `hot`, `warm`, and `cold` storage.


# Storage Tiers

* Cloud storage services commonly provide different `storage tiers` that allow you to balance:

  * `Cost`
  * `Access speed`
  * `Access frequency`
  * `Compliance requirements`

* A useful way to classify data is based on how frequently it is accessed:

  * `Hot data`
  * `Warm data`
  * `Cold data`

* `Hot data:`

  * Data that is accessed `frequently`.

  * Requires `fast retrieval times`.

  * Should be stored in high-performance storage systems using technologies such as:

    * `SSDs`
    * `Memory`

  * Example:

    * A product recommendation application may frequently access:

      * Product catalog.
      * User purchase history.
    * Frequently executed query results can also be stored in a `cache` to provide recommendations quickly.

  * Characteristics:

    * Highest storage cost.
    * Fastest retrieval.
    * Relatively low compute requirements for accessing the data.
    * Storage is optimized for fast access.

* `Warm data:`

  * Data accessed `less frequently than hot data`.

  * Still needs to be readily available when required.

  * Example:

    * Data used for regular reports and analysis that does not need to be updated in real time.

  * Warm data is typically stored using:

    * Lower-cost magnetic disks.
    * Hybrid storage systems.

  * Compared with hot data:

    * Lower storage cost.
    * Slower retrieval.
    * Greater compute resources may be required to retrieve the data.

* `Cold data:`

  * Data that is `rarely accessed`.

  * Often kept for `archival` purposes.

  * Examples:

    * Old project documentation.
    * Historical emails retained for compliance.

  * Cold data should be stored in the most cost-effective storage tier, typically using low-cost magnetic disks.

  * Characteristics:

    * Lowest storage cost.
    * Slowest retrieval compared with hot and warm storage.
    * Higher compute requirements when the data needs to be accessed.

* `Storage tier trade-off:`

  * As storage performance and access speed increase, `storage cost` generally increases.

  * As storage performance and access speed decrease, `storage cost` generally decreases.

  > There is a direct trade-off between storage cost and retrieval performance.

* Storing all data as `hot data`:

  * Provides very fast access.
  * Results in very high storage costs.

* Storing all data as `cold data`:

  * Minimizes storage costs.
  * Results in:

    * Longer retrieval times.

    * Higher compute requirements when data needs to be accessed.

  > The goal is usually not to put everything into one tier. A better approach is to use a `combination of storage tiers` based on the access patterns and requirements of different datasets.

* `Example using Amazon S3:`

  * Frequently accessed `hot data` used for real-time transactions can be stored in:

    * `S3 Express One Zone`
    * `S3 Standard`

  * `Warm data` accessed weekly or monthly, such as data used for fine-tuning a product recommendation system, can be stored in:

    * `S3 Standard-IA`
    * `S3 One Zone-IA`

  * Historical `cold data` can be archived using:

    * `S3 Glacier` tiers.

* Storage tier selection should not be based only on `access frequency`.

  * You should also consider:

    * `Scalability`

    * `Durability`

    * Cost.

    * Retrieval performance.

    * Compute requirements.

    * Compliance requirements.

  > The appropriate storage tier is determined by the overall requirements of the workload, not simply by whether data is hot, warm, or cold.


# Distributed Storage

* As data storage requirements grow and access patterns become more complex, a single machine will eventually become insufficient, requiring data to be distributed across multiple servers.

* `Distributed storage` is the default approach for storing data in the cloud, including:

  * `Block storage`
  * `File storage`
  * `Object storage`

* A distributed storage system distributes and replicates data across multiple servers called `nodes`.

  * Nodes are connected through a `network`.
  * A group of nodes forms a `cluster`.
  * Multiple clusters can collectively form a distributed storage system.

* Each node typically contains:

  * Physical storage such as `magnetic disks` or `SSDs`.
  * Processing capabilities for tasks such as:

    * Data management.
    * Replication.
    * Access control.

* The total storage capacity of a distributed system is generally based on the storage capacity available across its nodes, although replication means some of that capacity is used for redundant copies rather than unique data.

* `Horizontal scaling:`

  * Add more nodes to existing clusters as data volumes and workloads increase.
  * Allows storage capacity and processing capacity to grow by adding machines.

* `Vertical scaling:`

  * Upgrade the resources of a single machine.

  * In a single-machine architecture, scaling storage mainly means increasing the storage capacity of that server.

  > Distributed systems make horizontal scaling possible, while a single-machine architecture is primarily limited to vertical scaling.

* `Fault tolerance and durability:`

  * Distributing and replicating data across multiple nodes provides greater protection against failures.

  * Data can remain available even when one or more components fail.

  * Failures can include:

    * Hardware failures.
    * Software failures.
    * Network outages.
    * Other system disruptions.

* `High availability:`

  * If one node becomes unavailable, data can still be accessed from another available node containing the required data.

  * Replicas may even be located in different geographical locations.

  > Fault tolerance and replication help maintain availability because the system does not depend on a single machine remaining operational.

* `Distributed performance:`

  * Large processing tasks can be divided into smaller subtasks.

  * Different nodes process these subtasks in parallel.

  * This allows many `read` and `write` operations to happen simultaneously.

  * Replication can also improve read performance:

    * A read request can be served by a nearby replica.
    * The system can choose a less-congested replica.
    * This can reduce access time.

* Distributed storage architecture is used by many modern systems, including:

  * `Object storage`
  * `Cloud data warehouses`
  * `Hadoop Distributed File System (HDFS)`
  * `Apache Spark`

* There are two fundamental ways to distribute data across nodes:

  * `Replication`
  * `Partitioning`

* `Replication:`

  * Keeps copies of the same data on multiple nodes.

  * Replicas can potentially be located in different geographical locations.

  * Replication provides:

    * `Redundancy`
    * Higher `availability`
    * Greater `fault tolerance`
    * Potentially better read performance

* `Partitioning / Sharding:`

  * Splits a large dataset into smaller subsets called:

    * `Partitions`
    * `Shards`

  * Different partitions can then be assigned to different nodes.

  > Partitioning distributes different pieces of the dataset, while replication creates multiple copies of those pieces.

* In practice, distributed systems commonly combine both approaches:

  * First, partition a large dataset into smaller shards.
  * Distribute the shards across different nodes.
  * Replicate the nodes or partitions to provide redundancy.

* Many databases can automatically handle:

  * Partitioning.

  * Replication.

  * The database may abstract these details away from the user.

  * Alternatively, you may be able to configure replication and partitioning parameters for more control.

* `Replication challenge:`

  * Changes do not necessarily reach every replica instantly.

  * Replicating an update across multiple nodes takes time.

  * If you try to read from a node while it is being updated, the system may:

    * Wait until the update is available, providing more up-to-date data.

    * Return the data currently available on that node, which may be older.

  > This creates a fundamental trade-off between getting the latest data and making data immediately available.

* `CAP theorem:`

  * Describes a fundamental trade-off in distributed systems involving:

    * `Consistency`
    * `Availability`
    * `Partition tolerance`

  * The simplified statement is that a distributed system cannot guarantee all three properties simultaneously when a network partition occurs.

* `Consistency:`

  * Strong consistency means that every read reflects the latest successful write.

  > If a write has been committed, a subsequent read is guaranteed to see that latest value under strong consistency.

* `Availability:`

  * Every request receives a response, even if the response may not contain the most recent data.

* `Partition tolerance:`

  * The system continues operating despite network failures that prevent some nodes from communicating with others.

  > A network partition means that parts of a distributed system become unable to communicate even though the individual nodes may still be running.

* In real distributed systems, `partition tolerance` is generally considered necessary because network failures and communication disruptions cannot simply be assumed never to happen.

  > Therefore, when a partition occurs, the important trade-off is generally between maintaining `consistency` and maintaining `availability`.

* During a network partition, a system may choose to:

  * Reject or delay a request until the system can guarantee the latest data.

    * Improves `consistency`.
    * Reduces `availability`.

  * Return the data currently available from a replica.

    * Improves `availability`.

    * May return stale data and therefore temporarily reduce `consistency`.

  > CAP is specifically about distributed systems under network partition. It does not mean that a system permanently has to choose only two properties under normal operation.

* `ACID consistency vs distributed consistency:`

  * `ACID consistency` means that a transaction must preserve the rules and constraints defined by the database.

  * The database must move from one valid state to another valid state.

  * `Distributed consistency` concerns whether different nodes or replicas observe the same/latest data.

  > These are different concepts. ACID consistency is about preserving database correctness during transactions, while distributed consistency is about how data is observed across nodes.

* A system can use strong distributed consistency to help ensure that replicas reflect committed changes, but this does not make the terms `ACID consistency` and `distributed consistency` interchangeable.

* `BASE:`

  * BASE is a set of principles commonly associated with distributed systems that favor availability and eventual consistency.

  * `Basically Available:`

    * The system remains available and can continue responding to requests even when the data may not be perfectly up to date.

  * `Soft State:`

    * The state of the system can change over time, including as updates propagate between nodes.
    * A read may temporarily observe a state that is not yet final across all replicas.

  * `Eventually Consistent:`

    * If no new updates continue, replicas will eventually converge to consistent values.

  > BASE generally represents a different set of trade-offs from systems that prioritize strong consistency and strict transactional guarantees.

* When designing or using a distributed database, a data engineer needs to understand how that database handles consistency.

* Consistency behavior can depend on:

  * The underlying `database technology`.
  * Database `configuration parameters`.
  * Consistency settings at the `individual query level`.

* Consistency requirements should also be determined by the `business use case`.

  > Technical capabilities alone do not determine the correct consistency model. You may need to discuss requirements with stakeholders and decide whether the business needs the latest possible data or whether slightly stale data is acceptable in exchange for higher availability.

* `Example: Sales data with read replicas`

  * Suppose sales data needs to be provided to data scientists so they can update an analytics dashboard for the marketing team.
  * The production database has a `read replica`.
  * The data engineer can ingest, transform, store, and serve the required sales data from this distributed database setup.

* Suppose the database uses `Amazon RDS Aurora`.

  * Aurora is a distributed relational database service.
  * The main database instance provides `strict read-after-write consistency`.
  * This represents `strong consistency`.

* If the data scientist requires the most up-to-date sales data:

  * Read from the `main database instance`.
  * This provides strong consistency.

* If the data scientist prioritizes immediate access over having the absolute latest data:

  * Read from a `read replica`.
  * The replica may temporarily contain slightly older data.
  * This provides `eventual consistency`.

* Read replicas:

  * Track changes made to the main database instance.
  * Update their own copies of the data.
  * Can serve read requests independently from the main database.

* The application can therefore choose the consistency level on a `query-by-query` basis:

  * Main database → stronger consistency.

  * Read replica → potentially stale but highly available data.

  > This demonstrates that consistency is not always an all-or-nothing architectural decision. Depending on the database, different queries can sometimes use different consistency guarantees based on their requirements.

* The key trade-offs in distributed storage are:

  * `Scalability` through adding nodes.
  * `Durability` through replication.
  * `Availability` through redundant nodes and replicas.
  * `Performance` through parallel processing and local/less-congested replicas.
  * `Consistency` when replicated data needs to stay synchronized.
  * `Cost` of operating additional storage and compute nodes.

* A useful mental model:

  * `Replication` → copies the same data to multiple places.
  * `Partitioning` → splits different data across multiple places.
  * `Horizontal scaling` → adds more nodes.
  * `Fault tolerance` → continues operating despite failures.
  * `High availability` → continues responding despite failures.
  * `Strong consistency` → reads see the latest committed data.
  * `Eventual consistency` → replicas may temporarily differ but eventually converge.
  * `CAP` → during network partitions, distributed systems face a consistency vs availability trade-off.


# Database Storage Engines and Indexing

* `Database Management System (DBMS):`

  * A database typically has a software layer called a `DBMS` that manages interaction with the database.

  * DBMSs are used with:

    * `Relational databases`
    * `Non-relational databases`

      * Graph databases
      * Vector databases

  * A DBMS consists of several major components:

    * `Transport system`

    * `Query processor`

    * `Execution engine`

    * `Storage engine`

  > The storage engine is responsible for the physical storage and organization of data, while the query processor and execution engine are involved in processing queries.

* `Storage engine:`

  * Handles the physical storage of data on disk.

  * Important responsibilities include:

    * `Serialization`
    * Arrangement of data.
    * `Indexing`

  * Modern storage engines are commonly optimized for:

    * `SSD` performance characteristics.
    * Modern data types and structures.

  * They can support structures such as:

    * Variable-length strings.
    * Arrays.
    * Nested data.

  * As organizations began working with increasingly large datasets for analytics, storage engines evolved to provide strong support for `columnar storage`.

  > Columnar storage is particularly useful for analytical workloads because analytical queries often access only a subset of columns from very large datasets.

* `Database indexing:`

  * When querying very large tables, scanning every row can be expensive.

  * For example, imagine a table containing millions or billions of rows and a query that needs the average product price for records where `country = 'USA'`.

  * Without an index:

    * The query processor may need to scan the entire table.
    * Each row must be checked to determine whether its country matches `USA`.
    * This is called a `linear scan`.

  * An `index` is a special data structure that stores additional metadata to help locate required data more efficiently.

  > An index is essentially an additional lookup structure maintained alongside the table so the database does not have to search every row directly.

* Indexes in relational databases are commonly used for:

  * `Primary keys`

  * `Foreign keys`

  * Indexes can also be created on other columns when required by the application's access patterns.

  > Indexes should be created based on how data is queried. They improve retrieval performance but require additional storage and maintenance when the underlying data changes.

* `Example: country index`

  * Suppose a table contains a `country` column and millions of rows.

  * A separate index can contain:

    * The `country` values sorted alphabetically.
    * References to the corresponding rows in the original table.

  * When querying for `USA`:

    * The database can search the sorted index.

    * It can use `binary search` to quickly locate the relevant country values.

    * The index provides references to the corresponding rows.

    * The database can then retrieve those rows instead of scanning the entire table.

  > Searching a sorted index is much faster than checking every row individually.

* `Time complexity:`

  * A linear scan through `N` rows has a time complexity of approximately `O(N)`.

  * Binary search on a sorted index has a time complexity of approximately `O(log N)`.

  > You don't need to memorize the mathematical notation to understand the main idea: binary search eliminates large portions of the search space at each step, making it much faster than checking rows one by one.

* There are many different types of index structures.

  * The appropriate index depends on:

    * The database technology.
    * The type of data.
    * The application's query patterns.
    * The access requirements.

* `In-memory databases:`

  * Use `RAM` as the primary storage layer instead of relying primarily on disk storage.

  * RAM provides:

    * Extremely high transfer speeds.
    * Very low latency.

  * However, RAM is `volatile`.

    * Data can be lost when the machine loses power or restarts.

  * In-memory databases are therefore mainly useful for applications requiring `ultra-fast data retrieval`, such as:

    * Caching.

    * Real-time bidding.

    * Gaming leaderboards.

  > In-memory databases are generally not appropriate as the only storage layer when data must be retained permanently.

* `Memcached:`

  * A key-value storage system commonly used for caching.

  * Can be used to store:

    * Database query results.
    * API responses.

  * It is appropriate when losing cached data after a machine restart is acceptable.

  > Cached data can usually be regenerated from the underlying persistent data, which makes volatility acceptable for many caching workloads.

* `Redis:`

  * A popular in-memory `key-value store`.

  * Supports more complex data types than a basic key-value cache.

  * Provides built-in persistence mechanisms, including:

    * `Snapshotting`
    * `Journaling`

  * This allows Redis to be used for extremely high-performance applications where a small amount of data loss may be acceptable.

  > Redis can provide persistence, but its suitability depends on the application's durability requirements. Persistence mechanisms do not automatically make an in-memory system equivalent to a traditional durable database for every workload.

* Key concepts to remember:

  * `DBMS` → software layer that manages interaction with a database.
  * `Storage engine` → physically stores and organizes database data.
  * `Index` → additional data structure used to locate data faster.
  * `Linear scan` → checks rows one by one, approximately `O(N)`.
  * `Binary search` → searches sorted data efficiently, approximately `O(log N)`.
  * `In-memory database` → uses RAM for extremely fast access but must account for RAM's volatility.
  * `Memcached` → commonly used for simple caching where data loss is acceptable.
  * `Redis` → in-memory key-value store with richer data types and persistence mechanisms.


# Row-Oriented vs Column-Oriented Storage

* Structured tabular data and semi-structured data can be stored using two common storage patterns:

  * `Row-oriented storage`
  * `Column-oriented storage`

* The choice between row and column storage should be based primarily on the `data access pattern`, meaning how users and applications will access the data.

* `Row-oriented storage:`

  * Stores data `row by row`.

  * Each row represents a complete record.

  * On physical storage, each row is stored as a consecutive sequence of bytes.

  * For semi-structured data, each complete object can similarly be stored as a consecutive sequence of bytes.

  > Related fields are stored close together, which makes it efficient to retrieve or modify an entire record.

* Traditional `relational database management systems` commonly use row-oriented storage.

* Row storage is well suited for `OLTP` workloads because transactional operations often need to:

  * Find a specific record.
  * Read an entire record.
  * Update an entire record or several fields within it.
  * Perform small and frequent reads and writes with low latency.

* Example:

  * If a query needs to retrieve a specific record using an `id`:

    * The database locates the record.
    * Since all fields belonging to that record are stored together, the entire record can be efficiently read.
    * The record can also be efficiently updated.

* `Problem with row storage for analytical queries:`

  * Analytical queries often operate on an entire column rather than individual records.

  * Examples:

    * What was the `total revenue`?
    * Which product was sold most frequently?
    * What was the `average quantity`?

  * With row storage, even if the query needs only one column, the database may need to read the complete rows containing that column.

* Example with `1 million rows`:

  * Dataset contains:

    * `1,000,000 rows`
    * `30 columns`
    * `100 bytes per entry`

  * Assume the `price` column is the second column.

  * Query:

    ```sql
    SELECT SUM(price)
    FROM mytable;
    ```

  * With row storage:

    * All rows need to be transferred from persistent storage to `RAM`.
    * The rows are then sent to the `CPU`.
    * The CPU extracts the `price` value from each row and calculates the sum.

  * Total data transferred:

    * `1,000,000 rows × 30 columns × 100 bytes`
    * = `3,000,000,000 bytes`
    * ≈ `3 GB`

  * At a disk transfer speed of `200 MB/s`:

    * `3000 MB ÷ 200 MB/s`

    * = `15 seconds`

  > Row storage can work reasonably well for analytical queries on smaller datasets, but the amount of unnecessary data that must be read becomes a major problem as the dataset grows.

* Example with `1 billion rows`:

  * Same structure:

    * `1,000,000,000 rows`
    * `30 columns`
    * `100 bytes per entry`

  * Total data:

    * `3000 GB`

  * At `200 MB/s`, reading all rows and columns would take `just over 4 hours`.

  > The problem is that the query only needs the `price` column, but row storage requires reading the other columns as well.

* `Column-oriented storage:`

  * Stores data `column by column`.

  * All values belonging to one column are stored together.

  * The first column is stored together, then the second column, and so on.

  > Because values from the same column are physically stored together, a query that needs only a few columns can read only those columns instead of scanning complete rows.

* Column storage is particularly well suited for `analytical queries` because analytical workloads frequently operate on specific columns.

* Using the same example:

  * Dataset contains:

    * `1 billion rows`
    * `30 columns`
    * `100 bytes per entry`

  * Query:

    ```sql
    SELECT SUM(price)
    FROM mytable;
    ```

  * Since `price` is stored together in column storage:

    * Only the `price` column needs to be transferred from disk to RAM.
    * The other 29 columns do not need to be read.

  * Data transferred:

    * `1 billion entries × 100 bytes`
    * = `100 GB`

  * At `200 MB/s`:

    * `100,000 MB ÷ 200 MB/s`
    * ≈ `8.3 minutes`

  * Compared with row storage:

    * Row storage → `just over 4 hours`

    * Column storage → `just over 8 minutes`

  > Column storage is dramatically more efficient for analytical queries on large datasets because it avoids reading columns that the query does not need.

* `OLAP and column storage:`

  * Column-oriented databases are well suited for `OLAP`.
  * OLAP workloads commonly:

    * Process large datasets.
    * Perform aggregations.
    * Read only a subset of columns.
    * Scan large numbers of records.

* `OLTP and row storage:`

  * Row-oriented storage is well suited for `OLTP`.
  * OLTP workloads commonly:

    * Access individual records.
    * Read multiple fields from a record.
    * Perform frequent updates.
    * Require low-latency operations.

* `Problem with column storage for transactional workloads:`

  * Individual rows are not stored together.

  * To retrieve one complete record, the database may need to read the required values from multiple columns and reconstruct the row.

  * Updating a single field in a record can also be expensive:

    * The relevant column data must be deserialized.

    * The value must be modified.

    * The modified data must be written back to storage.

  > Column storage is optimized for reading selected columns across many rows, not for repeatedly reading and modifying individual records.

* `Key difference:`

  * `Row storage` → stores complete records together.

    * Efficient for accessing individual records.
    * Good for `OLTP`.
    * Poorer for large analytical scans that need only a few columns.

  * `Column storage` → stores values from the same column together.

    * Efficient for scanning selected columns across many records.
    * Good for `OLAP`.
    * Poorer for frequent transactional operations on individual records.

* A useful rule when choosing between them:

  * Use `row storage` for:

    * Transactional workloads.
    * Production application databases.
    * Frequent reads and writes of individual records.
    * `OLTP`.

  * Use `column storage` for:

    * Analytical processing.

    * Large-scale aggregations.

    * Queries that access a subset of columns.

    * `OLAP`.

  > The most important factor is the `access pattern`, not simply whether the data is structured or semi-structured.

* `Parquet` is a popular data format that combines concepts related to both row and column-oriented storage approaches.

* `Wide-column databases` are another storage approach and can be suitable for certain transactional workloads, although they are less commonly encountered than Parquet in modern data engineering.

* The fundamental reason for the row-vs-column trade-off is:

  * `Row storage` minimizes the work required to reconstruct a complete record.

  * `Column storage` minimizes the amount of data that must be read when a query needs only specific columns.

  > Think of it as **record access vs column access**: if you usually need the whole record, row storage is a natural fit; if you usually need a few columns from many records, column storage is a natural fit.


# Parquet, ORC, and Wide-Column Databases

* `Row-oriented vs column-oriented storage:`

  * `Column-oriented storage` is suitable for analytical workloads where operations such as aggregations are performed on columns.
  * It is not well suited for:

    * Reading individual rows.
    * Writing rows.
    * Updating rows.
  * `Row-oriented storage` is suitable for transactional workloads that require low-latency reads and writes.
  * It is not as efficient for large-scale analytical workloads.

* `Parquet and ORC:`

  * `Parquet` and `ORC (Optimized Row Columnar)` are file formats that use a `hybrid approach`.

  * They combine ideas from row-oriented and column-oriented storage.

  * The hybrid approach works by:

    * Horizontally partitioning the data into `row groups`.

    * Storing each row group using a `column-wise` format.

  > This provides the benefits of column-oriented storage within manageable groups of rows, making these formats well suited for analytical data processing.

* `ORC:`

  * Was particularly popular with `Apache Hive`.
  * Is generally less popular than Parquet today.
  * Has somewhat less support across modern cloud ecosystem tools.

* `Parquet:`

  * Is one of the most widely used columnar file formats in modern data engineering.
  * Its structure is based on `row groups`, `column chunks`, and `pages`.

* `Parquet file structure:`

  * A Parquet file is horizontally divided into `row groups`.

  * A row group has a default size of approximately `128 MB`.

  * Each `row group` contains a `column chunk` for every column in the dataset.

  * Each `column chunk` is further divided into `pages`.

  * Each `page` contains:

    * Encoded values for that column chunk.

    * Metadata such as:

      * `Minimum` value.
      * `Maximum` value.
      * `Count` of values.

    * `Repetition levels`.

    * `Definition levels`.

    * Other information needed to reconstruct nested data structures.

  > The hierarchy to remember is: `Parquet file → Row groups → Column chunks → Pages`.

* `Parquet and semi-structured data:`

  * Parquet can store:

    * Tabular data.
    * Nested semi-structured data such as `JSON`.

  * Repetition and definition levels help Parquet represent and reconstruct nested structures.

* `Parquet portability:`

  * One major advantage of Parquet is its `portability`.

  * It works well when exchanging data between different external tools.

  > Proprietary cloud data warehouse columnar formats may require data to be deserialized and re-serialized when moving data between systems, while Parquet provides a common interoperable format that many tools can read directly.

* `Wide-column databases:`

  * A `wide-column database` is a type of `NoSQL database`.

  * It can be thought of as combining characteristics of:

    * Relational databases.
    * Document stores.
    * Key-value databases.

  * In a traditional `key-value database`, data is represented as:

    * A `key`.
    * A corresponding `value`.

  * In a wide-column database, the value associated with a key can itself contain multiple key-value pairs.

* Wide-column databases organize data using a structure similar to a `two-dimensional key-value map`.

  * Each `row` generally represents a single entity.
  * Each row is identified by a unique `row key`.
  * Related columns are grouped into `column families`.
  * Each column family contains columns with unique column names.
  * Actual data is stored in `cells`.

* A cell is uniquely identified by the combination of:

  * `Row key`

  * `Column family`

  * `Column name`

  > Think of a cell's location as `(row key, column family, column name)`.

* `Example: customer and purchase data`

  * A customer can use `customer_id` as the `row key`.

  * For example:

    * `customer_id = 12345`

  * This row could contain two column families:

    * `Customer information`

      * `first_name`
      * `last_name`
      * `phone_num`
    * `Purchase information`

      * `invoice_num`
      * `store_id`

  * The row key identifies the customer, while the column families organize different types of information about that customer.

* `Cell versioning:`

  * Each cell value can be `versioned`.

  * A version is typically identified by a `timestamp`.

  * If a customer changes their phone number:

    * Multiple values can exist for the same cell.
    * Each value can have its own timestamp/version.

  * If a customer makes multiple purchases:

    * The corresponding cells can contain multiple versions.

    * Each version is associated with its timestamp.

  > Versioning allows a wide-column database to retain multiple historical values for a cell rather than simply replacing the previous value.

* `Flexible schema:`

  * Wide-column databases generally do not enforce a strict table schema.

  * Columns are only written when data exists for them.

  * Example:

    * Customer `45678` may have:

      * `first_name`
      * `last_name`
    * But no `phone_num` because the phone number is unknown.
    * They also may not have a `purchase information` column family because they have not made any purchases.

  * When new information becomes available:

    * The relevant column or column family can be added.

  > Unlike a traditional relational table with a fixed schema, a wide-column database allows different rows to contain different columns and column families. This makes it useful for sparse and evolving data.

* `Wide-column physical storage:`

  * Column families are typically stored `separately on disk`.

  * For the customer example:

    * `Customer information` is stored separately.
    * `Purchase information` is stored separately.

  * Within a column family, data is stored in a `row-oriented fashion`.

  * For the purchase information:

    * Data for a specific `row key` is stored together.

    * The values associated with that row are stored next to each other.

    * The database then moves to the next row key.

  > Wide-column databases therefore combine flexible, column-family-based data modeling with row-oriented storage within each column family.

* `Wide-column database characteristics:`

  * `NoSQL` database.
  * Uses `row keys`.
  * Groups related columns into `column families`.
  * Stores actual values in `cells`.
  * Supports cell `versioning`.
  * Allows flexible schemas.
  * Can efficiently represent sparse data because empty columns do not need to be stored.
  * Column families can be stored separately on disk.
  * Data within a column family is organized by row.

* Examples of wide-column databases include:

  * `Apache HBase`
  * `Google Bigtable`
  * `Apache Cassandra`
  * `Amazon Keyspaces`

* `Key points to remember:`

  * `Row storage` → good for transactional access to complete records.
  * `Column storage` → good for analytical operations over columns.
  * `Parquet / ORC` → hybrid file formats that partition data into row groups and store each group column-wise.
  * `Parquet` → widely used, portable, and well suited for modern data engineering workloads.
  * `Wide-column database` → NoSQL structure based on row keys, column families, and cells with a flexible schema.
  * `Column family` → group of related columns.
  * `Cell` → actual stored value identified by row key + column family + column name.
  * `Cell version` → historical value identified by a timestamp/version.
