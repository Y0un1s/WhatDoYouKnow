# Working with Source Systems
**Different types of data that you will deal with in the source systems:**

* Structured: data in a tabular format

* Semi-Structured: data that is not in a tabular format but still has some structure to it for example
    * JSON : A series of key-value pairs

* Unstructured: Data that does not have any predefined structure (they do have an inherent structure behind the scenes tho in the sense that there are dimensions pixels, colors etc)

**Different types of source systems**

* There are 3 general types of source systems that you will encounter
    
    * Databases: you will most likely be ingesting `structure` data and `semi-structured` data
        
        * they typically stored data in an organized way that allow you to preform CRUD 
        operations as a transactional pattern, CRUD is : create, read, update, delete
    
        * there is a software interface(DBMS) that sits between the physical database storage and the application or person interacting with the database to allow you to manipulate the stored data in the database
    
        * there are 2 types of databases: `Relational databases`: that store data in tables with rows and columns and has a predefined schema, and `Non-Relational databases` a.k.a `NoSQL` databases a.k.a `Not-Only SQL` databases that store Non-tabular data and doesn't have a predefined schema, they use dynamic a.k.a flexible schema instead
    
            >A schema is the blueprint or architecture of a database that defines how data is organized, structured, and related, including the tables, columns , data types, constraints, and how the data is related.
    
    * Files: anything 
    
        * files can be `structured` like a spreadsheet, `semi-structured` like a JSON or an xml file and `unstructured` like an image or a video
    
        > Spreadsheets (.csv, .xlsx, .gsheet) may contain either structured or semi-structured data. They are structured when they conform to a predefined schema and semi-structured when the schema is optional, inconsistent, or loosely enforced. CSV files do not enforce schemas, while Excel and Google Sheets provide only limited schema enforcement.
    
        * it's not strange to think of files like images, videos, csvs, audio etc.. as a source system, applications of all types write data to files so files are universal medium of data exchange, and they are one of the most common sources you will work with as a data engineer
    
        * You can access these files through any file system like Google Drive or an object storeage system like amazon S3 or simply as an attachment to an email
    
    * Streaming Systems: you will most likely be ingesting `semi-structured` data
    
        * Streaming systems is providing a continuous flow of data recorded as messages that contain information about events, in practice you might interact with a stream of events via messages queues like Amazon SQS `queue messages until a consumer process them usually each message is processed by one consumer` or other streaming platforms like kafka `store and distribute continuous stream of events, multiple consumers can read the same event independently` 
    
        > events are something that happened in the world or simply a change to the state of a system.

# Relational Databases
* The most common type of source system you will interact with is relational databases (it's everywhere)

* the name comes of storing data in different tables that are related to one another through set of keys or common attributes

    * structuring the data this way `reduces redundancy` by storing each piece of info once and makes data `easier to manage` by seperating different entities (customers orders products) and `improves data consistency and integrity` by using keys and constraints to maintain valid relationships and prevent anomalies during data insertion, update, and deletion.

    * > Keys and constraints maintain valid relationships between tables and prevent data anomalies. For example, a foreign key ensures that an order cannot be inserted for a customer who does not exist (insertion anomaly), prevents updates that would break the relationship between customers and orders (update anomaly), and prevents deleting a customer while related orders still exist unless the database is configured to cascade or restrict the deletion (deletion anomaly).

    * >**Integrity** is the broader concept of ensuring that data remains **accurate, valid, complete, and consistent** throughout its lifecycle. **Consistency** is one aspect of integrity and only ensures that data does not violate relationships or constraints. A database can be consistent yet still lack integrity. For example, a customer record with `Age = -5` may satisfy all key and relationship constraints, but it lacks integrity because the value is invalid according to the business rules.

    * Relational databases represent relationships between tables using keys.
        * Primary Key (PK): A column or a combination of columns whose values are unique and NOT NULL, used to uniquely identify each row in a table.
        * Foreign Key (FK): A column or a combination of columns that references the primary key (or another unique key) of another table, or the same table in the case of a self-referencing relationship. A foreign key establishes and enforces the relationship between tables, helping maintain referential integrity.

* Data Normalization (developed in the 1970s): Minimize Redundancy and ensure data integrity by storing data across tables in a logical way

* **Why worry?** about data redundancy or duplicate information in the first place, turns out that a normalized database provides a high degree of integrity and minimize redundancy but it can actually be slow when it comes to querying the data, data integrity is important of course
but the answer on how do you store your data depends on what you are trying to optimize for, as a data engineer you might ingesting data from a normalized database but depending on the end use case you are serving you might decide to organize the data according to a different model inside your own storage system like a one big table (OBT)

* when it comes to interacting with databases you will use a Relational database managment system (RDBMS) like MySQL or PostgreSQL or SQL Server they all support the Structured Query Language known as `SQL`

# Non Relational databases

* Non Relational databases were developed to trade certain RDBMS characteristics like strong consistency, joins, and a fixed schema for more schema flexibility scalability and improved performance
* NoSQL stands for Not Only SQL, it can still support SQL or SQL-Like query languages
* NoSQl Databases: it has Non-tabular structures and it supports variable data formats like `Key-Value`, `Document`, `Wide-Column`, `Graph` and others
    
    * unlike Relational databases NoSQL databases don't require a predefined schema, this mean you have more flexibility when deciding how to store your data 
    
* NoSQL databases excel in horizontal scaling which means automatically distributing data and workloads across multiple servers to meet increased traffic demands
    
* when a user writes data to a NoSQL database that is distributed across multiple nodes that write operation is first performed on a single node in this distributed system which is a location where one version of this database is running then there might be a slight delay before those changes are propagated to all other nodes
     
    * unlike relational databases NoSQL databases operate under the principle of eventual rather than strong data consistency meaning the database will allow you read from a node that has not received the latest write update and you may not get the most up to date data with a relational database with strong consistency you will not be able to read the data until all nodes are updated so NoSQL databases prioritize speed which is perfect for applications where system availability and scalability are more important than consistency such as social media platforms

* in terms of data integrity not all NoSQL databases guarantee the principles of ACID Compliance `Atomicity, Consistency, Isolation and Durability` but some do like mongoDB

    * that means when you are sourcing data from a NoSQL database that means you might need to take extra steps to ensure data integrity.

* Finally, NoSQL databases use specialized query languages tailored to their data model which are often but not always different from SQL.

* 2 Types of NoSQL databases: 
    
    * key-value databases: stores data as a collection of key-value pairs similar to what you might find in a python dictionary structure or a JSON file they key serves as a unique identifier to retrieve the corresponding value.
        *  This First type is perfect for scenarios such as where fast look-up is needed 

    * document databases: Stores data in JSON like documents each document has a unique key that identifies a document and allows you to retrieve that document's data and documents are organized into collections so you can think of a collection sort of like a table in a relational database and a document like a row
        
        * This locality makes it easier to retrieve all the information about a particular user compared to a relational database where the information is spread across multiple table, but documents don't support joins so it's less efficient to combine information from multiple documents
        * It has a very flexible schema, and it's usually used in applications that involves sensor readings, catalog, and content managements
        * This flexibility can be a downside, document databases can be an absolute nightmare to manage inquiry, and if you use it as a source system for your data as a data engineer this flexibility can make it easier for the sources system owners to change something that can break your pipelines

# ACID Compliance

* Both types of databases can support very high transaction rates, so they are used often in OLTP systems, and these systems need to store rapidly changing application states.

* Most traditional database systems are ACID compliant which means they support the principles of: `Atomicity, Consistency, Isolation, and Durability`. which helps ensure that transactions are accurate and reliable in an OLTP system.

* Many NoSQL databases are not ACID compliant by default, but some are ACID compliant

* `Atomicity:` ensures that transactions are atomic, treated as a single indivisible unit.
  
    * If a transaction contains multiple operations, Atomicity ensures that all operations are executed successfully or non of them are meaning if one failed the whole transaction fails.

* `Consistency:` ensures that any changes to the data made withing a transaction follow the set of rules or constraints defined by the database schema, to ensure that the database is consistent with a predefined schema
    >ACID consistency is different from distributed consistency (e.g., in the CAP theorem). ACID consistency ensures that each transaction preserves the correctness and validity of the data according to the database schema. Distributed consistency ensures that all replicas or nodes observe the same data at the same time. Although strong distributed consistency often relies on ACID-compliant transactions, the two concepts describe different aspects of database systems.

* `Isolation:` Ensures that concurrent transactions behave as if they were executed one at a time (serially). The database determines the execution order using its concurrency control mechanism, isolation does not guarantee first-come-first-served or first-to-finish ordering.
    > Isolation does not specify who "wins." It only guarantees that the final result is equivalent to some valid serial execution (as if the transactions had run one after another), Which transaction proceeds first depends on the database's concurrency control mechanism, such as locking, timestamp ordering, optimistic concurrency control, or MVCC.

* `Durability:` Once a transaction is completed it's effects are permanent and will survive any subsequent system failures such as power loss, this is essential in maintaining reliability of the database in events like natural disasters

* ACID guarantees that each transaction preserves the correctness and consistency of the database. In distributed ACID databases, additional mechanisms such as synchronous replication and consensus protocols ensure that all replicas observe the same committed data (strong consistency). Together, these mechanisms allow the database to present a single, consistent view of the data across multiple servers.

* Full ACID compliance is not necessary for all types of applications. Some database systems relax one or more ACID transaction guarantees, while others adopt weaker distributed consistency models (eventual consistency instead of strong consistency) to improve performance, scalability, and availability. The choice depends on the application's requirements and the trade-offs it is willing to make.



# Object Storage

* It's the most important mechanism for file storage and retrieval in your work as a data engineer.

* Object storage treats data a.k.a files as individual objects and stores them in a flat structure that doesn't adhere to a traditional file system hierarchy.
> There can be a logical structure for an object storage system like amazon S3, this resembles a physical file hierarchy system but it's purely cosmetic, the actual storage structure is flat so all files are on the top level, and this by design because it allows quick access to all objects

* Objects can be anything, this versatility makes object storage the perfect repository for semi-structured and unstructured data, which can be useful for supporting applications like serving data for training machine learning models

* In object storage each object is assigned a Universal  Unique Identifier a.k.a (UUID) key

* Each object has it's associated metadata

* After the initial write objects technically become immutable, and they don't support random write or append operations, a file in an object storage system is not like a table in RDB or a document in a NonRDB where you can update or append to, to change the data in an object you must overwrite the whole object and have the uuid point to this new object.

* With object storage you can enable object versioning, this adds the version info to the metadata, so instead of overwriting the whole object you can keep multiple versions of the same object

* Why Object storage?
    
    * Stores files of various data formats without a specific fie system structure

    * Easily scale in and out to provide virtually limitless space

    * Replicate data across several availability zones
        >This makes the data highly available and durable even if natural disasters occur.
    
    * Object storage is cheaper than other storage options
        > especially if you are storing data that you don't need access to on a regular basis 
    
    * Cloud object storage is the underlying storage for new architecture design such as data lakes and data lake houses because of it's `scalability, flexibility, cost-effectiveness, and duarbility`
    

# Logs

* Loosely speaking you can think of logs as the simplest type of streaming system, well a log isn't even a system at all it's just a record of information to track the activity of a system or an application 

* An append-only sequence of records ordered by time, capturing information about events that occur in systems.
    > an append-only means new records can only be added to the end of the sequence, and existing records are never modified or deleted, eventually, the system may delete old events according to a retention policy.

* You can use logs to support the analysis of user behavior patterns, track changes in the source database in a process called `change data capture a.k.a CDC`, you could use those changes to trigger the ingestion of processes so they can run based on the arrival of new data in the databases, or you can use log data in certain machine learning applications like anomaly detection. 

* This makes logs a rich data source that can support downstream use cases like data analysis, troubleshooting issues, monitoring performance, machine learning applications, and automation.

# Streaming Systems

* Event: is something that happened in the world or a change to the state of a system, like a user clicking on a link or a sensor measuring a change in temperature
    
    > You can think of all data as streaming data at it's source, this is because essentially all data consists of a record of events that happened out in the world or withing a system

* Message: A record of information about an event, it might include details about the event, like which button the user clicked or what temperature the sensor recorded as well as some metadata around the event including a timestamp of when the event happened
    
    * messages can be generated continuously to form a stream

* Stream: is a sequence of messages that might be a series of website clicks or sensor readings over a period of time

* Collectively messages and streams make up streaming data.
    
    * if you want to handle chunks of this data all at once like over a specific time, then that would be batch processing applied to a stream of messages

    * if you want to process each message as it's received then you need a system that is set up to take action based on the incoming messages, so then what you have is a system where messages record information based on events and action is taken as messages are received a.k.a `streaming system`

* The words events and messages are used interchangeably when it comes to describing the various components of an event-driven architecture, in data engineering it's not important to distinguish between the two, so when we talk about messages or events being produced, consumed, or stored in a queue it's all the same thing.

**Three components of a streaming system**

* Event/Message Producer: What generates the messages in a stream, like an IoT device, a mobile app, an API or a website, to name a few examples.

* Event/Message Router/ Streaming Broker: the way the messages find their eay to the consumer, such as Apache Kafka which acts as a buffer to filter and distribute the messages to the consumers, it's a router that helps decouple the producer from the consumer, which enables asynchronous communication between them so the producer doesn't have to wait for the event to be delivered to the consumer before it can send another one, this also prevents events from being lost if the consumer is not immediately available.

    > decouple means to reduce or eliminate direct dependencies between two components so that they can operate independently
    
    > Asynchronous communication means the sender does not wait for the receiver to process the message before continuing its own work

* Event/Message Consumer(s): sometimes known as the subscriber, is what processes each of these messages, and there can be more than one consumer in any given streaming system

* when you work with streaming systems it could be that your source system is a simple event producer like an IoT device and your system comprises of the event router and the consumer, or it could be that your upstream source system is made of multiple event producers, streaming brokers and consumers and the system you build is just another downstream consumer of events

* There are 2 main types of streaming systems they differ mainly in how the event router works:
    
    * message queues: the event router accts as a queue that accumulates the messages that are sent by the producer, the event consumer then reads the messages from the queue in a FIFO order, and once the consumer reads the message from the queue and acknowledges this, the message is deleted from the queue, with a message queue the event producer can push new messages to the queue at any time, and the event consumer can read them at anytime, you can think of a queue itself as a sort of temporary storage solution that allows event producers to be decoupled from event consumers, an example of a message queue is Amazon SQS

    * streaming platforms: like Apache Kafka or Amazon Kinesis Data Streams, the event producer streams events to a log, which is an append-only record of events, the event router then distributes messages in the log to appropriate event consumers, the consumer process the messages in the log sequentially as a read-only operation, and this means that unlike message queues the message do not get deleted from the log and the data is persistent, since the data is retained this way in a streaming platform, it's possible to replay batch or reprocess events in the log from a past point in time.

# IAM

* IAM is a framework for managing permissions, permissions define what an identity (a person, an application) can perform on a specific set of resources, like a database or an ETL tool

* IAM ensures that the appropriate set of identities have access to the right resources at the right time
    
    > IAM is how you exercise the least privilege principle as it allows you to grant people and applications access to only the essential resources they need to do their job and only for the duration that it's required.

* There is different types of AWS identities:
    
    * Root User: has unrestricted access to all resources

    * IAM User: who have specific permissions to certain resources

    * IAM Group: A collection of users that inherit the same permissions from a group policy
        
        > A policy is just a JSON document that includes the details of what resources and permissions the group should have.
    
    * IAM Role: A user, application, or service that's been granted temporary permissions to preform specified actions on a certain resource for a limited time
    
        > By default, AWS resources cannot perform actions on each other unless you explicitly grant permission.


# Basic of networking on the cloud

* A cloud data pipeline is a network of connected cloud resources, and networking determines how data flows between them.

* Cloud networking concepts are similar across AWS, Azure, and GCP.

* AWS infrastructure is organized as:

  * Region → Geographic location.
  * Availability Zone (AZ) → One or more physical data centers within a region.

* Resources and data can be replicated across Regions and AZs to improve availability, fault tolerance, and disaster recovery.

* When choosing an AWS Region, consider:

  * Compliance (data residency and legal requirements).
  * Latency (closer regions provide lower latency).
  * Availability (more AZs improve resilience).
  * Cost (pricing varies by region).

* Within any region you can make a custom virtual private cloud a.k.a VPCs, which are smaller networks that span multiple availability zones within the region 

* creating VPCs allows you to have more fine-grained control over who can access what resources.

* You can then divide the space in your vpc into subnetworks a.k.a subnets that house your actual data pipeline resources
    
    * each subnet can then have it's own security rules known as network access control a.k.a network ACL, as well as routing configurations through an internet gateway
        
        * this let's you create public subnets for internet-facing resources, like web servers, and private subnets for internal resources, like databases.

# VPC

* Every AWS Region contains a **default VPC**.

  > The default VPC is useful for testing, but in production you should create a custom VPC.

* A VPC (Virtual Private Cloud) is an isolated private network inside an AWS Region.

* A VPC can span multiple Availability Zones (AZs), and a Region can contain multiple VPCs.

  * Different VPCs are isolated by default and require explicit configuration to communicate.

* When creating a VPC you define:

  * A name.
  * The Region.
  * An IPv4 CIDR block (private IP range).

* CIDR (Classless Inter-Domain Routing) defines the private IP addresses available inside a VPC.

Example:

```text
10.0.0.0/16
```

* `10.0.0.0`

  * Starting network address.

* `/16`

  * First 16 bits identify the network.
  * Remaining bits are used for host IP addresses.

Example host addresses:

```text
10.0.1.10
10.0.50.20
10.0.200.5
```

---

# Subnets

* A subnet is a smaller network inside a VPC.

* Every subnet:

  * Belongs to one AZ.
  * Has its own CIDR block.
  * Must use a CIDR that is a subset of the VPC's CIDR.

Example:

```text
VPC
10.0.0.0/16

Public
10.0.1.0/24

Private
10.0.2.0/24
```

* `/24` means the first three numbers identify the subnet and the last number is used for host IP addresses.

* Resources are deployed into subnets, not directly into a VPC.

* Public subnets host internet-facing resources.

  * Examples:

    * Load Balancers.
    * NAT Gateways.
    * Bastion Hosts.

* Private subnets host internal resources.

  * Examples:

    * EC2 instances.

    * Amazon RDS databases.

    * Data engineering services.

  > Private subnets are not directly accessible from the Internet.

* A common production architecture consists of:

  * Two public subnets.
  * Two private subnets.
  * Distributed across multiple AZs for high availability and fault tolerance.

* Creating a VPC and subnets does **not** provide Internet access.

  > Internet connectivity requires components such as an Internet Gateway, Route Tables, and NAT Gateways.


# Internet Gateway (IGW)

* By default, a VPC and its subnets are isolated.

  * Resources cannot access the Internet.
  * The Internet cannot access resources inside the VPC.

* An Internet Gateway (IGW) enables Internet connectivity for a VPC.

  * Supports both inbound and outbound Internet traffic.

  * Only resources in **public subnets** can use the IGW.

  > Think of an Internet Gateway as the front door of a VPC.

* Attaching an Internet Gateway alone is **not enough**.

  > You must also configure Route Tables and security rules before Internet traffic can flow.

* An Internet Gateway has a one-to-one relationship with a VPC.

  * One VPC can have one Internet Gateway.
  * One Internet Gateway can only be attached to one VPC at a time.

---

# NAT Gateway

* A NAT (Network Address Translation) Gateway allows resources in **private subnets** to initiate outbound Internet connections.

  * Allows outbound traffic.

  * Blocks inbound connections initiated from the Internet.

  > Think of a NAT Gateway as a controlled one-way door that only lets private resources go out to the Internet.

* Common use cases:

  * Download software updates.
  * Install packages.
  * Retrieve patches.
  * Access AWS services without exposing resources to the Internet.

* A NAT Gateway is deployed in a **public subnet**.

* A common best practice is to deploy one NAT Gateway per Availability Zone (AZ) for high availability.

---

# Application Load Balancer (ALB)

* An Application Load Balancer (ALB) is the entry point for external users.

* The ALB:

  * Receives incoming requests from the Internet.
  * Distributes traffic across multiple EC2 instances.
  * Can route traffic across multiple Availability Zones.

* Using an ALB allows EC2 instances to remain in **private subnets** while still serving Internet traffic.

  > Clients communicate with the ALB, not directly with the EC2 instances.

---

* **Typical Architecture**

```text
Internet
    │
Internet Gateway
    │
Application Load Balancer
    │
Public Subnets
    │
NAT Gateway
    │
Private Subnets
    ├── EC2
    └── RDS
```

> Even after creating an Internet Gateway and NAT Gateways, Internet connectivity is not complete. Route Tables and security rules must still be configured.


# Route Tables

* A Route Table contains a set of routes that determine where network traffic is sent.

* Every subnet is associated with a Route Table.

* When a VPC is created, AWS automatically creates a **default Route Table**.

  * Allows communication between resources inside the VPC.

  * Does **not** provide Internet connectivity.

  > Internet connectivity requires custom routes.

* Every route consists of:

  * Destination → Where the traffic is going.
  * Target → Where the traffic should be forwarded.

* `0.0.0.0/0` matches every IPv4 address.

  > It represents the entire Internet.

* Public subnets route Internet-bound traffic to an **Internet Gateway (IGW)**.

Example:

```text
Destination        Target
0.0.0.0/0          Internet Gateway
```

* Private subnets route Internet-bound traffic to a **NAT Gateway**.

Example:

```text
Destination        Target
0.0.0.0/0          NAT Gateway
```

* This allows resources in private subnets to:

  * Access the Internet.

  * Download updates and patches.

  * Access AWS services.

  > The Internet cannot initiate connections to resources in private subnets through a NAT Gateway.

* A common production architecture associates:

  * Public Route Tables → Public Subnets.

  * Private Route Tables → Private Subnets.

  > A common best practice is to create a separate Route Table for each subnet to simplify management.


# Security Groups & Network ACLs

* By default, even with an Internet Gateway and Route Tables configured, resources cannot receive traffic until security rules allow it.

### Security Groups

* A Security Group acts as a **stateful virtual firewall** for individual AWS resources.

  * Controls inbound and outbound traffic.
  * By default:

    * Denies all inbound traffic.

    * Allows all outbound traffic.

  > Because Security Groups are stateful, return traffic is automatically allowed.

* Security Groups can be attached to resources such as:

  * EC2 instances.
  * Application Load Balancers.
  * Amazon RDS databases.

* Security Groups can reference other Security Groups.

  > This is known as **Security Group Chaining**.

Example:

```text
Internet
    │
ALB Security Group
    │
EC2 Security Group
    │
RDS Security Group
```

---

### Network ACLs (NACLs)

* A Network ACL is a **stateless firewall** that protects an entire subnet.

* By default:

  * Allows all inbound traffic.

  * Allows all outbound traffic.

  > Because Network ACLs are stateless, inbound and outbound rules must both be configured.

* Network ACLs provide an additional layer of security and are commonly used for subnet-level policies or troubleshooting connectivity issues.

---

### Troubleshooting Network Connectivity

If resources cannot communicate, check in this order:

1. Internet Gateway attachment.
2. Route Tables and subnet associations.
3. Security Group rules.
4. Network ACL rules.
5. Resource configuration (correct subnet and Security Group).

> A subnet becomes **public** or **private** based on its **Route Table**, not because of how it was created.


# Message Queues & Event Streaming Platforms

* Streaming ingestion generally uses two types of systems:

  * **Message Queues**
  * **Event Streaming Platforms**

---

### Message Queues

* A Message Queue is a temporary buffer that delivers messages asynchronously from producers to consumers.

* Messages are typically processed using **FIFO (First In, First Out)**.

  * The oldest message is consumed first.
  * Once consumed, the message is removed from the queue.

> Message queues are best when messages only need to be processed once.

---

### Event Streaming Platforms

* An Event Streaming Platform stores messages in a persistent **append-only log**.

* Unlike message queues:

  * Messages remain available after being consumed.
  * Consumers can replay or reprocess historical events.
  * Multiple consumers can read the same events independently.

* Common examples include:

  * Apache Kafka
  * Amazon Kinesis Data Streams

> Event streaming platforms are designed for long-term event storage and multiple independent consumers.

---

### Kafka Architecture

At a high level:

```text
Producer
    │
    ▼
Kafka Cluster (Brokers)
    │
    ▼
Consumers
```

* **Producers** push events to a Kafka cluster.

* A Kafka cluster consists of one or more **Brokers**.

* **Consumers** pull messages from the cluster.

---

### Topics

* A **Topic** is a logical category that groups related events.

Examples:

* Customer Orders
* Fraud Alerts
* IoT Temperature Readings

* Producers choose which topic each message belongs to.

> Think of a Topic as a highway carrying one category of traffic.

---

### Partitions

* Each Topic contains one or more **Partitions**.

* A Partition is an ordered, immutable log of messages.

* Multiple partitions allow messages to be processed in parallel.

* Producers determine which partition receives each message.

Common strategies include:

* Round Robin
* Message Key Hashing

> More partitions increase throughput by distributing messages across multiple logs.

---

### Kafka vs Kinesis

| Kafka | Amazon Kinesis |
|--------|----------------|
| Topic | Stream |
| Partition | Shard |

* The concepts are nearly identical.

* Kinesis simply uses different terminology.

---

### Consumer Groups

* Consumers are organized into **Consumer Groups**.

* A Consumer Group can subscribe to one or more Topics.

* Within a Consumer Group:

  * Each partition is assigned to only one consumer.
  * Different consumers process different partitions.
  * Together, the group processes the entire topic.

```text
Topic
 ├── Partition 1 → Consumer A
 ├── Partition 2 → Consumer B
 └── Partition 3 → Consumer C
```

> This allows work to be distributed across multiple consumers without processing the same partition twice.

---

### Message Retention

* Kafka retains messages for a configurable retention period.

* Messages remain available even after they are consumed.

* Consumers can replay historical events whenever needed.

Common use cases:

* Reprocessing failed jobs.
* Recovering from application failures.
* Building new downstream systems from historical data.

> Message consumption does not delete the event from Kafka.

---

### Streaming Data Sources

Events can originate from many sources, including:

* Web server logs.
* Application events.
* IoT devices.
* Database Change Data Capture (CDC).

---

### Continuous Change Data Capture (CDC)

* **Continuous CDC** monitors a database's transaction log.

* Every insert, update, or delete operation is streamed into the data pipeline.

* This keeps downstream systems synchronized with the source database.

> CDC enables near real-time replication without repeatedly querying the database.

---

### Message Queue vs Event Streaming

| Message Queue | Event Streaming Platform |
|--------------|--------------------------|
| Temporary message storage | Persistent event log |
| FIFO processing | Append-only log |
| Message removed after consumption | Message retained after consumption |
| Usually one consumer processes a message | Multiple consumer groups can process the same message |
| No replay | Supports replay and reprocessing |

> Choose a Message Queue when each message is processed once. Choose an Event Streaming Platform when events need to be stored, replayed, or consumed by multiple independent applications.

# Amazon Kinesis Data Streams

* Amazon Kinesis Data Streams is AWS's managed event streaming service.

* Like Apache Kafka, it consists of:

  * Producers
  * Streams
  * Consumers

* Producers write data to a **Stream**, and consumers read data from it.

---

### Streams & Shards

* A **Stream** is composed of one or more **Shards**.

* A Shard is the unit of capacity in Kinesis.

* To increase the stream's capacity, you add more shards.

```text
Producer
    │
    ▼
Kinesis Stream
├── Shard 1
├── Shard 2
└── Shard 3
```

> More shards allow Kinesis to handle higher read and write throughput.

---

### Shard Capacity

Each shard has fixed limits.

#### Write Capacity

* Up to **1,000 records per second**.
* Up to **1 MB/sec** of write throughput.

#### Read Capacity

* Up to **5 read operations per second**.
* Up to **2 MB/sec** of read throughput.

> If your application exceeds these limits, you need additional shards.

---

### Scaling a Stream

There are two ways to scale a Kinesis stream.

#### On-Demand Mode

* AWS automatically adds or removes shards based on traffic.

* Best for:

  * Unpredictable workloads.
  * Applications with sudden traffic spikes.
  * Reducing operational management.

> You only pay for what you use.

---

#### Provisioned Mode

* You manually specify the number of shards.

* You are responsible for increasing or decreasing capacity.

* Best for:

  * Predictable workloads.
  * Better cost control.

---

### Kinesis Data Record

Every record written to a stream contains:

* Partition Key.
* Sequence Number.
* Data (stored as a Binary Large Object or **Blob**).

```text
Record

Partition Key
Sequence Number
Data (Blob)
```

---

### Partition Key

* The **Partition Key** determines which shard receives a record.

* The producer chooses the partition key.

Example:

```text
Customer ID = 12345
```

All records with that customer ID are likely stored in the same shard.

This preserves ordering for that customer's events.

> Choosing a good partition key is important for balancing data across shards.

---

### Sequence Number

* Kinesis automatically assigns a **Sequence Number** to every record.

* Sequence numbers preserve the order of records **within a shard**.

Example:

```text
Shard

1001
1002
1003
1004
```

> Ordering is guaranteed only within a single shard.

---

### Shared Fan-Out

By default, multiple consumers share a shard's read capacity.

```text
Shard
   │
   ├── Consumer A
   ├── Consumer B
   └── Consumer C
```

* All consumers compete for the shard's **2 MB/sec** read capacity.

> Heavy consumers can reduce the throughput available to others.

---

### Enhanced Fan-Out

* Enhanced Fan-Out gives each consumer its own dedicated read throughput.

* Every consumer receives up to **2 MB/sec** independently.

```text
Shard
   │
   ├── Consumer A (2 MB/sec)
   ├── Consumer B (2 MB/sec)
   └── Consumer C (2 MB/sec)
```

> Enhanced Fan-Out eliminates contention between consumers.

---

### Processing Kinesis Streams

Consumers can process stream data using AWS managed services, including:

* AWS Lambda.
* Amazon Managed Service for Apache Flink.
* AWS Glue.

You can also build custom consumers using the:

* Amazon Kinesis Client Library (KCL).

---

### Stream Chaining

* The output of one Kinesis Stream can become the input of another stream.

```text
Producer
    │
    ▼
Stream A
    │
    ▼
Consumer
    │
    ▼
Stream B
```

> This allows you to build complex real-time processing pipelines.

---

### Integrations

Consumers can forward processed data to other AWS services.

Example:

```text
Kinesis Stream
      │
      ▼
Consumer
      │
      ▼
Amazon Data Firehose
      │
      ▼
Amazon S3
```

---

### Multiple Applications

* Multiple independent applications can consume the same Kinesis Stream.

Example:

```text
Kinesis Stream
       │
 ┌─────┴──────────┐
 ▼                ▼
Analytics App   Fraud Detection App
```

Each application:

* Reads the same records.
* Maintains its own processing state.
* Sends data to different downstream systems.

> Multiple applications can process the same stream simultaneously without affecting one another.

# Change Data Capture (CDC)

* **Change Data Capture (CDC)** is a technique for detecting and capturing changes made to a database.

* Instead of copying the entire database repeatedly, CDC captures only the changes:

  * Insert
  * Update
  * Delete

> CDC keeps downstream systems synchronized with the source database while minimizing unnecessary data transfer.

---

### Why CDC?

Suppose you have:

```text
Application Database
        │
        ▼
Data Warehouse
```

After some time, new data is added to the application database.

You now need to update the data warehouse.

There are two approaches.

---

### Full Load (Full Snapshot)

* Every refresh copies the entire dataset.

Process:

```text
Delete Existing Data
        │
        ▼
Copy Entire Source Database
```

Advantages:

* Simple to implement.
* Guarantees consistency.

Disadvantages:

* Slow for large datasets.
* Uses significant CPU, memory, and network bandwidth.
* Inefficient when only a few records changed.

> Best for small datasets or infrequent updates.

---

### Incremental Load

* Only the records that changed since the last load are copied.

Example:

```text
Source Table

ID  Last_Updated
1   10:00
2   10:15
3   10:45
```

If the previous load occurred at **10:20**, only record **3** is copied.

Advantages:

* Much faster.
* Transfers far less data.
* More efficient for large datasets.

Disadvantages:

* More complex to implement.

> CDC is an incremental loading technique for databases.

---

### What Changes Does CDC Capture?

CDC captures every database change event.

* INSERT
* UPDATE
* DELETE

Example:

```text
INSERT Customer

UPDATE Order

DELETE Product
```

These changes are forwarded to downstream systems instead of copying the entire table.

---

### Common CDC Use Cases

#### Database Replication

Synchronize data between databases.

Example:

```text
PostgreSQL
      │
      ▼
Data Warehouse
```

This keeps analytics data up to date.

---

#### Hybrid Cloud Synchronization

Replicate data between:

* On-premises databases.
* Cloud databases.

---

#### Auditing

Capture every historical change.

Useful for:

* Regulatory compliance.
* Purchase history.
* Financial records.

---

#### Event-Driven Microservices

Database changes automatically trigger downstream services.

Example:

```text
Customer Places Order
        │
        ▼
Database Update
        │
        ▼
CDC
   ├── Shipping Service
   ├── Customer Notification
   └── Inventory Service
```

---

### CDC Communication Models

There are two ways changes can be delivered.

---

#### Push-Based CDC

* The source database immediately sends changes to downstream systems.

```text
Database
    │
    ▼
Target System
```

Advantages:

* Near real-time updates.

Disadvantages:

* Updates may be lost if the target system is unavailable.

---

#### Pull-Based CDC

* The target system periodically checks the source database for changes.

```text
Target System
      │
      ▼
Database
```

Advantages:

* Simpler and more reliable.

Disadvantages:

* Data arrives with some delay because polling happens at intervals.

---

### CDC Implementation Methods

There are several ways to implement CDC.

---

#### 1. Query-Based CDC (Batch CDC)

* The target system periodically queries the source database.

* Typically relies on a timestamp column such as:

  * `updated_at`
  * `last_updated`
  * `last_modified`

Example query:

```sql
SELECT *
FROM orders
WHERE updated_at > last_sync_time;
```

Advantages:

* Easy to implement.
* Works with most databases.

Disadvantages:

* Requires scanning tables.
* Adds load to the database.
* Not truly real-time.

> This is a **Pull-Based** approach.

---

#### 2. Log-Based CDC (Continuous CDC)

* Reads the database's transaction log instead of querying tables.

The transaction log records every database operation.

```text
Transaction Log

INSERT
UPDATE
DELETE
INSERT
UPDATE
```

CDC tools read these events and stream them to downstream systems.

Example pipeline:

```text
Database
      │
Transaction Log
      │
      ▼
Debezium
      │
      ▼
Apache Kafka
      │
      ▼
Consumers
```

Advantages:

* Near real-time updates.
* No table scanning.
* No `updated_at` column required.
* Minimal impact on database performance.

> This is the most common CDC implementation in modern data engineering.

---

#### 3. Trigger-Based CDC

* Database triggers detect changes automatically.

Example:

```text
UPDATE Customer
       │
       ▼
Database Trigger
       │
       ▼
CDC Pipeline
```

Advantages:

* Immediate notification of changes.

Disadvantages:

* Large numbers of triggers can reduce database write performance.

> This is a **Push-Based** approach.

---

### CDC Methods Comparison

| Method | Push / Pull | Real-Time | Extra Column Needed | Database Overhead |
|---------|-------------|-----------|---------------------|-------------------|
| Full Load | Pull | No | No | High |
| Query-Based CDC | Pull | No | Yes (`updated_at`) | Moderate |
| Log-Based CDC | Pull | Yes | No | Low |
| Trigger-Based CDC | Push | Yes | No | Can impact writes |

> Modern streaming architectures typically use **Log-Based CDC** because it provides near real-time updates with minimal impact on the source database.
