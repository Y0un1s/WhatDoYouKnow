# Data Pipelines Before Orchestration

Before tools like Airflow, pipelines were commonly automated using **Cron**.

---

### What is Cron?

**Cron** is a Unix utility for scheduling commands to run automatically.

A scheduled command is called a **Cron Job**.

---

### Cron Job Format

```text
Minute Hour Day Month DayOfWeek Command
```

Example:

```text
0 0 * * * python ingest_data.py
```

Runs every day at **12:00 AM**.

---

### Cron Fields

| Field | Values |
|--------|--------|
| Minute | 0–59 |
| Hour | 0–23 |
| Day of Month | 1–31 |
| Month | 1–12 |
| Day of Week | 0–6 (Sunday–Saturday) |

* `*` means **any value**.

---

### Using Cron for Pipelines

Each pipeline task is scheduled separately.

Example:

```text
12:00 AM → Ingest Data
1:00 AM  → Transform Data
2:00 AM  → Load Data
```

Tasks are expected to finish before the next one starts.

---

### Limitations of Cron

Cron provides scheduling only.

It does **not** provide:

* Dependency management.
* Automatic retries.
* Monitoring.
* Alerts.
* Failure tracking.

---

### When is Cron Useful?

Cron is suitable for:

* Simple scheduled scripts.
* Independent tasks.
* Small automation jobs.
* Prototyping.

---

### Cron vs Airflow

| Cron | Airflow |
|------|----------|
| Scheduling only | Workflow orchestration |
| No dependency management | Handles dependencies |
| No retries | Automatic retries |
| No monitoring | Built-in monitoring |
| No alerts | Alerting support |
| Best for simple tasks | Best for complex pipelines |

---

### Summary

* **Orchestration** automates and manages complete data workflows.
* **Apache Airflow** is the most widely used orchestration platform.
* **Cron** can schedule simple tasks but lacks workflow management.
* Modern data pipelines use orchestration tools to build reliable, scalable, and observable workflows.

# Core Orchestration Concepts

---

### Directed Acyclic Graph (DAG)

A **DAG (Directed Acyclic Graph)** represents a workflow.

* **Nodes** represent tasks.
* **Edges** represent dependencies.
* Data flows in one direction.
* No loops or cycles are allowed.

Example:

```text
Extract
   │
   ▼
Transform
   │
   ▼
Load
```

---

### Task Dependencies

Dependencies ensure tasks execute in the correct order.

Example:

```text
Extract
   │
   ▼
Transform
   │
   ▼
Load
```

A task starts only after its dependent task finishes successfully.

---

### Defining DAGs

In Airflow, DAGs are defined using **Python code**.

The Airflow UI allows you to:

* View DAGs.
* Trigger workflows.
* Monitor task execution.
* Debug failures.

---

### Scheduling DAGs

DAGs can run:

* On a schedule (time-based).
* When an event occurs (event-based).

Examples:

* Every day at midnight.
* Every hour.
* Every Monday.

---

### Event-Based Workflows

Instead of running on a schedule, a DAG can start when an event occurs.

Examples:

* A dataset is updated.
* A file is uploaded.
* A message arrives in a queue.

---

### Sensors

A **Sensor** waits for a specific event before allowing the workflow to continue.

Examples:

* Wait for a file in Amazon S3.
* Wait for a database table.
* Wait for an external job to finish.

---

### Monitoring & Logging

Orchestration tools provide:

* Task status.
* Execution logs.
* Runtime metrics.
* Failure tracking.

---

### Alerts

Alerts notify engineers when:

* A task fails.
* A workflow fails.
* A task exceeds its expected runtime.

---

### Data Quality Checks

Data quality tests can be included as workflow tasks.

Common checks:

* NULL values.
* Value ranges.
* Schema validation.
* Duplicate records.
* Required columns.

# Airflow Architecture & User Interface

* **Apache Airflow** is a workflow orchestration platform that automates and manages data pipelines.

* Airflow executes workflows defined as **DAGs (Directed Acyclic Graphs)**.

---

### Airflow Architecture

The main Airflow components are:

* Web Server.
* Scheduler.
* Workers.
* Metadata Database.
* DAG Directory.

---

### Web Server

The **Web Server** hosts the Airflow User Interface (UI).

It is used to:

* View DAGs.
* Trigger workflows.
* Monitor execution.
* Troubleshoot failures.

---

### DAG Directory

The **DAG Directory** stores the Python files that define your DAGs.

Any DAG placed in this directory automatically appears in the Airflow UI.

---

### Scheduler

The **Scheduler** continuously checks:

* DAG schedules.
* Task dependencies.

When a task is ready, it sends it to the execution queue.

---

### Executor

The **Executor** manages task execution.

It takes queued tasks and assigns them to workers.

---

### Workers

**Workers** execute the tasks defined in your DAG.

Task status changes typically follow this order:

```text
Scheduled
    ↓
Queued
    ↓
Running
    ↓
Success / Failed
```

---

### Metadata Database

The **Metadata Database** stores:

* DAG states.
* Task states.
* Execution history.
* Logs and metadata.

The Airflow UI reads this information to display pipeline status.

---

### Managed Airflow

Managed services automatically create and manage the Airflow environment.

Examples:

* Amazon MWAA.
* Google Cloud Composer.

On Amazon MWAA:

* Amazon S3 stores DAG files.
* Aurora PostgreSQL stores metadata.

| Parameter | Purpose |
|-----------|---------|

---

### Scheduling

The `schedule` parameter determines when the DAG runs.

Examples:

* Cron expressions.
* Presets such as `@daily`, `@hourly`, `@weekly`, `@monthly`.
* `timedelta` objects.

---

### Understanding `start_date`

`start_date` defines the **logical date** of the first DAG run.

Important concepts:

* Every DAG run has a **Data Interval**.
* The **Logical Date** is the **start** of that interval.
* The DAG executes **after** the data interval ends.

Example:

```text
Data Interval

March 1 ─────────────► March 2

Logical Date          DAG Runs
```

> Airflow runs the DAG at the **end** of the interval because all data for that period is then available.

---

### Catchup

`catchup` controls whether Airflow runs missed schedules.

* `True` → Execute all missed DAG runs.
* `False` → Skip missed runs and execute only future schedules. :contentReference[oaicite:2]{index=2}

# Airflow Best Practices

Writing DAGs using best practices makes them **efficient, readable, maintainable, idempotent, and reproducible.**

---

### Keep Tasks Simple & Atomic

Each task should perform **one operation**.

Example ETL workflow:

```text
Extract
   │
   ▼
Transform
   │
   ▼
Load
```

Benefits:

* Easier debugging.
* Better monitoring.
* Clear dependencies.
* Reusable tasks.

---

### Avoid Top-Level Code

Top-level code is any code outside the DAG definition or task operators.

Bad:

```python
call_api()

with DAG(...):
    ...
```

The scheduler parses DAG files regularly, so top-level code executes every time the DAG is parsed, even if the DAG doesn't run.

Instead, place code inside task functions.

---

### Use Variables Instead of Hard-Coded Values

Avoid:

```python
country = "US"
count = 20
```

Instead use:

* Airflow Variables
* Environment Variables

Retrieve values with:

```python
Variable.get("country")
```

Use Variables for configuration values that may change without editing the DAG.

Examples:

* API URLs
* Bucket names
* Countries
* Batch sizes

---

### Built-in Variables & Jinja Templates

Airflow provides built-in variables for every DAG run.

Example:

```text
{{ ds }}
```

`ds` is the logical date of the DAG run.

Example:

```python
PythonOperator(
    task_id="load",
    python_callable=load_to_s3,
    op_kwargs={
        "file_name": "data/{{ ds }}/file.csv"
    }
)
```

Jinja templates allow values to be generated dynamically at runtime.

---

### XCom Usage

**XCom (Cross-Communication)** is used to pass **small amounts of data** between tasks.

Suitable for:

* IDs
* Dates
* Metrics
* Status values
* File paths

Example:

```text
Task A
   │
xcom_push()
   │
   ▼
Metadata Database
   │
xcom_pull()
   │
   ▼
Task B
```

Avoid using XCom for:

* DataFrames
* Large JSON objects
* Large datasets

Instead, store large data in:

* Amazon S3
* Google Cloud Storage
* Azure Blob Storage
* Databases

---

### Task Groups

Use **Task Groups** to organize related tasks in the Airflow UI.

Example:

```text
Start
   │
   ▼
┌───────────────┐
│ Extract       │
│ Transform     │
└───────────────┘
   │
   ▼
Load
```

Benefits:

* Cleaner DAG visualization.
* Better organization.
* Easier navigation.

---

### Airflow is an Orchestrator

Airflow schedules and coordinates workflows.

Heavy processing should be performed by specialized tools such as:

* Apache Spark
* Apache Flink
* dbt
* SQL engines

Airflow should trigger these jobs rather than execute the processing itself.

---

### Keep Business Logic Separate

A DAG file should contain orchestration logic only.

Place business logic in separate Python modules.

Benefits:

* Cleaner DAG files.
* Better code reuse.
* Easier testing.

---

### Common DAG Parameters

| Parameter | Purpose |
|-----------|---------|
| `dag_id` | Unique DAG name |
| `schedule` | Execution schedule |
| `start_date` | First logical date |
| `catchup` | Run missed intervals |
| `tags` | Organize DAGs |
| `description` | DAG description |

---

### Common Operator Parameters

| Parameter | Purpose |
|-----------|---------|
| `task_id` | Task name |
| `python_callable` | Function to execute |
| `email` | Alert recipients |
| `email_on_failure` | Send email on failure |
| `email_on_retry` | Send email on retry |
| `retries` | Number of retry attempts |
| `op_kwargs` | Arguments passed to the task function |

---

### Summary

* Keep tasks **small and atomic**.
* Avoid **top-level code**.
* Store configuration using **Airflow Variables**.
* Use **Jinja templates** for dynamic values.
* Use **XCom** only for small data shared between tasks.
* Use **Task Groups** to organize complex DAGs.
* Airflow should **orchestrate**, not perform heavy computation.
* Keep business logic separate from DAG definitions.

# Airflow TaskFlow API

The **TaskFlow API** is an alternative way to write Airflow DAGs introduced in **Airflow 2.0**.

Its goal is to make DAGs **simpler, cleaner, and more concise**, especially when using many Python functions.

---

### Traditional Paradigm vs TaskFlow API

| Traditional Paradigm | TaskFlow API |
|-----------------------|--------------|
| Create a `DAG` object | Use the `@dag` decorator |
| Create tasks with `PythonOperator` | Create tasks with the `@task` decorator |
| Specify `python_callable` | Decorate the Python function directly |
| Use `xcom_push()` and `xcom_pull()` manually | Return values are automatically stored in XCom |

---

### The `@dag` Decorator

Instead of explicitly creating a `DAG` object, use the `@dag` decorator.

```python
@dag(
    schedule="@daily",
    start_date=...
)
def my_dag():
    ...
```

The function name becomes the **DAG ID** in the Airflow UI.

After defining the function, you must instantiate it:

```python
my_dag()
```

Otherwise, the DAG will not appear in the Airflow UI.

---

### The `@task` Decorator

Instead of creating a `PythonOperator`, decorate the Python function directly.

Traditional:

```python
PythonOperator(
    task_id="extract",
    python_callable=extract_data
)
```

TaskFlow:

```python
@task
def extract_data():
    ...
```

The function name automatically becomes the **task ID**.

Benefits:

* Less boilerplate code.
* Fewer names to manage.
* Better readability.

---

### Defining Dependencies

Dependencies are still created using the bit-shift operator (`>>`).

Example:

```python
extract() >> transform() >> load()
```

---

### XCom in the Traditional Paradigm

Data is shared manually using:

* `xcom_push()` to store data.
* `xcom_pull()` to retrieve data.

```text
Task A
   │
xcom_push()
   │
Metadata Database
   │
xcom_pull()
   │
Task B
```

---

### XCom in TaskFlow API

TaskFlow automatically uses **XCom** when a task returns a value.

```python
@task
def extract():
    return data
```

The returned value is automatically stored in XCom.

Another task simply accepts it as a function parameter.

```python
@task
def transform(data):
    ...
```

Dependencies are created by passing the returned value.

```python
data = extract()
transform(data)
```

or

```python
transform(extract())
```

No explicit `xcom_push()` or `xcom_pull()` is required.

---

### Manual XCom with TaskFlow

TaskFlow still supports manual XCom operations.

You can access the task instance from the Airflow context and call:

* `xcom_push()`
* `xcom_pull()`

when more control is needed.

---

### Mixing Both Paradigms

TaskFlow **does not replace** all Airflow operators.

You can combine:

* TaskFlow API
* Traditional Operators

within the same DAG whenever appropriate.

---

### Benefits of TaskFlow API

* Less boilerplate code.
* Cleaner Python syntax.
* Automatic XCom handling.
* Easier to read and maintain.
* Ideal for Python-based workflows.

---

### When to Use Each Approach

| Traditional | TaskFlow API |
|-------------|--------------|
| Using many Airflow operators | Mostly Python functions |
| Need explicit operator configuration | Prefer concise code |
| Manual XCom management | Automatic XCom handling |
| Complex workflows with many operator types | Python-centric pipelines |

---

### Summary

* TaskFlow API was introduced in **Airflow 2.0**.
* Use `@dag` instead of creating a `DAG` object manually.
* Use `@task` instead of `PythonOperator`.
* Function names automatically become **DAG IDs** and **Task IDs**.
* Returning a value from a task automatically stores it in **XCom**.
* Passing a returned value to another task automatically creates dependencies and shares data.
* Traditional operators and TaskFlow API can be used together in the same DAG.

# Branching in Apache Airflow

**Branching** allows a DAG to choose which task to execute next based on a condition.

It is used to create **dynamic workflows** where different execution paths are taken depending on runtime results.

---

### How Branching Works

A branch task evaluates a condition and returns the **task ID** of the next task to execute.

Example:

```text
          Extract Data
               │
               ▼
         Check Ratio
         /         \
        ▼           ▼
Print Greater   Print Less
        \           /
         ▼         ▼
         Do Nothing
```

If:

* `ratio > 0.5` → Execute `print_greater`
* `ratio <= 0.5` → Execute `print_less`

---

### Branching in the Traditional Paradigm

Use the **BranchPythonOperator**.

```python
BranchPythonOperator(
    task_id="check_ratio",
    python_callable=check_ratio
)
```

The branching function contains normal Python logic.

Example:

```python
def check_ratio(**context):
    if ratio > 0.5:
        return "print_greater"
    return "print_less"
```

The returned value **must be the `task_id`** of the task to execute next.

---

### Using XCom in Branching

The branch task often reads values produced by previous tasks using **XCom**.

Example:

```python
ratio = context["ti"].xcom_pull(
    key="ratio_us",
    task_ids="extract_data"
)
```

The extracted value is then used in the branching condition.

---

### Skipped Tasks

When a branch is chosen:

* The selected branch executes.
* All other branches are automatically **skipped**.

Example:

```text
Check Ratio
     │
     ├── Print Greater   Executed
     └── Print Less      Skipped
```

---

### Continuing After a Branch

Suppose both branches should continue to a common task.

```text
Print Greater
      │
      ▼
  Do Nothing
      ▲
      │
Print Less
```

By default, the final task is skipped because one upstream task was skipped.

To prevent this, set:

```python
trigger_rule="none_failed_min_one_success"
```

Meaning:

* No upstream task failed.
* At least one upstream task succeeded.

This allows the common task to run regardless of which branch was taken.

---

### Branching with TaskFlow API

TaskFlow replaces `BranchPythonOperator` with the **`@task.branch` decorator**.

```python
@task.branch
def check_ratio():
    ...
```

The function still returns the **task ID** of the next task.

Example:

```python
@task.branch
def check_ratio():
    if ratio > 0.5:
        return "print_greater"
    return "print_less"
```

The branching behavior remains the same.

---

### Empty Tasks in TaskFlow

Instead of using an `EmptyOperator`, you can create a task that performs no work.

```python
@task(trigger_rule="none_failed_min_one_success")
def do_nothing():
    pass
```

This task serves as a merge point after branching.

---

### Accessing XCom in TaskFlow

There are two common approaches.

Pass the entire Airflow context:

```python
def check_ratio(**context):
    context["ti"].xcom_pull(...)
```

Or pass only the task instance:

```python
def check_ratio(ti=None):
    ti.xcom_pull(...)
```

Passing `ti` directly is simpler when only the task instance is needed.

---

### Traditional vs TaskFlow Branching

| Traditional | TaskFlow API |
|-------------|--------------|
| `BranchPythonOperator` | `@task.branch` |
| Uses `python_callable` | Decorates the function directly |
| Returns next `task_id` | Returns next `task_id` |
| Uses `EmptyOperator` for merge | Can use a normal `@task` |
| Configure `trigger_rule` on the operator | Configure `trigger_rule` in the decorator |

---

### Summary

* Branching creates **dynamic execution paths**.
* Use **`BranchPythonOperator`** or **`@task.branch`**.
* The branch function returns the **`task_id`** of the next task.
* Branch conditions commonly use values retrieved from **XCom**.
* Non-selected branches are automatically skipped.
* Use `trigger_rule="none_failed_min_one_success"` when multiple branches merge into one task.
* In TaskFlow, the task instance can be accessed using `ti` instead of the full context dictionary.

AWS automatically provisions and manages the Airflow environment while keeping the standard Airflow interface.

MWAA architecture:

```text
        Amazon S3
      (DAG Files)
            │
            ▼
      Airflow Scheduler
       Airflow Workers
     (AWS Fargate)
            │
            ▼
   Amazon Aurora
 (Metadata Database)
            │
            ▼
 Amazon CloudWatch
 (Logs & Metrics)
```

Other integrations:

* AWS Key Management Service (KMS) for encryption.
* Amazon S3 for DAG storage.
* Amazon CloudWatch for monitoring.

Advantages:

* No infrastructure management.
* Automatic scaling.
* Managed upgrades.
* Standard Apache Airflow experience.

Disadvantages:

* Less infrastructure control.
* AWS-managed environment.

---

### AWS Glue Workflows

AWS Glue Workflows orchestrate **ETL pipelines** built with AWS Glue.

A workflow consists of:

* Glue Jobs
* Glue Crawlers
* Triggers

Workflow:

```text
Trigger
   │
   ▼
Crawler
   │
   ▼
Glue Job
   │
   ▼
Next Job
```

Triggers can start workflows:

* On a schedule.
* On demand.
* From Amazon EventBridge events.

Best suited for:

* ETL pipelines running entirely on AWS Glue.

---

### AWS Step Functions

AWS Step Functions orchestrate workflows using **State Machines**.

A state can represent:

* AWS Lambda
* AWS Glue Job
* Amazon ECS Task
* Decision logic
* Data transformation

Example:

```text
Start
   │
   ▼
 Lambda
   │
   ▼
Decision
 ┌───┴───┐
 ▼       ▼
Glue    ECS
  │       │
  └───┬───┘
      ▼
     End
```

Features:

* Native AWS integration.
* Serverless.
* Conditional branching.
* Error handling.
* Parallel execution.

Best suited for:

* AWS-centric workflows.
* Event-driven applications.
* Serverless architectures.

| Best for | Complex workflows | Managed Airflow | Glue ETL | AWS service orchestration |

### When to Choose Each

**Open-Source Airflow**

Choose when you need:

* Full control.
* Custom configurations.
* Maximum flexibility.

---

**Amazon MWAA**

Choose when you want:

* Airflow without managing infrastructure.
* Automatic scaling.
* Managed Airflow on AWS.

---

**AWS Glue Workflows**

Choose when your pipeline consists mainly of:

* Glue Jobs.
* Glue Crawlers.
* ETL workloads.

---

**AWS Step Functions**

Choose when your workflow coordinates:

* Lambda functions.
* ECS tasks.
* Glue jobs.
* Other AWS services.

---

### Key Trade-Off

```text
More Control <-----------------------------> More Convenience

Open-Source Airflow        MWAA      Glue / Step Functions
```

* More control generally means more operational responsibility.
* More convenience generally means less infrastructure management.

---

### Summary

* AWS offers multiple orchestration services, each optimized for different use cases.
* **Open-Source Airflow** provides maximum flexibility but requires infrastructure management.
* **MWAA** is a managed version of Apache Airflow that automates deployment, scaling, and maintenance.
* **AWS Glue Workflows** are designed specifically for orchestrating ETL jobs and crawlers.
* **AWS Step Functions** orchestrate AWS services using serverless state machines.
* The best orchestration tool depends on your requirements, balancing **control**, **convenience**, and the complexity of your workflows.
