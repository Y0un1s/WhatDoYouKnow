# 1- What is Linux:
* Unlike Windows or macOS, which are closed-source operating systems owned by Microsoft and Apple, **Linux is an open-source kernel**. Built through global collaboration, it serves as the core engine powering hundreds of distinct operating systems known as **Linux distributions** (like Ubuntu, Red Hat, or Debian). Most Linux distributions are open source. Some enterprise distributions, such as Red Hat Enterprise Linux (RHEL), provide commercial support subscriptions, certified software, and long-term maintenance for businesses.

* The kernel is the brain of Linux. It schedules processes and handles the hardware, such as the CPU, RAM, and disk.

* The shell, as its name suggests, is a protective layer around the kernel. It takes human-readable commands and translates them into something the kernel understands. The shell is usually accessed through a terminal, which is the interface where you type commands.





# 2- Why Linux in Big Data:
* **Software Conflicts & POSIX:** Many Big Data tools conflict with Windows (like HDFS or Apache Sqoop). Despite Sqoop being deprecated, other modern tools face the same problem.
    * Sqoop, Apache Hadoop, Apache Spark, and Apache Kafka were built by developers *on* Linux, *for* Linux.
    * Linux uses **POSIX** (Portable Operating System Interface), a standard rulebook that software follows to run seamlessly across different operating systems.
    * Think of POSIX as a shared kitchen rulebook. Windows is the chef who decided *not* to join the gang; they built a rival kitchen instead.
    * Microsoft finally admitted developers need POSIX. Instead of rewriting Windows, they created **WSL (Windows Subsystem for Linux)**, embedding a real, lightweight Linux kernel directly inside Windows.

* **Security:** Linux is inherently more secure due to its **strict privilege model** and its open-source nature. With thousands of eyes on the code, vulnerabilities are often patched quickly.

* **Containerization:** Modern Big Data is heavily driven by **Docker and Kubernetes**.
    * Containers are natively a Linux technology. While you *can* run Docker on Windows, it spins up a hidden Linux virtual machine behind the scenes. Running Linux inside Windows wastes a massive amount of compute power.

* **Horizontal Scaling & Economics:** Most Big Data tools are distributed systems that rely on horizontal scaling (clustering). Instead of buying one massive, expensive computer, you link 100 or 1,000 cheaper computers together.
    * Scaling out on Windows requires paying an OS licensing fee for *every single server node* in your cluster. On Linux, it's free.

* **The Bottom Line:** This is why roughly **90%+** of the world's cloud and Big Data infrastructure runs on Linux.

# 3- Linux commands

* **Before we start here are some tips to make your experience with the terminal better :** To make the Terminal quiet, you can disable Bell sound temporarily with: `bind 'set bell-style none'`, To make it permanent, add it to your `~/.inputrc` file: `echo 'set bell-style none' >> ~/.inputrc`, **Tab Completion:** Pressing Tab while typing a command or path automatically completes it when there is a unique match. For example, if `Pictures` is the only directory starting with `P`, typing `cd P` and pressing Tab becomes `cd Pictures/`. If multiple matches exist (e.g., `Documents` and `Downloads` for `cd Do`), nothing is completed; pressing Tab twice displays all available matches. Tab completion also works recursively for longer paths, so `cd Doc` + Tab becomes `cd Documents/`, and pressing Tab again may further complete to `cd Documents/python-examples/` if it is the only matching entry.

* **Built-in vs External Commands :** Not all commands in Linux work the same way. Some commands are built into the shell currently being used (called built-in commands), while others are separate executable programs stored on disk (called external commands). For example, `cd`, `echo`, and `read` are built-in commands, while `ls`, `cat`, `grep`, and `man` are external commands.
  When you run a command, the shell first checks whether it is a built-in command. If it is not, the shell searches the directories listed in the `PATH` environment variable for an executable file with the same name and runs it. (`PATH` is not a single directory; it is a list of directories.)
  This makes external commands feel just like native shell commands even though they are separate programs. You can use the `type` command to determine how the shell interprets a command. If `type` reports that a command is a shell builtin, then it is built into the shell; otherwise it may be an alias, a function, or an external command.
  Sometimes a command exists both as a shell builtin and as an external executable. In such cases, the shell usually prefers the built-in version. For example, Bash provides a built-in version of `echo` even though an external version may also exist on the system.

* **Scouting and Checking Files, Folders:**
* `ls` (list) : it displays the files and folders in your current path pretty much like the dir command in Windows. In modern distros, the interface highlights directories, which are folders, in blue and files in white. Running the plain vanilla `ls` command won't show you the hidden stuff.

* `ls -a` (list all) : it displays all files and directories including the hidden ones. In Linux hidden files and directories start with a dot (`.`), and hiding these is mainly for organizational purposes because you don't use these for 99% of the time, so they shouldn't always be present in front of you.

* `ls -l` (list long) and `ls -a -l` or `ls -al` (to include hidden ones) : This displays your files vertically with a ton of extra details that might seem gibberish at first, things like permissions, group, user, file size, timestamp of modification.

* `File Ownership and Permissions`: Every file and directory in Linux has an owner user and an owner group. You can view ownership and permissions using `ls -l`. The first character of the output indicates the file type: `d` means a directory, while `-` means a regular file. The next nine characters represent permissions and are divided into three groups of `rwx`: the first group applies to the owner (user), the second to the group, and the third to everyone else on the system (others). Here, `r` stands for read, `w` for write, and `x` for execute, while `-` means that the permission is not granted. For files, read allows viewing the contents, write allows modifying them, and execute allows running the file as a program. For directories, read allows listing its contents, write allows creating or deleting files within it, and execute allows entering or accessing the directory. Permissions can be changed with `chmod`, while ownership can be changed with `chown`.

* **Navigating and Tracking paths:**
* `pwd` (print working directory) : It shows you exactly where you are, the full path. Keep in mind that Linux uses forward slash `/` unlike Windows that uses backward slash `\`.

* `cd` (change directory) : it moves you around from your current directory to another one, note that Linux is case sensitive so if you type Documents instead of documents it will strictly look and try to move to a directory named Documents, typing `cd ..` takes you back one level but typing `cd .` does nothing because `.` stands for the current directory you are in.

* **Managing Files and Directories:**
* `find` : searches for files and directories starting from a specified directory and recursively traverses subdirectories looking for matches, `find <specified directory> -name <name>`, you can use `-iname` to perform case-insensitive search.

* `touch` : the name comes from touching a file to alter its modification time or access time without altering its content, but if the file doesn't exist it creates that file with a size of zero

* `mkdir` (make directory): creates a directory

* `cp` (copy) : Copies a file or directory from one location to another. To copy a directory and all of its contents, use the `-r` (recursive) option. The source and destination can be specified using either relative or absolute paths, a relative path starts from your current working directory, an absolute path starts from the root directory `/` and gives the complete location of a file or folder.

* `mv` (move) : moves a file or directory from one location to another. It can also be used to rename files and directories. Like the `cp` command, it accepts both relative and absolute paths for the source and destination. Unlike `cp`, it does not require the `-r` option when moving directories.

* `rm` (remove) : it deletes a file

* `rm -r` (remove recursively): In Linux, `rm` cannot remove a non-empty directory by default. The `-r` (or `--recursive`) option tells `rm` to delete the directory and all of its contents, including any subdirectories and files.

* `tar` (tape archive) : Combines files/directories into a single archive. Does not compress by default. Usually you combine it with gzip, `tar -czf myfolder_archive.tar.gz myfolder/`, c = create archive, z = compress with gzip, f = archive filename. The extension depends on whether compression is used and which compression algorithm is used, if not `.tar`, if compressed with gzip `.tar.gz`. You can extract a gzip-compressed archive using `tar -xzf myfolder_archive.tar.gz` or a regular tar archive using `tar -xf myfolder_archive.tar`.

* `curl` (client uniform resource locator) : Sends requests to a URL and returns the server's response. The response can be any data type (HTML, JSON, CSV, images, PDFs, etc.). Use `-o` like here `curl -o <file_name> <url>` to save the response to a file, making curl useful as a downloader too.

* `wget` (World Wide Web + get) : You can think of it as a specialized version of curl focused on downloading. It automatically saves downloaded content to a file and can even recursively download an entire website. For example, if you are downloading a website, wget tries to recreate the website's structure on your disk.

* **Viewing Content, Printing, and Help:**
* `cat` (Concatenate) : it combines text together and displays all file contents straight onto your screen.

* `more` : Displays a file page by page instead of printing the entire file like `cat`. Controls: Space = next page, b = previous page, q = quit.

* `head`, `tail` : Display the beginning or end of a file, 10 lines by default. You can change that using `-n <no>`. `tail -f` = follow a file as new content is added.

* `wc` : views the number of lines, words and characters in a file. For lines only use `-l`, for words only `-w`, for characters count `-c`. Note that it counts the new line characters too.

* `grep` : Searches for lines that match a specific word or a phrase in a file. Use `-i` to make it case insensitive.

* `echo` : Prints text to the terminal. It is a built-in shell command commonly used to display messages, variables, or command output. Use the `-e` option to enable special characters such as `\n` (new line). It is good practice to wrap the text in double quotes. Examples: `echo -e "Hello\nWorld"`. You can redirect the output to a file using `>` or `>>`: `echo "echo Hello World" >> test.sh`. This creates `test.sh` if it does not exist and appends the line to the end of the file. Using a single `>` instead of `>>` overwrites the entire contents of the file.

* `clear` or the shortcut pressing `ctrl l` : clears the terminal screen by moving the current output out of view, giving you a clean screen to work with. It does not delete your command history or stop any running processes.

* `history` : spits out the list of commands you executed since starting this session in the terminal (you don't have to press the up and down arrow keys to find your commands)

* `type` : it tells you if a command is built-in in the shell or external sitting somewhere on the disk. For example, `cd` (if you type `man cd` you won't get anything cause there is no manual for it outside the shell, it's built-in in the shell) unlike `man` which is an external command

* `command --help` : it prints out a small documentation of that specific command and its available options.

* `man command` (manual command) : it prints out a full manual of the command. Press `q` to quit.

* **Identity and Networking:**
* `uname -a`: kernel name and related info, like version and release date.

* `whoami` : it tells you who you logged in as

* `hostname` : prints the name of the Linux machine or VM you are currently operating on.

* `ip address` : Shows your network interfaces and current IP addresses.

* **Ownership & Privileges:**
* **Users and Groups** : A Linux user is an account with permissions. `root` has UID `0`, system users are used by services, and normal users are the people who log in. On many systems, system users have low UIDs and normal users start around `1000`, but the exact range can vary by distro. Basic account info is stored in `/etc/passwd`, and password hashes are stored in `/etc/shadow`. `sudo` lets an allowed user run commands with root privileges. `useradd` creates a user, `groupadd` creates a group, and `usermod -aG` adds a user to another group.

* `sudo` (superuser do): Executes a command with the absolute power and privileges of the Root user (the Administrator). If you try to step outside your home directory and touch critical system directories, Linux will slap you with a "Permission Denied" error. To bypass this, you prefix your command with `sudo`. It will ask for your account password, not the Root password (which is usually disabled or unknown by default in distros like Ubuntu). Not everyone can run `sudo` commands, the system administrator has to explicitly add your name to a VIP list of allowed users first.

* `sudo useradd (username)` : when you create a user Linux typically creates a primary group for that user with the same username. You can add a new user using this command

* `sudo groupadd (group-name)` : You can create a group using this command

* `sudo usermod -aG (group-name) (username)` (user modification) : you can add a user to a group by using this command, `-aG` means append supplementary groups. `sudo usermod -G developers ahmed` does not append. It replaces all supplementary groups. For example: **Before:** ahmed : ahmed docker sudo **After:** `sudo usermod -G developers ahmed`, ahmed : ahmed developers.
  **Note:** Every user has one **primary group** and can belong to multiple **supplementary groups**. The primary group is usually created automatically when the user is created and often has the same name as the username. The `-aG` option only adds or modifies supplementary groups and does not change the primary group. To change a user's primary group, use: `sudo usermod -g developers ahmed`, `developers` becomes Ahmed's new primary group. You can view a user's primary and supplementary groups using: `id ahmed`


* `su` (switch user): you can swap users using `su <user>`

* `chmod xxx` (change mode) : changes the permissions of a file or a directory. Imagine permissions as numbers where `(r)` read = 4 `(w)` write = 2 `(x)` execute = 1. The permission value for each group is the sum of the permissions you want to grant. So if you wanna give all permissions for the 3 groups for a file you can use the command `chmod 777 (file or directory name)`. If you want to give the group permissions of read, write and deny any permissions to the others group you use `chmod 760 (file or directory name)`. Owner = rwx = 7 Group = rw- = 6 Others = --- = 0 and so on. Note that only the super user or the owner of a directory or a file can change its permissions.

* `sudo chown (new owner)` (change owner) : changes the owner of a file or a directory. It can also change the group ownership, for example: `chown newowner:newgroup file`.

* **Process Handling:**

* **Processes, Daemons and Jobs** : A process is a running program managed by the Linux kernel and identified by a PID (Process ID). Most processes either run in the foreground (attached to a terminal) or in the background. A daemon is a special type of background process that runs independently of any user's terminal and usually provides a system service. Examples include the SSH daemon (`sshd`), Docker daemon (`dockerd`), and cron daemon (`crond`). Daemons are typically started during system boot and continue running until the system shuts down, A job is a shell-managed collection of one or more processes.


* `|` (pipe a.k.a pipeline): Creates a pipeline in which the output of one command becomes the input of another command. For example: `command1 | command2`.

* `ps` (process status) : Displays information about running processes. By default, it shows a limited set of processes associated with the current terminal as a snapshot. Use `-e` to display all running processes on the system, including those belonging to other terminals and users.

* `top` (table of processes) : Shows a live, constantly updating view of system activity. Unlike `ps`, which takes a snapshot, `top` refreshes every few seconds. You can press `k` to kill a process while in there and you can quit by pressing `q`.

* `nice` : nice controls the scheduling priority of a process when it starts. Linux uses a nice value ranging from: `-20` (highest priority), `0` (default), `19` (lowest priority). Example: `nice -n 10 python script.py`. This starts the process with a lower priority so it is "nice" to other processes, a higher value means even less CPU priority.

* `renice` : changes the nice value of a process that is already running, Example: `renice -20 -p (PID)`

* `;`, `&` : `;` runs commands sequentially, while `&` runs a command in the background and immediately returns control to the shell. For example: `command1 ; command2` runs the second command after the first finishes, while `command1 &` starts the command in the background.

* Job Control (`jobs`, `fg`, `bg`) : You can view background jobs using `jobs`, which displays job numbers. When referring to a job in commands such as `fg`, `bg`, and `kill`, prefix the job number with `%`, for example: `%1`, `%2`, etc. To bring a background job back to the foreground, use `fg %1`. To continue a stopped job in the background, use `bg %1`.

  A simple background command such as `python script.py &` is typically one job containing one process. However, a pipeline such as `grep ERROR log.txt | sort | uniq &` is still considered a single job even though it contains three processes (`grep`, `sort`, and `uniq`) because they are connected together as one pipeline.

  Running multiple background commands creates multiple jobs. For example:

  ```bash 
  python script1.py & python script2.py & python script3.py &
  ```

  creates three separate jobs because each command was started independently. The shell may display something similar to:

  ```text
  [1] Running  python script1.py &
  [2] Running  python script2.py &
  [3] Running  python script3.py &
  ```

  Jobs provide a convenient way for the shell to manage foreground and background execution without requiring you to remember PIDs.
  * Note: Pressing Ctrl + Z suspends (stops) the currently running foreground job and returns control to the shell. The job is not terminated and can be resumed with either fg or bg.

* `kill` : Terminates a process by sending it a signal. The most common usage is `kill PID`, where PID is the Process ID of the process you want to stop. By default, `kill` sends the `SIGTERM` (Signal Terminate) signal, which politely asks the process to terminate and gives it a chance to clean up resources before exiting. `kill` sends a signal to a process; it does not necessarily kill it immediately. If the process refuses to stop, you can use `kill -9 PID`, which sends the `SIGKILL` signal and forces the kernel to terminate the process immediately. When working with jobs, you can also terminate a job using its job number, for example: `kill %1`. To view the PID of a process, use commands such as `ps` or `top`, and to view job numbers use `jobs`.

* `systemctl` (system control) : Command for controlling services, often used with `sudo` for system services. `sudo systemctl start docker`, `sudo systemctl stop docker`, `sudo systemctl restart docker`

* **Related To Commands:**
* `#`, `?`, `*`: `#` starts a comment. `?` matches exactly one character. `*` matches zero or more characters and is commonly used as a wildcard.

* `" "`, `' '`, `\` : Both double quotes (`"..."`) and single quotes (`'...'`) prevent characters such as `;`, `|`, `>`, `<`, and `&` from being treated as shell syntax. The difference is that double quotes still allow variable expansion (`$name`), command substitution (`$(date)`), and arithmetic expansion (`$((2+3))`), while single quotes treat everything literally, A backslash (`\`) escapes a single character, causing Bash to treat it literally. For example, `"\$HOME is my home"` prints `$HOME is my home` instead of expanding the variable.

* `stdin` (Standard Input): The default input stream a program reads from when it needs input. By default, it is connected to the keyboard. For example, running `cat` without a filename makes it wait for input from `stdin`. `stdin` can be redirected to come from another source, such as a file `cat < file.txt` or another command `echo "Hello" | cat`.

* `stdout` (Standard Output): The default place where a program sends its normal output. By default, `stdout` is connected to your terminal screen.

* `>`, `>>`, `<` : `>` redirects the standard output of a command to a file, overwriting the file if it already exists. For example: `echo "Hello" > file.txt`. `>>` redirects the standard output of a command to a file, appending to the end instead of overwriting it. For example: `echo "Hello" >> file.txt`. `<` redirects a file to a command's `stdin`. For example: `cat < file.txt`. You can also redirect error output using file descriptor `2`, such as `2>` and `2>>`.

* `$(command)` : Replaces a command with its output. For example: `working_directory=$(pwd)` stores the output of `pwd` in the variable `working_directory`.

* Variables: Their scope is limited to the current shell session unless exported. You can create a shell variable using `variable=value` (do not leave spaces around `=`). Variables can be created directly in the shell or inside a script. You can display all shell variables using `set`, or combine it with a pipe such as `set | head -4`. To access the value of a variable, prefix its name with `$`, for example: `$variable`. You can remove a variable using `unset variable`.

  * Shell Variable: Exists only within the current shell session and is not inherited by child processes.
  * Environment Variable: A shell variable that has been exported, making it available to child processes and programs started from the shell.
  * You can make a shell variable an environment variable using `export`, for example: `export variable`.

* Arguments (args): You can pass arguments to a script when executing it, for example: `./script.sh arg1 arg2`. Inside the script, arguments can be accessed using special variables such as `$1`, `$2`, `$3`, and so on, where `$1` represents the first argument, `$2` the second, and so forth.



# 4- Linux File System Hierarchy

* **Before we start**: there are 2 concepts that you need to understand.
  * First, every file in Linux is associated with an inode, which stores the file's metadata such as permissions, owner, timestamps, and the locations of the file's data blocks on disk. You can view a file's metadata using the `stat` command. A file name itself is stored separately and points to the inode.

  * The second concept is the difference between a hard link and a soft link (symbolic link). A hard link `ln <file_name> <hard_link_name>` creates another file name that points to the same inode, so both names refer to the same file. If you delete one name, the file remains accessible through the other name. A soft link a.k.a symbolic link `ln -s <file_name> <soft_link_name>` creates a reference to the file's path instead of directly to its inode. If the original file is deleted or moved, the soft link will no longer work. This is similar to a shortcut in Windows.

  * **Note:** when you delete a file using `rm`, Linux removes the file name (directory entry), not the inode itself. If other hard links still exist, the file remains accessible through them. When the last hard link is removed and no process is still using the file, Linux frees the inode and its data blocks. This is somewhat similar to reference counting in programming languages such as Python, where an object can only be removed after all references to it are gone.

* `/` **(Root directory)** : This is the top-level directory of the entire Linux filesystem. Everything starts from here. To navigate to it, use `cd /`. **Note:** The root directory is different from the root user. The root directory exists to organize the entire operating system and serves as the starting point of the filesystem hierarchy. The root user, on the other hand, is the administrator account in Linux and has full control over the system. On Ubuntu and WSL, you can obtain a root shell using `sudo -i` or switch to root using `sudo su`. To exit the root shell, use `exit`.

* **Absolute vs Relative Paths** : An absolute path starts from the root directory `/` and specifies the complete location of a file or directory, such as `/home/ahmed/projects/script.py`. A relative path starts from your current working directory, such as `projects/script.py`. Every absolute path in Linux begins with `/`.

* `/root` : The home directory of the root user (administrator). This is different from the root directory `/`. To navigate to it, use `cd /root`. Access to this directory typically requires root privileges.

* `/home` : Contains the home directory of each user. A user's home directory is the personal folder where that user's files, documents, downloads, projects, and settings are stored. To go there, use `cd ~` for your own home directory or `cd /home` to view the parent directory that contains all users' home folders. It may contain many users, such as `/home/ahmed` and `/home/mahmoud`.

* `/etc` *(Et Cetera, Latin for "and so on")* : Contains system-wide configuration files and settings. Many applications store their configuration files here. Examples include network settings, user account information, and SSH configuration. To navigate to it, use `cd /etc`.

* `/usr` *(Unix System Resources)* : Contains most user-installed programs, libraries, documentation, and shared resources. Many executable programs are stored in `/usr/bin`. To navigate to it, use `cd /usr`.

  **Note:** Despite the name, `/usr` does not contain user home directories. Historically the name came from Unix, but today it mainly stores programs, libraries, documentation, and shared resources.

* `/bin` : Contains essential command-line programs required for basic system operation, such as `ls`, `cp`, `mv`, and `cat`. To navigate to it, use `cd /bin`. On modern Linux distributions, `/bin` is often a symbolic link to `/usr/bin`, meaning both paths point to the same location. This change was made to simplify the filesystem layout, while keeping `/bin` for backward compatibility so older programs and scripts that expect binaries in `/bin` continue to work without modification.

* `/sbin` *(System Binaries)* : Contains important system administration programs such as filesystem management, networking, and system recovery tools. On modern Linux distributions, `/sbin` is often linked to `/usr/sbin`, similar to how `/bin` is often linked to `/usr/bin`.

* `/var` *(Variable data that changes frequently)* : Contains data that changes frequently while the system is running. This includes log files, caches, temporary application data, and databases. To navigate to it, use `cd /var`. A commonly used subdirectory is `/var/log`, which stores system and application logs.

  **Note:** Logs can be thought of as a diary or journal that Linux and applications write to as they run.

* `/tmp` : Stores temporary files created by the system and applications. Files in this directory may be automatically deleted when the system reboots. To navigate to it, use `cd /tmp`.

* `/dev` : Contains device files. In Windows, hardware is usually accessed through drivers and special APIs. In Linux, there is a philosophy that says, "Everything is a file." As a result, many hardware devices are represented as files inside this directory. To navigate to it, use `cd /dev`.

* `/boot` : Contains files required for the Linux boot process, including the Linux kernel and bootloader-related files. Modifying files in this directory incorrectly can make the system unbootable. To navigate to it, use `cd /boot`.

* **Mounting** : Linux uses a single directory tree that starts at `/`. Disks and partitions do not appear as separate drive letters like `C:` or `D:` in Windows. Instead, they are attached (mounted) somewhere inside the directory tree. For example, one partition may be mounted as `/`, another as `/home`, and another as `/mnt/data`.

  This means that directories that appear closely related in the filesystem hierarchy may actually reside on completely different physical disks. In an extreme example, `/home` could be stored on one disk, while `/home/younis` could be a separate filesystem mounted from an entirely different disk, SSD, network share, or storage device. The filesystem hierarchy describes how files are organized logically, while mounting determines where the data is physically stored.

* `/proc` *(Process Information Filesystem)* : A virtual filesystem created by the Linux kernel. It does not contain real files stored on disk. Instead, it exposes information about running processes and the system itself. Examples include `/proc/cpuinfo`, which contains CPU information, and `/proc/meminfo`, which contains memory information. To navigate to it, use `cd /proc`.

* `/sys` *(System Filesystem)* : Another virtual filesystem that exposes information about devices, drivers, and kernel subsystems. It provides a structured way for the kernel to expose hardware and system information to userspace programs. To navigate to it, use `cd /sys`.

* `/mnt` *(Mount)* : Traditionally used as a temporary location for manually mounting filesystems, disks, or network shares. For example, an administrator may temporarily mount a disk under `/mnt/data`. To navigate to it, use `cd /mnt`.

* `/media` : Commonly used for automatically mounted removable devices such as USB drives, CDs, and DVDs. Modern desktop Linux distributions often mount removable media somewhere under this directory. To navigate to it, use `cd /media`.

* `/lib` and `/lib64` *(Libraries)* : Contain essential shared libraries required by programs in directories such as `/bin` and `/sbin`. These libraries are similar in concept to DLL files in Windows. To navigate to them, use `cd /lib` or `cd /lib64`.

* `/opt` *(Optional Software)* : Typically used for installing optional third-party software that does not follow the standard Linux package layout. Some commercial applications and manually installed software place their files here. To navigate to it, use `cd /opt`.

* **Filesystem Hierarchy Standard (FHS)** : Linux distributions generally follow a specification called the Filesystem Hierarchy Standard (FHS). It defines the purpose of major directories such as `/etc`, `/usr`, `/var`, and `/home`, helping software behave consistently across different Linux distributions.

# 5- Linux Packages

* Packages are archive files that contain the software, metadata, configuration files, and dependency information required to install or update an application. Packages have different file types; for example, Debian-based distributions use `.deb` files, while Red Hat-based distributions use `.rpm` files.

* `apt` (Advanced Package Tool) : It is the default package manager on Debian-based distributions. You can use it to install, remove, and update packages. For example: `sudo apt install gedit`. Note that it requires `sudo` because installing software changes **system-wide files**, which means it affects the operating system and can affect all users of the system. Linux protects those files from normal users.

  * `apt list --installed` : Lists all packages currently installed on the system.

  * `apt show <package>` : Displays detailed information about a package such as its version, dependencies, description, and size. Example: `apt show docker.io`.

  * `-y` (yes) : Automatically answers "yes" to confirmation prompts. Example: `sudo apt install openssh-server -y`.

* `sudo apt update` : Updates the local package database by downloading the latest package information from the configured repositories. It also lets you know how many installed packages can be upgraded.

* `sudo apt upgrade` : Upgrades all installed packages on your system to the newest available versions.

* **Repositories, Updates, and Manual Installation** : A repository (repo) is a server on the internet that stores software packages (`.deb` files) and metadata about those packages, such as their versions, dependencies, and descriptions. Ubuntu comes preconfigured with official repositories, and software vendors may provide additional repositories for their own software.

  When you run:

  ```bash
  sudo apt update
  ```

  APT contacts all configured repositories and downloads the latest package metadata, updating its local package database. It does **not** install or upgrade any software at this stage.

  When you run:

  ```bash
  sudo apt upgrade
  ```

  APT compares the versions installed on your system against the versions available in the configured repositories. If a newer version exists, APT downloads and installs it.

  Software vendors are responsible for maintaining their repositories and publishing new package versions. Your system does not automatically search the internet for updates; it only checks the repositories that have been configured on your machine.

  If you manually install a package using a downloaded `.deb` file:

  ```bash
  sudo apt install ./package.deb
  ```

  the package is installed and recorded in the local package database. However, this does **not** automatically connect the package to the developer's website or update server. Automatic updates are only possible if a configured repository contains newer versions of that package.
  Some software vendors provide a repository and either ask you to add it manually or configure it automatically during installation. In that case, the repository is added to the system's list of configured repositories, and future updates follow the normal APT workflow. Without a configured repository, you would typically need to manually download and install newer versions of the package whenever updates are released.

  In summary:

  ```text
  Manual Installation with Vendor Repository
  
  Download package.deb
           ↓
  sudo apt install ./package.deb
           ↓
  Package Installed
           ↓
  Vendor repository added to
  APT's configured repositories
           ↓
  #(the above was One-time setup)
  sudo apt update
           ↓
  APT downloads package metadata
  from the vendor repository
           ↓
  Vendor publishes a newer version
           ↓
  sudo apt update
           ↓
  APT discovers the newer version
           ↓
  sudo apt upgrade
           ↓
  Package Upgraded
  ```


  A useful way to think about a repository is as a package catalog server that APT periodically contacts to learn which software versions are available.


* `dpkg` (Debian Package Manager) : The low-level package management tool used by Debian and Ubuntu systems. The `apt` package manager uses `dpkg` underneath to perform the actual installation, removal, and querying of `.deb` packages. Unlike `apt`, `dpkg` does not automatically download packages from repositories or resolve dependencies. Example: `sudo dpkg -i package.deb`. In most cases, users interact with `apt` rather than `dpkg` directly because `apt` automatically handles repositories and dependencies.

  * `dpkg -l` : Lists installed packages on the system.

  * `dpkg -S <file>` : Finds which installed package owns a specific file. Example: `dpkg -S /usr/bin/git`.

* `alien` : You can convert packages from one type to another using the `alien` command. To convert an RPM package to Debian format, use: `alien (package_name).rpm`. To convert a Debian package to RPM format, use: `alien -r (package_name).deb`.

* **Note :** There are other package managers, such as `pip`, which is used to install Python packages. Ubuntu discourages `sudo pip install` because it can overwrite Python packages managed by `apt`, causing dependency conflicts and potentially breaking system tools. Modern Ubuntu systems may require a flag such as `pip install pandas --break-system-packages` to override this protection. Use `sudo` only if you are installing a package into system Python directories (those owned by root), not just for your own account. In general, use `apt` for system packages and `pip` inside a virtual environment for your own projects.


# 6- SSH



* **Protocol** : A protocol is a standardized set of rules that defines how devices, applications, or systems communicate with each other over a network. It specifies how data is formatted, transmitted, received, interpreted, and how errors are handled.

  A protocol by itself is just a specification (a set of rules). To actually use a protocol, you need software that implements those rules.

  A protocol typically defines:
  - Message format (how data is structured)
  - Communication sequence (who sends what and when)
  - Error detection and handling
  - Authentication methods
  - Encryption and security mechanisms

  Client-Server Communication Model:

  ```text
  Client -----------------> Server
           Request

  Client <----------------- Server
            Response
  ```

  **Client:**
  - Requests a service or resource.
  - Initiates communication.
  - Receives and processes responses.

  **Server:**
  - Provides a service or resource.
  - Waits for incoming requests.
  - Processes requests and sends responses.

  Example (HTTP):

  ```text
  Browser (Client)
         |
         | "Give me index.html"
         |
         v
  Web Server

  Web Server
         |
         | "Here is index.html"
         |
         v
  Browser (Client)
  ```

  In this example:
  - HTTP is the protocol (the communication rules).
  - The browser is the client application implementing HTTP.
  - The web server is the server application implementing HTTP.

  Real-world analogy:
  - A protocol is like a language combined with a set of conversation rules.
  - The client is the person asking a question.
  - The server is the person providing the answer.
  - Both must follow the same language and rules to understand each other.

* **SSH (Secure Shell)** : A secure protocol that lets a client connect to a remote machine, authenticate, and execute commands over a network.

  By default, SSH uses port **22**.
  - IP Address = Building Address
  - Port       = Apartment Number
  - An IP address identifies which machine you want to reach.
  - A port identifies which application or service on that machine you want to reach.

  SSH follows the same client-server model:

  ```text
  ssh Client -----------------> sshd Server
               Request

  ssh Client <----------------- sshd Server
                Response
  ```
  * **sshd (SSH Daemon)** : The SSH server process that runs on a machine and listens for incoming SSH connections.

  Example:

  ```text
  Local Machine
       |
       | SSH Connection
       |
       v
  Linux Server
       |
       v
  Remote Shell
  ```

  After connecting, commands entered on the local machine are executed on the remote machine.

  ```text
  Local Terminal
       |
       | ls -l
       |
       v
  Executed on Remote Server
  ```

  Like any protocol, SSH is not an application. To use SSH, you need software that implements the SSH protocol.

  Examples of SSH implementations:
  - OpenSSH
  - PuTTY

  The minimum components required are:

  ```text
  Local Machine                 Remote Machine
  -------------                 --------------
  SSH Client  ----------------> SSH Server
  ```

  The SSH client initiates the connection, while the SSH server waits for incoming connections.

  Traditional SSH authentication requires:
  - A username
  - A password
  - The server hostname or IP address

  Example:

  ```bash
  ssh username@192.168.1.10
  ```

  Where:
  - `username` is an account on the remote machine.
  - `192.168.1.10` is the IP address of the remote machine.

  SSH can also use key-based authentication, which is more secure than passwords and is commonly used in production environments.

  To use key-based authentication, you first generate a key pair on your local machine You can also give the key a descriptive name:

  ```bash
  ssh-keygen -t ed25519 -f ~/.ssh/linux_server_key
  ```

  This creates:

  ```text
  Private Key: ~/.ssh/linux_server_key
  Public Key : ~/.ssh/linux_server_key.pub
  ```
  Where:
  - `-t` stands for type
  - `ed25519` is the cryptographic algorithm used to create the key pair.
  - `-f` where you after specify the file location and name
  - The private key must stay secret on your machine.
  - The public key can be shared and copied to remote servers.

  - When creating a cloud VM, the cloud provider often generates the SSH key pair for you. You typically download only the private key (for example, my-server.pem), while the matching public key is automatically placed on the server

  Before key-based authentication works, the public key must be copied to the remote server. A common way to do this is:
  - **How can the server allow me to copy something there in the first place?**
  Because the first time, you usually authenticate with a password.
  
  ```bash
  ssh-copy-id -i ~/.ssh/linux_server_key.pub username@192.168.1.10
  ```

  This command:
  - `-i` identify file
  - Connects to the remote machine using password authentication.
  - Creates the `~/.ssh` directory if it does not exist.
  - Appends your public key to:

  ```text
  ~/.ssh/authorized_keys (a file that contains all authorized keys)
  ```

  on the remote machine.
  

  Once the server has your public key, it can verify that you own the matching private key.
  - After that, password authentication can even be disabled if desired.

  When you connect:

  ```bash
  ssh username@192.168.1.10
  ```

  the authentication process works like this:

  1. The server checks whether your public key exists in `authorized_keys`.
  2. The server sends a challenge that can only be answered using the matching private key.
  3. Your SSH client uses the private key to answer the challenge.
  4. If the answer is correct, the server allows the connection.

  ```text
  Local Machine                           Remote Machine
  -------------                           --------------
  Private Key (secret)                    Public Key
          |                                      |
          +---------- Authentication ----------->|
                                                 |
                                          Access Granted
  ```

  Unlike password authentication, the private key is never sent across the network.

  Common SSH key algorithms:
  - ED25519 (recommended for most users)
  - RSA
  - ECDSA

  SSH can also be used for:
  - Secure file transfer (SCP)
  - Secure file synchronization (rsync over SSH)
  - Port forwarding and tunneling (Securely forward network traffic through an SSH connection. This is commonly used to access services that are not directly exposed to the network.)
  - Forwarding graphical applications

# 7- What is Shell Scripting 
* Shell scripting is the process of writing a sequence of commands in a text file that can be interpreted and executed by a shell. Commands can be entered interactively in a terminal or stored line by line in a script file. Shell scripts are interpreted at runtime rather than compiled beforehand, similar to Python. While this generally makes them slower than compiled languages, one of their main advantages is that they are specifically designed for executing commands, launching programs, and automating command-line tasks. Shell scripting is an essential skill for anyone working on a POSIX-based system (POSIX is a standard that defines how Unix-like operating systems should behave, allowing scripts to work across systems such as Linux, macOS, and Unix). Scripts are commonly used to automate repetitive tasks and system administration processes (tasks involved in managing and maintaining computer systems, such as user management, backups, software installation, monitoring, and log analysis).

* A shell script can be made executable and may begin with an interpreter directive, commonly known as a shebang. The shebang syntax is: `#!<path-to-interpreter> [optional-arg]`, where the interpreter is specified using an absolute path, for example: `#!/bin/bash`. The shebang is not limited to shell scripts; it can be used with other interpreted languages as well. For example, a Python script may start with: `#!/usr/bin/env python3`.

* When an executable script is run, the operating system uses the shebang to determine which program should execute the file. Although shebangs are most commonly used with interpreted languages, they can also be used with compiled executables. For example: `#!/path/to/my_program`, where `my_program` is a compiled C program created from source code such as `my_program.c` using `gcc my_program.c -o my_program`. When the script is executed, the operating system runs `my_program` and passes the script file to it. Compiled languages such as C and C++ generally do not require a shebang because their executables already contain machine code and can be run directly by the operating system. However, any executable program capable of processing a file can be specified in a shebang.

* One of the main strengths of shell scripting is its ability to combine built-in shell commands (such as echo, cd, and read) and external programs (such as curl, grep, wc, and python3) into a single executable file. The shell acts as a coordinator, running each command and program in the required order, allowing complex tasks to be automated using simple text-based commands. External programs often feel like native shell commands because the shell automatically searches the directories listed in the PATH environment variable for an executable with the requested name and runs it. Shell scripts can also execute programs that are not in PATH if their path is explicitly specified. In this example, Bash uses curl to download a file, python3 to process it, and wc to count its lines and then runs a local program that is not in PATH. This demonstrates how shell scripts can connect multiple tools into a single automated workflow.

```bash
#!/bin/bash

echo "Downloading file..."
curl https://example.com/data.csv -o data.csv

echo "Processing data..."
python3 process.py

echo "Counting lines..."
wc -l data.csv

echo "Running a local program..."
./my_program

echo "Done!"
```

Bash is commonly used for this because it can run shell commands and external programs directly. While the same workflow could be written in a Python script, Python would require additional code, such as the `subprocess` module, to launch external programs. This makes Bash a simple and popular choice for automating command-line tasks and connecting multiple tools into a single workflow. More generally, shell scripting is designed for executing commands, launching programs, and automating system tasks, whereas languages such as Python are general-purpose programming languages designed for building applications, implementing algorithms, processing data, and solving a wider range of programming problems.
