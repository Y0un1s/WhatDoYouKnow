# 1- What is Linux:
* Unlike Windows or macOS, which are closed-source operating systems owned by Microsoft and Apple, **Linux is an open-source kernel**. Built through global collaboration, it serves as the core engine powering hundreds of distinct operating systems known as **Linux distributions** (like Ubuntu, Red Hat, or Debian). Most Linux distributions are open source. Some enterprise distributions, such as Red Hat Enterprise Linux (RHEL), provide commercial support subscriptions, certified software, and long-term maintenance for businesses.

* The kernel is the brain of Linux. It schedules processes and handles the hardware, such as the CPU, RAM, and disk.

* The shell, as its name suggests, is a protective layer around the kernel. It takes human-readable commands and translates them into something the kernel understands. The shell is usually accessed through a terminal, which is the interface where you type commands.





# 2- Why Linux in Big Data:
* **Software Conflicts & POSIX:** Many Big Data tools conflict with Windows (like HDFS or Apache Sqoop). Despite Sqoop being deprecated, other modern tools face the same problem.
    * Sqoop, Apache Hadoop, Apache Spark, and Apache Kafka were built by developers *on* Linux, *for* Linux.
    * Linux uses **POSIX** (Portable Operating System Interface), a standard rulebook that software follows to run seamlessly across different operating systems.
    * Think of POSIX as a shared kitchen rulebook. Windows is the chef who decided *not* to join the gang; they built a rival kitchen instead.
    * Microsoft finally admitted developers need POSIX. Instead of rewriting Windows, they created **WSL (Windows Subsystem for Linux)**, embedding a real, lightweight Linux kernel directly inside Windows.

* **Security:** Linux is inherently more secure due to its **strict privilege model** and its open-source nature. With thousands of eyes on the code, vulnerabilities are often patched quickly.

* **Containerization:** Modern Big Data is heavily driven by **Docker and Kubernetes**.
    * Containers are natively a Linux technology. While you *can* run Docker on Windows, it spins up a hidden Linux virtual machine behind the scenes. Running Linux inside Windows wastes a massive amount of compute power.

* **Horizontal Scaling & Economics:** Most Big Data tools are distributed systems that rely on horizontal scaling (clustering). Instead of buying one massive, expensive computer, you link 100 or 1,000 cheaper computers together.
    * Scaling out on Windows requires paying an OS licensing fee for *every single server node* in your cluster. On Linux, it's free.

* **The Bottom Line:** This is why roughly **90%+** of the world's cloud and Big Data infrastructure runs on Linux.

# 3- Linux commands

* **Before we start here are some tips to make your experience with the terminal better :** To make the Terminal quiet, you can disable Bell sound temporarily with: `bind 'set bell-style none'`, To make it permanent, add it to your `~/.inputrc` file: `echo 'set bell-style none' >> ~/.inputrc`, **Tab Completion:** Pressing Tab while typing a command or path automatically completes it when there is a unique match. For example, if `Pictures` is the only directory starting with `P`, typing `cd P` and pressing Tab becomes `cd Pictures/`. If multiple matches exist (e.g., `Documents` and `Downloads` for `cd Do`), nothing is completed; pressing Tab twice displays all available matches. Tab completion also works recursively for longer paths, so `cd Doc` + Tab becomes `cd Documents/`, and pressing Tab again may further complete to `cd Documents/python-examples/` if it is the only matching entry.

* **Built-in vs External Commands :** Not all commands in Linux work the same way. Some commands are built into the shell currently being used (called built-in commands), while others are separate executable programs stored on disk (called external commands). For example, `cd`, `echo`, and `read` are built-in commands, while `ls`, `cat`, `grep`, and `man` are external commands. When you run a command, the shell first checks if it is a built-in command. If it is not, the shell searches the directories listed in the `PATH` environment variable for an executable file with the same name and runs it (PATH is not one directory. It's a list of directories). This makes external commands feel just like native shell commands even though they are separate programs. You can use the `type` command to determine whether a command is built-in or external.

* **Scouting and Checking Files, Folders**
* `ls` (list) : it displays the files and folders in your current path pretty much like the dir command in Windows. In modern distros, the interface highlights directories, which are folders, in blue and files in white. Running the plain vanilla `ls` command won't show you the hidden stuff.

* `ls -a` (list all) : it displays all files and directories including the hidden ones. In Linux hidden files and directories start with a dot(.), and hiding these is not a trickery (like oh no one would be able to see them), but it's mainly for organizational purposes because you don't use these for 99% of the time, so they shouldn't be always present in front of you so you won't be annoyed with them.

* `ls -l` (list long) and `ls -a -l` or `ls -all` (to include hidden ones) : This displays your files vertically with a ton of extra details that might seem gibberish at first, things like permissions, group, user, file size, timestamp of modification.
Let's explore the permissions first, if the first character is `d` then it's a directory or `-` then it's a file. And then there are `rwx` in order = read write execute. If any of them is `-` then it means that permission is denied. First `rwx` is for the user and second `rwx` is for the group and third `rwx` is for anyone else on the system, so it's basically 10 characters, first one determines the type is a directory or a file and then there are 3 groups of permissions for the user, the group, and anyone else.

* **Navigating and Tracking paths**
* `pwd` (print working directory) : It shows you exactly where you are, the full path. Keep in mind that Linux uses forward slash `/` unlike Windows that uses backward slash `\`.

* `cd` (change directory) : it moves you around from your current directory to another one, note that Linux is case sensitive so if you type Documents instead of documents it will strictly look and try to move to a directory named Documents, typing `cd ..` takes you back one level but typing `cd .` does absolutely nothing cause it stands for the current directory you are in.


* **Managing Files and Directories**
* `find` : searches for files and directories starting from a specified directory and recursively traverses subdirectories looking for matches, `find <specified directory> -name <name>`, you can use `-iname` to perform case-insensitive search.

* `touch` : the name comes from touching a file to alter its modification time or access time without altering its content, but if the file doesn't exist it creates that file with a size of zero

* `mkdir` (make directory): creates a directory

* `cp` (copy) : Copies a file or directory from one location to another. To copy a directory and all of its contents, use the `-r` (recursive) option. The source and destination can be specified using either relative or absolute paths, a relative path starts from your current working directory, an absolute path starts from the root directory `/` and gives the complete location of a file or folder.

* `mv` (move) : moves a file or directory from one location to another. It can also be used to rename files and directories. Like the `cp` command, it accepts both relative and absolute paths for the source and destination. Unlike `cp`, it does not require the `-r` option when moving directories.

* `rm` (remove) : it deletes a file

* `rm -r` (remove recursively): In Linux, `rm` cannot remove a non-empty directory by default. The `-r` (or `--recursive`) option tells `rm` to delete the directory and all of its contents, including any subdirectories and files.

* `tar` (tape archive) : Combines files/directories into a single archive. Does not compress by default. Usually you combine it with gzip, `tar -czf myfolder_archive.tar.gz myfolder/`, c = create archive, z = compress with gzip, f = archive filename. The extension depends on whether compression is used and which compression algorithm is used, if not `.tar`, if compressed with gzip `.tar.gz`. You can decompress or dearchive using `tar -xf myfolder_archive.tar.gz myfolder_archive` for example.

* `curl` (client uniform resource locator) : Sends requests to a URL and returns the server's response. The response can be any data type (HTML, JSON, CSV, images, PDFs, etc.). Use `-o` like here `curl -o <file_name> <url>` to save the response to a file, making curl useful as a downloader too.

* `wget` (World Wide Web + get) : You can think of it as a specialized version of curl focused on downloading. It automatically saves downloaded content to a file and can even recursively download an entire website. For example, if you are downloading a website, wget tries to recreate the website's structure on your disk.

* **Viewing Content, Printing, and Help**
* `cat` (Concatenate) : it combines text together and displays all file contents straight onto your screen.

* `more` : Displays a file page by page instead of printing the entire file like `cat`. Controls: Space = next page, b = previous page, q = quit.

* `head`, `tail` : Display the beginning or end of a file, 10 by default. You can change that using `-n <no>`. `tail -f` = follow a file as new content is added.

* `wc` : views the number of lines, words and characters in a file. For lines only use `-l`, for words only `-w`, for characters count `-c`. Note that it counts the new line characters too. 

* `grep` : Searches for lines that match a specific word or a phrase in a file. Use `-i` to make it case insensitive.

* `echo` : Prints text to the terminal. It is a built-in shell command commonly used to display messages, variables, or command output. Use the `-e` option to enable special characters such as `\n` (new line). It is good practice to wrap the text in double quotes. Examples: `echo -e "Hello\nWorld"`. You can redirect the output to a file using `>` or `>>`: `echo "echo Hello World" >> test.sh`. This creates `test.sh` if it does not exist and appends the line to the end of the file. Using a single `>` instead of `>>` overwrites the entire contents of the file.

* `clear` or the shortcut pressing `ctrl l` : clears the terminal screen by moving the current output out of view, giving you a clean screen to work with. It does not delete your command history or stop any running processes.

* `history` : spits out the list of commands you executed since starting this session in the terminal (you don't have to press the up and down arrow keys to find your commands)

* `type` : it tells you if a command is built-in in the shell or external sitting somewhere on the disk. For example, `cd` (if you type `man cd` you won't get anything cause there is no manual for it outside the shell, it's built-in in the shell) unlike `man` which is an external command

* `command --help` : it prints out a small documentation of that specific command and what its associated tags etc..

* `man command` (manual command) : it prints out a full manual of the command, if you wanna quit you can just press `q`



* **Identity and Networking**
* `whoami` : it tells you who you logged in as

* `hostname` : prints the name of the Linux machine or VM you are currently operating on.

* `ip address` : Shows your network interfaces and current IP addresses.

* **Ownership & Privileges**
* `sudo` (superuser do): Executes a command with the absolute power and privileges of the Root user (the Administrator). If you try to step outside your home directory and touch critical system directories, Linux will slap you with a "Permission Denied" error. To bypass this, you prefix your command with `sudo`. It will ask for your account password, not the Root password (which is usually disabled or unknown by default in distros like Ubuntu). Not anyone can run `sudo` commands, the system administrator has to explicitly add your name to a VIP list of allowed users first.

* `sudo useradd (username)` : when you create a user Linux typically creates a primary group for that user with the same username. You can add a new user using this command

* `sudo groupadd (group-name)` : You can create a group using this command

* `sudo usermod -aG (group-name) (username)` (user modification) : you can add a user to a group by using this command, `-aG` means append supplementary groups. `sudo usermod -G developers ahmed` does not append. It replaces all supplementary groups. For example: **Before:** ahmed : ahmed docker sudo **After:** `sudo usermod -G developers ahmed`, ahmed : ahmed developers.



* `chmod xxx` (change mode) : changes the permissions of a file or a directory. Imagine permissions as numbers where `(r)` read=4 `(w)` write=2 `(x)` execute=1. The permission value for each group is the sum of the permissions you want to grant. So if you wanna give all permissions for the 3 groups for a file you can use the command `chmod 777 (file or directory name)`. If you want to give the group permissions of read, write and deny any permissions to the others group you use `chmod 760 (file or directory name)`. Owner = rwx = 7 Group = rw- = 6 Others = --- = 0 and so on. Note that only the super user or an owner of a directory or a file can change its permissions.

* `sudo chown (new owner)` (change owner) : changes the owner of a file or a directory

* `ps` (process status) : Displays information about running processes. By default, it shows a limited set of processes associated with the current terminal as a snapshot. Use `e` to display all running processes on the system, including those belonging to other terminals and users.

* `top` (table of processes) : Shows a live, constantly updating view of system activity. Unlike `ps`, which takes a snapshot, `top` refreshes every few seconds. You can press `k` to kill a process while in there and you can quit by pressing `q`.

* `kill (PID)` (kill a process using its id) : terminate gracefully. If the process refuses to stop: `kill -9 (PID)`. The `-9` signal (SIGKILL) forces the kernel to immediately terminate the process.

* `nice` : nice controls the scheduling priority of a process when it starts. Linux uses a nice value ranging from: `-20` (highest priority), `0` (default), `19` (lowest priority). Example: `nice -n 10 python script.py`. This starts the process with a lower priority so it is "nice" to other processes, a higher value means even less CPU priority.

* `renice` : changes the nice value of a process that is already running, Example: `renice -20 -p (PID)`

* `&` : places a process directly in the background. Example: `python script.py &`. This process now runs in the background and returns terminal control back to you, and returns the job number followed by the PID. A process is a running program managed by the Linux kernel and identified by a PID (Process ID). A job is a shell-managed collection of one or more processes that lets you move commands between the foreground and background using commands like jobs, bg, and fg

* `sudo systemctl` (system control) : Command for controlling services, `sudo systemctl start docker`, `sudo systemctl stop docker`, `sudo systemctl restart docker`

# 4- Linux File System Hierarchy

* `/` **(Root directory)** : This is the top-level directory of the entire Linux filesystem. Everything starts from here. To navigate to it, use `cd /`. **Note:** The root directory is different from the root user. The root directory exists to organize the entire operating system and serves as the starting point of the filesystem hierarchy. The root user, on the other hand, is the administrator account in Linux and has full control over the system. On Ubuntu and WSL, you can obtain a root shell using `sudo -i`. To exit the root shell, use `exit`.

* `/root` : The home directory of the root user (administrator). This is different from the root directory `/`. To navigate to it, use `cd /root`. Access to this directory typically requires root privileges.

* `/home` : Contains the home directory of each user. A user's home directory is the personal folder where that user's files, documents, downloads, projects, and settings are stored. To go there, use `cd ~` or `cd /home`. It may contain many users, such as `/home/ahmed` and `/home/mahmoud`.

* `/etc` *(Et Cetera, Latin for "and so on")* : Contains system-wide configuration files and settings. Many applications store their configuration files here. Examples include network settings, user account information, and SSH configuration. To navigate to it, use `cd /etc`.

* `/usr` *(Unix System Resources)* : Contains most user-installed programs, libraries, documentation, and shared resources. Many executable programs are stored in `/usr/bin`. To navigate to it, use `cd /usr`.

* `/bin` : Contains essential command-line programs required for basic system operation, such as `ls`, `cp`, `mv`, and `cat`. To navigate to it, use `cd /bin`. On modern Linux distributions, `/bin` is often linked to `/usr/bin`, meaning they may point to the same location.

* `/var` *(Variable data that changes frequently)* : Contains data that changes frequently while the system is running. This includes log files, caches, temporary application data, and databases. To navigate to it, use `cd /var`. A commonly used subdirectory is `/var/log`, which stores system and application logs. **Note:** Logs can be thought of as a diary or journal that Linux and applications write to as they run.

* `/tmp` : Stores temporary files created by the system and applications. Files in this directory may be automatically deleted when the system reboots. To navigate to it, use `cd /tmp`.

* `/dev` : Contains device files. In Windows, hardware is usually accessed through drivers and special APIs. In Linux, there is a philosophy that says, "Everything is a file." As a result, many hardware devices are represented as files inside this directory. To navigate to it, use `cd /dev`.


# 5- Linux Packages

* Packages are archive files that contain the software, metadata, configuration files, and dependency information required to install or update an application. Packages have different file types; for example, Debian-based distributions use `.deb` files, while Red Hat-based distributions use `.rpm` files.

* `apt` (Advanced Package Tool) : It is the default package manager on Debian-based distributions. You can use it to install, remove, and update packages. For example: `sudo apt install gedit`. Note that it requires `sudo` because installing software changes **system-wide files**, which means it affects the operating system and can affect all users of the system. Linux protects those files from normal users.

* `sudo apt update` : Updates the local package database by downloading the latest package information from the configured repositories. It also lets you know how many installed packages can be upgraded.

* `sudo apt upgrade` : Upgrades all installed packages on your system to the newest available versions.

* `alien` : You can convert packages from one type to another using the `alien` command. To convert an RPM package to Debian format, use: `alien (package_name).rpm`. To convert a Debian package to RPM format, use: `alien -r (package_name).deb`.

* **Note :** There are other package managers, such as `pip`, which is used to install Python packages. Ubuntu discourages `sudo pip install` because it can overwrite Python packages managed by `apt`, causing dependency conflicts and potentially breaking system tools. Modern Ubuntu systems may require a flag such as `pip install pandas --break-system-packages` to override this protection. Use `sudo` only if you are installing a package into system Python directories (those owned by root), not just for your own account. In general, use `apt` for system packages and `pip` inside a virtual environment for your own projects.


# 6- What is Shell Scripting

* Shell scripting is the process of writing a sequence of commands in a text file that can be interpreted and executed by a shell. Commands can be entered interactively in a terminal or stored line by line in a script file. Shell scripts are interpreted at runtime rather than compiled beforehand, similar to Python. While this generally makes them slower than compiled languages, one of their main advantages is that they are specifically designed for executing commands, launching programs, and automating command-line tasks. Shell scripting is an essential skill for anyone working on a POSIX-based system (POSIX is a standard that defines how Unix-like operating systems should behave, allowing scripts to work across systems such as Linux, macOS, and Unix). Scripts are commonly used to automate repetitive tasks and system administration processes (tasks involved in managing and maintaining computer systems, such as user management, backups, software installation, monitoring, and log analysis).

* A shell script can be made executable and may begin with an interpreter directive, commonly known as a shebang. The shebang syntax is: `#!<path-to-interpreter> [optional-arg]`, where the interpreter is specified using an absolute path, for example: `#!/bin/bash`. The shebang is not limited to shell scripts; it can be used with other interpreted languages as well. For example, a Python script may start with: `#!/usr/bin/env python3`.

* When an executable script is run, the operating system uses the shebang to determine which program should execute the file. Although shebangs are most commonly used with interpreted languages, they can also be used with compiled executables. For example: `#!/path/to/my_program`, where `my_program` is a compiled C program created from source code such as `my_program.c` using `gcc my_program.c -o my_program`. When the script is executed, the operating system runs `my_program` and passes the script file to it. Compiled languages such as C and C++ generally do not require a shebang because their executables already contain machine code and can be run directly by the operating system. However, any executable program capable of processing a file can be specified in a shebang.

* One of the main strengths of shell scripting is its ability to combine built-in shell commands (such as echo, cd, and read) and external programs (such as curl, grep, wc, and python3) into a single executable file. The shell acts as a coordinator, running each command and program in the required order, allowing complex tasks to be automated using simple text-based commands. External programs often feel like native shell commands because the shell automatically searches the directories listed in the PATH environment variable for an executable with the requested name and runs it. Shell scripts can also execute programs that are not in PATH if their path is explicitly specified. In this example, Bash uses curl to download a file, python3 to process it, and wc to count its lines and then runs a local program that is not in PATH. This demonstrates how shell scripts can connect multiple tools into a single automated workflow.

```bash
#!/bin/bash

echo "Downloading file..."
curl https://example.com/data.csv -o data.csv

echo "Processing data..."
python3 process.py

echo "Counting lines..."
wc -l data.csv

echo "Running a local program..."
./my_program

echo "Done!"
```

Bash is commonly used for this because it can run shell commands and external programs directly. While the same workflow could be written in a Python script, Python would require additional code, such as the `subprocess` module, to launch external programs. This makes Bash a simple and popular choice for automating command-line tasks and connecting multiple tools into a single workflow. More generally, shell scripting is designed for executing commands, launching programs, and automating system tasks, whereas languages such as Python are general-purpose programming languages designed for building applications, implementing algorithms, processing data, and solving a wider range of programming problems.
